`Sets are mainly about uniqueness and membership.`

`A set is a collection of unique values.`

````python
# A set automatically removes duplicate values.

citis = {"Mumbai", "Delhi", "Mumbai", "Pune"}

print(citis)
````

`It removed duplicate value Mumbai`

`Set = collection where every value is unique.`

### **1. CREATING A SET**

````python
# Curly braces create a set when you provide values.

numbers = {10, 20, 30, 40}

print(numbers)
````

`You can also create one from another collection.`

````python
# list -> Set
numbers = [10, 10, 20, 30, 40, 40]

unique_nums = set(numbers)

print(f"Unique Numbers: {unique_nums}")
````

`This is very useful pattern.`  

### **2. REMOVING DUPLICATES**

`Suppose a dataset contains:`


````python
customer_id = [101, 102, 101, 103, 102, 104]
````

`We want to know which customers are unique.`

````python
unique_customers = set(customer_id)

print(f"Unique Customers: {(unique_customers)}")
````

`This is one of the practical reason you will use sets.`

` 
Remember : Sets are unordered.
`

### **3. MEMBERSHIP TESTING**

`Sets are excellent for asking: Does this value exist?`

````python
cities = {"Mumbai", "Delhi", "Pune"}

print("Mumbai" in cities)
# True

print("Chennai" in cities)
# False
````

` This is called membership test.`

### **4. ADDING VALUES**

`Set are mutable.`

````python
cities = {"Mumbai", "Delhi"}

cities.add("Pune")

print(cities)
````

`If we add a duplicate.`

````python
cities.add("Mumbai")

print(cities)
````

`Nothing new is added.`

### **5. REMOVING VALUES**

````python
cities = {"Mumbai", "Delhi", "Pune"}

cities.remove("Delhi")

print(cities)   
````

`There is also discard():`

````python
cities.discard("Chennai")
print(citis)
````

`The Difference: remove() -> raises an error if value does not exist.`

`discard() -> does nothing if the value does not exist.`

### **6. SET OPERATIONS - HIGH VALUE**

````python
customers_a = {101, 102, 103, 104}
customers_b = {103, 104, 105, 106}
````

### **Intersection**

`Who appears in both groups?`

````python
common = customers_a.intersection(customers_b)

print(common)
````

`Think: A & B -> Common values`

### **Union**

`Who appears in either group?.`

````python
all_customers = customers_a.union(customers_b)

print(all_customers)
# {101, 102, 103, 104, 105, 106}
````

`Think: A | B -> everthing from both, without duplicates.`

### **Difference**

`Who is in A and not in B?`

````python
only_a = customers_a.difference(customers_b)

print(only_a)
# {101, 102}
````

`Think: A - B -> values unique to A.`
