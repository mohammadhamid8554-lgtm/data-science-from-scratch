### **3. Your practice exercise: Write a JSON file**

- You already know how to read JSON. Now let's learn how to save Python data into a JSON file using json.dump().

````python
customer =  {
    "customer_name": "Priya",
    "age": 21,
    "premium_member": False
}
````

Imports the json module.



Opens data/customer.json in write mode.



Uses with open().



Saves the dictionary to the file using json.dump().

````python
import json

with open("data/customer.json", "w") as file:

    json.dump(customer, file, indent=4)
````

`The indent=4 option controls indentation. It changes the formatting, not the meaning of the data.`

### **2. Next concept: Handle errors when reading JSON**

`What happens if a JSON file is missing? Or if it exists but contains invalid JSON?`

- FileNotFoundError: Python cannot find the file at the specified path.



- json.JSONDecodeError: Python finds and opens the file, but its contents are not valid JSON.

`For example, this is valid JSON:`

````python
{"product_name": "Laptop", "price": 50000}
````

`But this is invalid JSON because the property name isn't enclosed in double quotes:`

````python
{product_name: "Laptop", "price": 50000}
````

`When you call json.load(file) on invalid JSON, Python raises JSONDecodeError.`

### **Your practice exercise**

````python
import json

try:
    with open("data/product.json", "r") as file:
        product = json.load(file)
        print(product)

except FileNotFoundError:
    print("File not found.")


except json.JSONDecodeError:
    print("The file contains invalid JSON.")
````

product.json exists and contains valid JSON.



product.json does not exist.



product.json exists but contains {product_name: "Laptop"}.

1. It will print content present in the file.

2. File not found.

3. The file contains invalid JSON.

`Remember: A file can exist and still contain data that Python cannot interpret as valid JSON. File existence and data validity are two separate checks.`

### **2. Final JSON practice: Read, update, and save**

````python
{
    "product_name": "Laptop",
    "price": 50000,
    "in_stock": True
}
````

Imports json.



Opens the file in "r" mode and loads the data with json.load().



Changes the price to 52000.



Opens the same file in "w" mode.



Saves the updated dictionary with json.dump() and indent=4.

````python
import json

try: 
    with open("data/product.json", "r") as file:
        product = json.load(file)

        product.update({"price": 52000})

    with open("data/product.json", "w") as file:
        json.dump(product, file, indent=4)

except FileNotFoundError:
    print("File not found.")

except json.JSONDecodeError:
    print("File does not contain a valid JSON format.")
````
