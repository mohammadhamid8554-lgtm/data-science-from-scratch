### **Concept 3: Reading a file safely with `with open()`**

````python
# So far we have used this pattern

file = open("sales.txt", "r")

content = file.read()

print(content)

file.close()
````

```
This works, but there's a potential problem.

What happens if an error occurs before file.close() executes?

The file might remain open longer than necessary.

Python provides a cleaner approach: with open().

```

````python
with open("sales.txt", "r") as file:
    content = file.read()
    print(content)
````

`Notice that we no longer need to call file.close() ourselves.`

### **Understant the structure**

1. with -> Manages the file's use safely.

2. open("sales.txt", "r") -> Opens file in read mode.

3. as file  -> Gives us a variable to work with the opened file.

4. Indented statements -> Operations performed while the file is open.

`Data scientist's rule: Prefer with open() for ordinary file operations because it manages file closure automatically.`

### **2. Next concept: Check before reading a file**

- Imagine you're building a sales analysis program. Your program needs to read sales.csv, but the file might be missing.

- You already know that opening a missing file in "r" mode raises FileNotFoundError.

- There are two common approaches.

### **Approach A: Handle the error**

````python
try:
    with open("data/sales.csv", "r") as file:
        content = file.read()

        print(content)

except FileNotFoundError:
    print("Sales file not found.") 
````

### **Appraoch B: Check before opening**

````python
from pathlib import Path

file_path = Path("data/sales.csv")

if file_path.is_file():
    with open(file_path , "r") as file:
        content = file.read()
        print(content)

else:
    print("Sales file not found.")
````

### **3. Your practice exercise**

Imports Path.

Creates the path data/customers.txt.

Checks whether that path points to a file.

If it is a file, opens it in "r" mode and prints its contents.

Handles FileNotFoundError with an appropriate message.

````python
from pathlib import Path

try:
    file_path = Path("data/customers.txt")

    if file_path.is_file():
        with open(file_path, "r") as file:
            content = file.read()
            print(content)
    else:
        print("The path does not point to a file.")

except FileNotFoundError:
    print("Could not find customers.txt.")
````
