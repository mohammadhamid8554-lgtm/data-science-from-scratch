## **3. if + else**

`Sometimes you have two possible paths.`

````python
score = 45

if score >= 50:
    print("Pass")

else:
    print("Fail")
````

`Only one branch executes.`

## **4. if + elif + else**

`When there are multiple possible categories:`

````python
score = 78

if score >=90:
    print("Excellent")

elif score >= 75:
    print("Good")

else:
    print("Fail")
````

```

Python checks from top to bottom.

Once it finds a condition that is True, it executes that block and stops 

checking the remaining branches.

```

### **5. THE ORDER MATTER**

`This is a classic mistake.`

````python
# Bad logic:

score = 95

if score >= 50:
    print("Pass")

elif score >= 90:
    print("Excellent")
````

```

Output: Pass

Why?

Because 95 >= 50 is already True.

Python never reaches the elif.

```

````python
# Better: 

if score >= 90:
    print("Excellent")

elif score >= 50: 
    print("Pass")
````

```

Data Scientist lesson

Conditions should usually be ordered from the most specific/high-priority 

rule to the broader rule.

```

### **6. REAL DATA SCIENCE EXAMPLE**

````python
# Suppose you're detecting suspicious transactions.

transaction_amount = 15_000

if transaction_amount > 10_000:
    print("Review transaction")

else:
    print("Normal transaction")

# This is a simple rule-based classification.
````

```
Later, machine learning will learn such patterns from data instead of you 

manually writing every rule.

So conditional logic is one of the basic building blocks behind data 

processing and decision systems.

```

### **7. MULTIPLE CONDITIONS**

````python
# You can combine conditions using: 1. and, 2. or, 3. not.

age = 25
income = 60_000

if age >= 18 and income >= 50_000:
    print("Meets criteria")
````

` Both conditions must be True.`

````python
if age < 18 or age > 60:
    print("Special age group")
````

`Only one condition needs to be true.`

````python
is_verified = False

if not is_verified:
    print("Verification required")
````

`not flips the Boolean.`

### **8. ELITE CODE - DATA SCIENTIST STYLE**

````python
# Example: classifly a customer's spending level

spending = 750

# Python checks conditions from top to bottom.

if spending >= 1000:
    category = "High"

elif spending >= 500:
    category = "Medium"

else:
    category = "Low"

# We store the decision instead of immediatly printing it. 
# this is mor useful in data-processing code.
print(category)
````
