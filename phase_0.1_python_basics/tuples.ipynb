`A tuple is an ordered collection of values that cannot be changed after creation`

```python
#  A tuple sotres multiple values in an ordered structure.
# Parentheses () are commonly used to create a tuple

coordinates = (19.07, 72.87)

print(coordinates)
```

`Think:`

`List: I can change this collection.`

`Tuple: This collection should stay fixed.`

### **1. ACCESSING TUPLE ELEMENTS**

`Tuples use the same indexing system as lists.`

```python
customer = ("Tom", 20, "Mumbai")

print(customer[0])
# Tom

print(customer[1])
# 20

print(customer[2])
# Mumbai
```

`Indexing still starts at 0.`

### **2. SLICING**

`Tuples also support slicing.`

```python
data = ("Tom", 20, "Mumbai", True)

print(data[:3])
```

`Same rule:`

`start -> included`

`stop -> excluded`

### **3. THE BIG DIFFERENCE: IMMUTABILIBTY**

`This is the concept you actually need to remember.`

```python
data = ["Tom", 20]
data[1] = 23

print(data)
```

`It works because lists are mutable`

```python
data = ("Tom", 20)

data[1] = 32
print(data)
```

` List: Mutable `

`Tuple: Immutable`

### **4. WHY SHOULD WE WANT SOMETHING IMMUTABLE?**

```python
coordinates = (91.23, 72.8777)

# These two numbers represent a fiexed coordinate.
# You don't want some later piece of code accidentally doing.

coordinates[1] = 25


# A tuple prevents that
```

### **5. TUPLE UNPACKING**

`This is one tuple feature worth knowing well`

```python
customer = ("Tom", 20, "Mumbai")

name, age, city = customer

print(name)
print(age)
print(city)
```

`This becomes surprisingly useful when working with functions, database result, and structured data`

### **6. TUPLE WITH DIFFERENT DATA TYPES**

`A tuple can contain different types`

```python
customer = ("Tom", 21, "Mumbai", 5500.0, True)
print(type(customer))
```

`The important point is: Tuple is about the structure and mutability, not about what type of data it contains.`

### **7. A DATA SCIENCE EXAMPLE**

`Imagine a model configuration that shouldn't accidentally change:`

```python
model_settings = ("Random Forest", 100, 10)

model_name, n_trees, max_depth = model_settings

print([model_name, n_trees, max_depth])
```

`You can think of this as a small fixed record.`
