### **1. FOR LOOP - THE IMPORTANT ONE**

````python
# Suppose:

sales = [100, 200, 300, 400]

# You want to look at every sale.

sales = [100, 200, 300, 400]

for index,sale in enumerate(sales, start = 1 ):
    print(f"{index}. {sale}")
````

### **2. DON'T CONFUSE THE LOOP VARIABLE WITH THE LIST**

````python
for sale in sales:
    print(sale)

# Sales -> entire collection
# sale -> one item at a time
# this distinction is important when working with datasets.
````

### **3. LOOP THROUGH DICTIONARY**

````python
customer = {
    "name": "Alex",
    "age": 25,
    "city": "Mumbai"
}

for key, value in customer.items():
    print(f"{key} : {value}")
````

` You get each key-value pair.`

```
Conceptually:

"name" → "Alex"

"age"  → 25

"city" → "Mumbai"

```

`This is why you learned .itmes() earlier.`

### **4. range()**

````python
# You will frequently see: 

for i in range(5):
    print(i)

# Last element is excluded.
````

### **5. while Loop**

`A while loop keeps running while a condition is True.`

````python
count = 1

while count <= 3:
    print(count)
    count += 1
````

```
Is count <= 3?
       ↓
      Yes
       ↓
    execute
       ↓
   increase count
       ↓
check again

```

### **6. The dangerous part of while**

`You must eventually make the condition become False.`

````python
count = 1

while count <= 3:
    # print(count)

#count never changes.
````

```
So the condition remains True forever.

That's called an infinite loop.
```

### **7. break**

`break means: stop the loop immediatly.`

````python
for sale in sales:
    if sale > 250:
        break
    
    print(sale)
````

`The loop stops when it reaches a sale greater than 250.`

### **8. continue** 

```
continue means:

Skip this iteration and move to the next one.

```

````python
for sale in sales:
    if sale == 200:
        continue
    
    print(sale)
````

`200 was skipped.`

### **10. Elite Code**

````python
# A small example fo precessing customer spending

spending  = [200, 1500, 700, 3000]

for amount in spending:

    # Each loop handles one observation.
    if amount >= 1000:
        category = "High"

    elif amount >= 500:
        category  = "Medium"

    else: 
        category = "Low"

    # Show the decision for the current observation.
    print(f"{amount}: {category}")
````

```
Notice the connection:

LIST
 ↓
LOOP through observations
 ↓
CONDITION
 ↓
DECISION

That is already starting to resemble real data processing.
```
