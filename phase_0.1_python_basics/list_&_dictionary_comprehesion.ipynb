### **1. The Big Idea**

````python
# Imagine you have a lst of sales.

sales = [100, 200, 350, 400, 50]

# Your task is to create a new list containing only sales greater than 200.

# You already know how to use a loop:

high_sale = []

for sale in sales:
    if sale > 200:
        high_sale.append(sale)

print(f"High sales: {high_sale}")
print("This works perfectly.")
````

````python
# But Python gives us a shorter way to express the same logic.

sales = [100, 250, 80, 400, 150]

high_sale = [sale for sale in sales if sale > 200]

print(f"High Sales: {high_sale}")
print("This is called list comprehension.")
````

```

Both versions produce the same result.

The difference is how we express the solution.

```

### **2. Understand the Pattern**

````python
# Look carefully at this expression:

high_sale = [sale for sale in sales if sale > 200]
print(f"High sales: {high_sale}")

# Read it in plain English:

print("\nCreate a list containing each sale, but only when that sale is greater than 200.")
````

`new_list = [expression for item in collection if condition]`

`A list comprehension can transform every item without filtering by omitting the optional if condition.`



### **3. Three Patterns You Must Master**

`Pattern A: Transform every value`

````python
# Suppose a company wants to icreaase every product price by 10%.

prices = [200, 130, 450, 800, 300]

new_prices = [price * (1 - 0.10) for price in prices]

print(f"New Prices: {new_prices}")
print("The important idea is that every input value is transformed into a new value.")
````

`Pattern B: Filter values`

````python
# Suppose we want to identify customers who spent at least 500.


customer_spending = [200, 750, 450, 1200, 500]

# Keep only spending values of at least 500.

high_spending = [amount for amount in customer_spending if amount >= 500]

print(f"High Spending Customers: {high_spending}")
print("\nNotice that the values stay unchanged. The condition decides which values survive.")
````

`Pattern C: Transform and filter together`

````python
# Suppose a business wants to give a 10% discount only on products costing moe than 500.

prices = [200, 560, 700, 900, 1000]

discounted_prices = [price * (1 - 0.10) for price in prices if price > 500]

print(f"Discounted Prices: {discounted_prices}")
````

`Comprehensions efficiently filter and transform data, but complex business rules still need careful consideration.`

### **4. Dictionary Comprehensions**

`A dictionary comprehension follows the same idea but creates a dictionary instead of a list.`

````python
# Suppose we have a list of product prices:

prices = [100, 200, 300]

# We want to create a dictionary in which each price is the key and its tax amount in the value.

tax_by_price = {price: round(price * 0.18, 2) for price in prices}
print(f"Tax by Prices: {tax_by_price}")
````

```

The general pattern is:

new_dict = {
    
    key_expression: value_expression

    for item in collection
}

```

````python
# You can also filter items:

prices = [100, 200, 300, 400]

tax_by_price = {price: round(price * 0.18, 2) for price in prices if price >= 300}
print(f"Tax by prices: {tax_by_price}")
print("\nOnly prices of 300 or more are included.")
````

```
A critical detail

Dictionary keys must be unique.

```

````python
# For example:

numbers = [10, 20, 10, 30]

result = {number: number * 2 for number in numbers}

print(f"Uniue dictioanry values: {result}")
````

`Duplicate dictionary keys are overwritten by the last value, so each key appears only once`

`This matters when working with customer IDs, transaction records, and other real datasets.`

### **5. The Most Important Data Science Lesson**


````python
# Consider this dataset:

transactions = [
    {"customer_id" : 101, "amount" : 500},
    {"customer_id": 102, "amount" :  300},
    {"customer_id" : 101, "amount" : 700}
]

# Suppose we want a list of all transaction amounts.

amounts = [amount["amount"] for amount in transactions]
print(f"List of amounts: {amounts}")
````

`Understand what each row represents before transforming data, and use Pandas or NumPy for large-scale operations.`

### **6. Common Mistakes**

` Mistake 1: Confusing tranformtion with filtering.`

````python
numbers = [1, 2, 3, 4]

result = [number * 2 for number in numbers]
print(result)
print("\nEvery value is tranformed.")
````

````python
# But: 

result = [number for number in numbers if number % 2 == 0]
print(result)
print("\nValues are filtered, not transformed.")
````

#### `Mistake 2: Making everything a comprehension`

`Use comprehensions for simple transformations, but choose clear loops or functions when the logic becomes complex`

#### `Mistake 3: Forgetting that the result is a new collection`

````python
numbers = [1, 2, 3]

doubled = [number ** 2 for number in numbers]
print(numbers)
print(doubled)
````

### Challenge 5 — Your First Dictionary Comprehension

A company stores product prices:

````python
products = {
    "Laptop": 60000,
    "Mouse": 800,
    "Keyboard": 2500,
    "Monitor": 12000
}
````

The company wants a new dictionary containing only products costing more than ₹5,000, with a 10% discount applied to their prices.

### **Your task**

Create a dictionary named discounted_products using a single dictionary comprehension.

Requirements:

Keep the original product names as dictionary keys.

Include only products whose original prices are greater than 5000.

Apply a 10% discount to each included price.

Print the resulting dictionary.

````python
discounted_products = {product: price * 0.90 for product, price in products.items() if price > 5000}
print(f"Discounted Dictionary: {discounted_products}")
````
