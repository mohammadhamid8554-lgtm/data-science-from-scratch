### 1. **What Is a Built-in Function**?

```


Python comes with many functions already available. These are called built-in functions.

Think of them as tools in a toolbox.

Need to count items? Use len().

Need to calculate a total? Use sum().

Need to find the largest value? Use max().

Need to sort values? Use sorted()

```

````python
# For example:

sales = [200, 500, 300, 800]

total_sales = sum(sales)

print(f"Total Sales: {total_sales}")
````

`You didn't need to write a loop to calculate the total`

`Data Scientist mindset: Before writing your own logic, ask whether Python already provides a reliable tool for the task.`

### **2. The Six Essential Function**

#### **A. len() -- Count the Items**

````python
# Suppose you have customer transcation amounts.

transactions = [200, 500, 300, 800]

number_of_transactions = len(transactions)
print(f"Total number of transactions: {number_of_transactions}")
````

`len() returns the number of items in a collection.`

````python
# It also works with strings and dictionaries.

customer_name = "Alice"

print(f"Lenght of the name: {len(customer_name)}")
````

````python
# For dictionary, it counts the number of keys.

customer = {
    "name" :  "Alice",
    "age" : 25, 
    "city" : "Mumbai"
}

print(f"Lenght of the dictionary: {len(customer)}")
````

`Important: len(transactions) counts records, not unique customers. One customer may have several transactions.`

#### **B. sum() — Calculate the Total**

````python
sales = [200, 500, 300, 800]

total_sales = sum(sales)

print(f"Total Sales: {total_sales}")
````

```
This is useful for calculating total sales, expenses, revenue, and other numerical totals.

You can also provide a starting value:
```

````python
numbers = [10, 20, 30]

print(sum(numbers, 100))
````

`Python starts at 100 and adds the numbers.`

### **A common mistake**

`sum() is designed for numerical values, not strings.`

````python
names = ["Alice", "Bob"]

# print(sum(names))
````

### **C. min() — Find the Smallest Value**

````python
prices = [800, 250, 1200, 400]

print(f"Smallest Price: {min(prices)}")
````

`Useful for finding the lowest price, smallest measurement, or minimum transaction`

`You can also find the smallest string according to Python's string ordering, but that does not necessarily mean the shortest string.`

````python
names = ["Zoe", "Alice", "Bob"]

print(min(names))
````

`Python compares strings lexicographically, based on their character values. It isn't comparing their lengths.`

### **D. max() — Find the Largest Value**

````python
sales = [200, 500, 300, 800]

print(f"Lagest Sale: {max(sales)}")
````

```

Useful for finding the highest sale, largest transaction, or maximum measurement.

Together, min() and max() can help you inspect a dataset's range.

```

````python
print(f"Minimum Sale: {min(sales)}")
print(f"Maximum Sale: {max(sales)}")
````

`Data Scientist warning: Minimum and maximum values can be strongly affected by unusual observations. They don't tell you the typical value of a dataset.`

### **E. sorted() — Arrange Values**

`Suppose you want to arrange sales from smallest to largest.`

````python
sorted_sales = sorted(sales)
print(f"Sorted Sales: {sorted_sales}")
````

`To arrange values from largest to smallest:`

````python
sorted_sales = sorted(sales,reverse= True)
print(f"Sorted Sales: {sorted_sales}")
````

```
The important difference: sorted() versus .sort()

Look at these two approaches.

```

### **Approach - 1: `sorted()`**

````python
sales = [300, 200, 500, 800]

result = sorted(sales)

print(result)
````

`sorted() returns a new sorted list. The original list remains unchanged.`

### **Approach - 2: `.sort()`**

````python
sales = [200, 800, 300, 500]

sales.sort()

print(sales)
````

`.sort changes the original list.`

`Remember this distinction. Accidentally overwriting your original data can create problems during analysis.`

### **F. round() — Round a Number**

````python
price = 123.4567

print(round(price, 2))
````

`The second argument specifies how many decimal places to round to.`

````python
print(round(2.5))
print(round(3.4))
````

### **3. Combine Functions to Answer Business Questions**

````python
# A compay has the following sales:

sales = [200, 500, 800, 700]
````

```

Suppose the manager asks:

How many sales are recorded?

What is the total sales amount?

What is the highest sale?

What is the lowest sale?

What do the sales look like in ascending order?

You can answer all five questions with the functions you've learned.

```

````python
sales = [200, 500, 800, 700]

# Count the sales records.
number_of_sales = len(sales)

# Calculate the total sales.
total_sales = sum(sales)

# Find the highest and lowest sales.
highest_sale = max(sales)
lowest_sale = min(sales)

# Create a sorted compy without changing the original list.
sorted_sales = sorted(sales)

print(f"Number of sales: {number_of_sales}")
print(f"Totla sales: {total_sales}")
print(f"Highest Sale: {highest_sale}")
print(f"Lowest Sale: {lowest_sale}")
print(f"Sorted sales: {sorted_sales}")
````

### **Challenge 12 — `Learn zip()`**


````python
products = ["Laptop", "Mouse", "Keyboard"]
prices = [50000, 500, 1500]
````

You want to connect each product with its price.

You could access the lists using indexes, but zip() makes this simpler:

````python
for product, price in zip(products, prices):
    print(f"{product}: ₹{price}")
````

```

It pairs items according to their positions: first with first, second with second, and so on.

Important: By default, zip() stops when the shortest input iterable runs out of items.
```

### **Challenge 15 — Test Your Understanding**

````python
sales = [300, 500, 300, 900, 1200, 500]
````

Count the total number of sales records.

Count the number of unique sale amounts.

Check whether any sale is greater than ₹1,000.

Check whether all sales are greater than ₹200.

Print the sales sorted from largest to smallest


### **Solution**

````python
# total number of sales.
number_of_sales = len(sales)

# unique sale amount.
sale_unique = len(set(sales))


# Sale greater than 1000.
highest_sale = any(sale > 1000 for sale in sales)

# All sales greater than 200
lowest_sales = all(sale > 200 for sale in sales)

# Sorted Sales.
sorted_sales = sorted(sales, reverse=True)

print(f"Total number of sales: {number_of_sales}")
print(f"Unique sales amount: {sale_unique}")
print(f"Highest sale amount: {highest_sale}" )
print(f"Lowest sale amount: {lowest_sales}")
print(f"Sorted sales amount from largest to smallest: {sorted_sales}")
````
