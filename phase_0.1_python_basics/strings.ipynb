### **1. STRINGS**

`A string is a sequence of character`

```python
name = "Tom"
city = "Mumbai"
customer_id = "C10222"
```

`The quotes tell Python: 'Treat everything inside as text'`

### **2. CREATING STRINGS**

```python
# Single quotes
name = "Tom"

# Double quotes
city = "Mumbai"

print("Both creates a string.")
```

### **CHECK THE TYPE**

```python
print(type(name))
```

### **3. STRINGS ARE SEQUENCES(INDEXING)**

`This is important: Python treats a string as a sequence of individual characters`  

`Indexing: Acess one element at a time.`

```python
name = "Tom"

# Index:     0  1  2
# character: T  o  m

print(name[0])
print(name[1])
print(name[2])
```

`Python starts counting from 0, not 1.`

### **NEGATIVE INDEXING**

```python
print(name[-1])
print(name[-2])
print(name[-3])


# "Tom"

#  T   o   m
#  0   1   2
# -1  -2  -3
```

### **4. SLICING**

`slicing means taking a portion of a string`

```python
name = "Tom"
print(name[0:2])
```

`Rule: [start:stop:step]`

`Start is included. Stop is excluded`

### **Example**

```python
word = "Python"

print(word[0:6])
print(word[2:6])

# We can also omit the begining or end:
print(word[:6])
print(word[2:])
```

### **5. STRING METHODS: VERY IMPORTANT FOR DATA SCEINCE**

#### **We might recieve:**

"  Mumbai"

 "mumbai"

 "MUMBAI"

 " MUMBAI   "

`Human understand these as the same city.`

`A computer doesn't automatically know that.`


### `.strip()` 

`removes all the white spaces from both sides of a string.  `

```python
city = "  Mumbai  "

print(f"City with white spaces:{city}")
clean_city = city.strip()
print(f"Clean City:{clean_city}")
```

### `.lower()`

`Converts the text into lowercase.`

```python
city = "MUMBAI"
print(f"Lowercase: {city.lower()}")
```

### `.upper()`

`Converts the text into uppercase.`

```python
city = "   mumbai   "
print(f"Uppercase: {city.strip().upper()}")
```

### `.replace()`

`Replaces an element and place new element.`

```python
text = "Data Science is difficult"
text = text.replace("difficult", "interesting")

print(text)
# Data Science is interesting
```

### `.split()`

`Splits a string into individual parts.`

```python
email = "tom@gmail.com"
username = email.split("@")[0]
print(username)

# tom
```

### **6. JOINING STRINGS (CONCATENATION)**

`You can ask quesions about text`

```python
email = "tom@gmail.com"
print("@" in email)

# True
```

`You can also check the beginning/end`

```python
filename = "sales.csv"
print(filename.startswith("sales"))
# True
print(filename.endswith(".csv"))
# True
```

### **7. f-string**

`One of the most useful ways to combine variables with text`

```python
name = "Tom"
age = 20

message = f"My name is {name} and I am {age} years old."
print(message)
```
