### **2. What is an exception?**

`An exception is an error that occurs while a Python program is running and interrupts its normal execution.`

`````python
# For example

number = 10/0

# Python  raise: ZeroDivisionError
# Why? Because division by zero is not allowed.
`````

`````python
# Another example: 

customer_age = int("hello")

# Python raises: ValueError
# Why? Because hello cannot be converted into an integer.
`````

### **3. Your first exception-handling tool: try and except**

`Python provides try and except to handle exceptions.`


`````python
# consider this example:

try:
    result = 10 / 0
    print(result)

except ZeroDivisionError:
    print("Cannot divide by zero.")
`````

### **Let's understand it.**

`````python
try: 
    result = 10 / 0

# Python attempts to execute the code inside try.

# Except 

except ZeroDivisionError:
    print("Cannot divide by zero.")

# If Python raises a ZeroDivisionError inside the try block, 
# it executes the matching except block.
`````

`Important: Exception handling does not magically fix the underlying problem. It lets us decide what the program should do when a known problem occurs.`

### **4. Apply this to a data science problem**

`We can protect against an empty sales list:`

`````python
sales = []

try:
    average_sales = sum(sales) / len(sales)
    print(f"Average sales: {average_sales}")

except ZeroDivisionError:
    print("Cannot calculate the average because there are no sales records.")
`````

`````python
sales = []

if len(sales) > 0:
    average_sales = sum(sales) / len(sales)
    print(f"Average sales : {average_sales}")

else:
    print("No sales records are available.")
`````

`Both approches handle the empty-list situation.`

### **Next Challenge: else and finally**

`else — Runs when the try block finishes without an exception.`

`
finally — Runs whether an exception occurs or not, under normal control flow.`

`````python
try:
    number = int("50")
    result = 100 / number

except ValueError:
    print("Invalid number.")

except ZeroDivisionError:
    print("Cannot divide by zero.")

else:
    print("Calculation successful.")

finally:
    print("Calculation attempt finished.")
`````
