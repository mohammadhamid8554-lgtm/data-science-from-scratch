### **1. VARIABLS**

`Definition: A variable is a name that refers to a value.`

`Analogy --> Varible = labeled box containing data.`

```python
name : str = "Tom"          # str --> text
age : int = 54              # int --> whole numbers
salary : float = 45_000.0   # float --> decimal number
is_student : bool = True    # bool --> True/False
```

### **2. CHECKING DATA TYPES**

`type() function tells us what kind of data a varible contains.`

`This becoms VERY important when working with real dataset.`

```python
print(f"Name Type: {type(name)}")
print(f"Age Type: {type(age)}")
print(f"Salary Type: {type(salary)}")
print(f"Is_Student Type: {type(is_student)}")
```

### **3. BASIC NUMERICAL OPERATIONS**

`Python can perform mathematical operations direclyt.`

```python
income : float  = 50_000.0
expenses : float  = 20_000.0

profit = income - expenses

print(f"Profit: {profit}")
```

### **4. BASIC STRING OPERATIONS**

`Strings represent text data.`

`Text cleaning become a major part of Data Science later.`

```python
first_name : str = "Tom"
last_name : str = "Smith"

full_name = f"{first_name} {last_name}"

print(f"My name is: {full_name}.")
```

### **5. BOOLEAN VALUES**

`Boolean values represent a condition: True or False`

```python
age:int = 20
is_adult:bool = age>=18
print(is_adult)
```

### **6. COMPARISION OPERATORS**

`Comparision produce Boolean values.`

`These become the foundation of filtering data.`

```python
age:int = 20

print(age == 20)    # Equal to
print(age != 20)    # not equal to
print(age>18)       # grater than
print(age<18)       # less than
print(age>=18)      # grater than equal to
print(age<=18)      # less than equal to
```

### **7. TYPE CONVERSION**

`Sometimes data arrives in the WRONG type.`

`Example: '25' is text, not a number.`

```python
age_text:str = "25"

# Convert string -> integer
age_number:int = int(age_text)
print(f"Age number: {age_number}")
print(f"Age number type: {type(age_number)}")

# Another common conversion:
# integer -> float

price:int = 100
price_decimal:float = float(price)
print(f"Price decimal: {price_decimal}")

# Number -> String

customer_id:int = 12345
customer_id_text:str = str(customer_id)
print(f"Type of customer id text: {type(customer_id_text)}")
```

### **8. MULTIPLE VARIABLS**

`Python allows multiple assignments in one line.`

```python
age, salary, experience = 20, 50000.0, 2
print([age, salary, experience])
```

### **9. REASSIGNING A VARIABLE**

`A variable can point to a new value.`

```python
age:int = 20
age:int = 21
print(age)
```

### **10. A SMALL DATA-SCIENCE EXAMPLE**

`Imagine one row from a customer dataset.`

```python
customer_name:str = "Alice"
customer_age:int = 32
monthly_income:float = 75_000.0
is_premium_customer:bool = True

# Inspect the data before working with it.

print("Quick Inpection:")
print((customer_name))
print((customer_age))
print((monthly_income))
print((is_premium_customer))
print("\nData Types:")
print(type(customer_name))
print(type(customer_age))
print(type(monthly_income))
print(type(is_premium_customer))
```
