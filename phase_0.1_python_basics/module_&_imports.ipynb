## **1. Modules**

#### `A module is a Python file that contains reusable code, such as functions and variables`

`Think of it like a toolbox.`

## **2. import**

#### `Python doesn't automatically give you direct access to every module's contents.`

`You can make a module available to your code using import.`

### **Practical example**

````python
import math

number = 25

result = math.sqrt(number)
print(result)
````

#### Let's understand every line.

line 1. import math -> This makes the math module available to your code.

line 2. number= 25 -> We store the number 25 in a variable.

line 3. result = sqrt(number) -> math is module, sqrt func, number is 
        the   input, result stores the answer.

line 4. print(result) -> This displays the result.

### **Understand this pattern**

````python
# module_name.function_name()

math.sqrt(25)
math.factorial(5)
````

`The first calculates a sqrt root. The second calculates a factorial.`

### **3. Your first practical data science example**

````python
import statistics

daily_sales = [1000, 1500, 1200, 1800, 1500]

average_sales = statistics.mean(daily_sales)

print(average_sales)
````

## What happened?

We imported the statistics module.

We stored the sales records in a list.

We used statistics.mean() to calculate the average.

We displayed the result.

The important idea is that you used an existing function rather than writing the average calculation yourself.

This is how data scientists work: they combine existing tools with their understanding of the problem.

### **Next concept: Two ways to import functions**

#### **`Method 1: Import the module`**

````python
import math 

result = math.sqrt(81)

print(result)
````

`Here, we write math.sqrt() because the function belongs to the math module.`

#### **`Method 2: Import a specific function`**

````python
from math import sqrt

result = sqrt(81)

print(result)
````

`Here, we imported sqrt directly, so we can call sqrt() without writing math..`

` prefer import math for now.`

### **Next concept: Importing multiple functions**

````python
import math 

number = 10

root  = math.sqrt(number)
rounded_root = round(root, 2)

print(f"Rounded root of the number: {rounded_root}")
````
