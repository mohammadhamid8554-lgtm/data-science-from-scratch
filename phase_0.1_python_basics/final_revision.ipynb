# 4. Loops

Loops repeat code.

### for Loop

Used when you want to iterate over a sequence or collection.

```python
for number in [10, 20, 30]:
    print(number)
```

### range()

```python
range(5)        # 0, 1, 2, 3, 4
range(1, 6)     # 1, 2, 3, 4, 5
range(1, 10, 2) # 1, 3, 5, 7, 9
```

Example:

```python
for i in range(1, 6):
    print(i)
```

### while Loop

Repeats while a condition is true.

```python
count = 1

while count <= 5:
    print(count)
    count += 1
```

### break

Stops the loop immediately.

```python
for number in range(10):
    if number == 5:
        break
    print(number)
```

### continue

Skips the current iteration.

```python
for number in range(5):
    if number == 2:
        continue
    print(number)
```

**Remember:**
- `break` = stop the loop.
- `continue` = skip this iteration.
- Always make sure a `while` loop can eventually stop.

# 10. Sets

A set stores unique values.

```python
numbers = {10, 20, 20, 30}

# Result: {10, 20, 30}
```

### Important Set Methods

```python
numbers.add(40)
numbers.remove(20)
numbers.discard(50)
numbers.clear()
```

### Set Operations

```python
a = {1, 2, 3}
b = {3, 4, 5}

a.union(b)        # {1, 2, 3, 4, 5}
a.intersection(b) # {3}
a.difference(b)  # {1, 2}
```

**Remember:**
- Sets remove duplicates.
- Sets are not indexed.
- `remove()` raises an error if the item is missing.
- `discard()` does not raise an error if the item is missing.
- `{}` creates an empty dictionary, not an empty set. Use `set()`.

# 11. List and Dictionary Comprehensions

Comprehensions provide a shorter way to create collections.

### List Comprehension

Normal approach:

```python
squares = []

for number in range(1, 6):
    squares.append(number ** 2)
```

Comprehension:

```python
squares = [number ** 2 for number in range(1, 6)]
```

### With Conditions

```python
even_numbers = [
    number for number in range(1, 11)
    if number % 2 == 0
]
```

### Dictionary Comprehension

```python
squares = {
    number: number ** 2
    for number in range(1, 6)
}
```

**Remember:** Use comprehensions when the logic is simple and readable. Do not force complex logic into one line.

# 14. Exception Handling

Exceptions are errors that happen while a program runs.

### try and except

```python
try:
    number = int(input("Enter a number: "))
    print(100 / number)

except ValueError:
    print("Please enter a valid number.")

except ZeroDivisionError:
    print("Cannot divide by zero.")
```

### else

Runs when no exception occurs.

```python
try:
    number = int(input("Enter a number: "))

except ValueError:
    print("Invalid input.")

else:
    print(number)
```

### finally

Runs whether an exception occurs or not.

```python
try:
    print("Working")

except Exception:
    print("Error")

finally:
    print("Program finished")
```

### Common Exceptions

| Exception | Meaning |
|---|---|
| `ValueError` | Correct type, invalid value |
| `TypeError` | Operation used with an unsuitable type |
| `ZeroDivisionError` | Division by zero |
| `FileNotFoundError` | File does not exist at the given path |
| `PermissionError` | Operation not permitted |
| `KeyError` | Dictionary key not found |
| `IndexError` | Index out of range |
| `JSONDecodeError` | Invalid JSON format |

**Remember:** Handle specific exceptions whenever possible. Avoid using a broad `except` without a good reason.

# 15. File Handling — Text Files

File handling allows Python to read and write data stored in files.

### Opening a File

```python
file = open("data/sales.txt", "r")
```

### File Modes

| Mode | Meaning |
|---|---|
| `"r"` | Read |
| `"w"` | Write; overwrites existing contents |
| `"a"` | Append; adds content at the end |

### Reading a File

```python
with open("data/sales.txt", "r") as file:
    content = file.read()
    print(content)
```

### Reading One Line

```python
with open("data/sales.txt", "r") as file:
    line = file.readline()
```

```python
with open("data/sales.txt", "w") as file:
    file.write("Product A, 500\n")
```

```python
with open("data/sales.txt", "a") as file:
    file.write("Product B, 700\n")
```

```python
with open("data/sales.txt", "r") as file:
    content = file.read()
    file.seek(0)
    first_line = file.readline()
```

### Important Concepts

- `read()` returns a string.
- `readline()` reads one line.
- `readlines()` returns a list of remaining lines.
- `seek(0)` moves the file position to the beginning.
- `with open()` automatically closes the file when the block ends.
- `"w"` can destroy existing file contents by overwriting them.
- `"r"` does not create a missing file.
- `write()` does not automatically add a newline.

# 17. Date and Time — datetime

The `datetime` module helps work with dates and times.

```python
from datetime import date, datetime, time, timedelta
```

### Today's Date

```python
today = date.today()

today.year
today.month
today.day
```

### Current Date and Time

```python
current = datetime.now()
```

### Create a Time

```python
t = time(14, 30, 0)

t.hour
t.minute
t.second
```

### Convert a String into a Date

Use `strptime()`.

```python
date_string = "03-11-2026"

converted_date = datetime.strptime(
    date_string,
    "%d-%m-%Y"
)
```

### Convert a Date into a String

Use `strftime()`.

```python
today = datetime.now()

formatted_date = today.strftime("%d-%m-%Y")
```

### Important Difference

| Method | Purpose |
|---|---|
| `strptime()` | String → datetime |
| `strftime()` | datetime → formatted string |

### Date Difference

```python
date1 = date(2026, 11, 3)
date2 = date(2026, 11, 10)

difference = date2 - date1

difference.days
```

### timedelta

Used to represent a duration.

```python
now = datetime.now()

future = now + timedelta(days=7)
past = now - timedelta(days=3)
```

### Time Difference

```python
start = datetime.now()
end = start + timedelta(hours=2)

difference = end - start

difference.total_seconds()
difference.total_seconds() / 3600
```

**Remember:**
- `.days` gives the whole-day component of a duration.
- `.total_seconds()` gives the total duration in seconds.
- Divide by `3600` to convert seconds into hours.
- `datetime` can include both date and time.
- `date` represents a date without a time.

# 18. CSV File Handling

CSV means **Comma-Separated Values**. It is commonly used to store tabular data.

Example `students.csv`:

```text
name,subject,marks
Aman,Math,85
Riya,Science,90
Tom,English,78
```

### Import CSV

```python
import csv
```

### Read Using csv.reader()

```python
with open("data/students.csv", "r", newline="") as file:
    reader = csv.reader(file)

    header = next(reader)

    for row in reader:
        print(row)
```

Each row is a list of strings.

Example:

```python
["Aman", "Math", "85"]
```

### Read Using csv.DictReader()

```python
with open("data/students.csv", "r", newline="") as file:
    reader = csv.DictReader(file)

    for student in reader:
        print(student["name"])
        print(student["marks"])
```

Each data row is a dictionary.

### Convert CSV Values

CSV values are usually read as strings.

```python
marks = int(student["marks"])
```

### Write Using csv.writer()

```python
import csv

with open("data/output.csv", "w", newline="") as file:
    writer = csv.writer(file)

    writer.writerow(["Name", "Marks"])
    writer.writerow(["Tom", 85])
```

### Write Using csv.DictWriter()

```python
with open("data/output.csv", "w", newline="") as file:
    fieldnames = ["name", "marks"]

    writer = csv.DictWriter(
        file,
        fieldnames=fieldnames
    )

    writer.writeheader()

    writer.writerow({
        "name": "Tom",
        "marks": 85
    })
```

### Missing and Invalid Values

```python
value = ""

if value == "":
    print("Missing value")
else:
    try:
        number = float(value)
    except ValueError:
        print("Invalid number")
```

**Remember:**
- `csv.reader()` → rows as lists.
- `csv.DictReader()` → rows as dictionaries.
- CSV values are generally strings when read.
- Convert numbers before doing calculations.
- Empty string `""` is not the same as zero `0`.
- Use `newline=""` when opening CSV files for reading or writing with the CSV module.

# 19. JSON File Handling

JSON means **JavaScript Object Notation**. It is a common format for storing and exchanging structured data.

Example `product.json`:

```json
{
    "name": "Laptop",
    "price": 50000,
    "available": true
}
```

### Import JSON

```python
import json
```

### Read JSON Using json.load()

```python
with open("data/product.json", "r") as file:
    product = json.load(file)
```

### JSON to Python Conversion

| JSON | Python |
|---|---|
| Object `{}` | Dictionary |
| Array `[]` | List |
| String | String |
| Number | `int` or `float` |
| `true` | `True` |
| `false` | `False` |
| `null` | `None` |

### Access JSON Data

```python
print(product["name"])
print(product["price"])
```

### Update JSON Data

```python
product["price"] = 52000
```

### Save Using json.dump()

```python
with open("data/product.json", "w") as file:
    json.dump(product, file, indent=4)
```

### Handle JSON Errors

```python
try:
    with open("data/product.json", "r") as file:
        product = json.load(file)

except FileNotFoundError:
    print("File not found.")

except json.JSONDecodeError:
    print("Invalid JSON format.")
```

### Important Difference

| Function | Purpose |
|---|---|
| `json.load()` | Reads JSON from a file |
| `json.dump()` | Writes Python data into a file |
| `json.loads()` | Converts JSON string into Python data |
| `json.dumps()` | Converts Python data into JSON string |

**Remember:**
- `json.load()` returns Python data, not a string.
- `json.dump()` writes data into a file.
- `indent=4` makes the JSON easier to read.
- Opening the file with `"w"` overwrites existing contents.
- Invalid JSON syntax raises `json.JSONDecodeError`.

# 21. Common Mistakes to Avoid

| Mistake | Correct Understanding |
|---|---|
| Assuming `input()` returns an integer | It returns a string |
| Confusing `=` and `==` | `=` assigns; `==` compares |
| Forgetting indentation | Python uses indentation to define blocks |
| Using `print()` instead of `return` | `return` sends a value back |
| Assuming `sort()` returns a new list | It changes the original list |
| Assuming `sorted()` changes the original list | It returns a new sorted list |
| Assuming sets preserve duplicates | Sets store unique values |
| Accessing missing dictionary keys blindly | Use `.get()` when appropriate |
| Forgetting CSV values are strings | Convert numeric values |
| Treating missing CSV values as zero | Check for empty values |
| Using `"w"` without considering existing data | It overwrites existing contents |
| Confusing `strptime()` and `strftime()` | One parses; the other formats |
| Assuming a path is relative to the script | Relative paths use the current working directory |
| Assuming `json.load()` returns a string | It returns a Python data structure |

# 22. Data Scientist's Python Mental Checklist

Before writing a Python program, ask yourself:

1. **Input:** What data am I receiving?
2. **Data type:** Is it a string, integer, float, list, dictionary, or another type?
3. **Validation:** Can the input be missing or invalid?
4. **Logic:** Do I need conditions, loops, or functions?
5. **Data structure:** Which collection is best for this problem?
6. **Calculation:** Do I need to convert values before calculating?
7. **Errors:** What can go wrong, and which exceptions should I handle?
8. **Output:** What result should the program produce?
9. **Storage:** Do I need to save the result in a TXT, CSV, or JSON file?
