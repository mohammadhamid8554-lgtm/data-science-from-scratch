#### `Indexing = get specific data.`

#### `Slicing = get a range of data.`

### **1. INDEXING**

`Python uses zero-based indexing.`

````python
# The first element has index 0, not 1.

sales = [1200, 1500, 900, 2000]

print(sales[0])  # 1200
print(sales[1])  # 1500
print(sales[2])  # 900
print(sales[3])  # 2000


# Think:

# Value:   1200   1500   900   2000

# Index:      0      1     2      3
````

`This becomes extremely important in Numpy and Pandas.`

### **2. NEGATIVE INDEXING**

`Python also lets you count from the end.`

````python
sales = [1200, 1500, 900, 2000]

print(sales[-1])  # 2000
print(sales[-2])  # 900
print(sales[-3])  # 1500

# Think:

# Value:   1200   1500   900   2000
# Index:      0      1     2      3
#           -4     -3    -2     -1
````

`This is especially useful when you want the last few observations.`

### **3. SLICING**

````python
# Suppose:

sales = [1200, 1500, 900, 2000, 1800]


# You want the first three values.

print(sales[:3])

# [start : stop]

# Last element excluded.
````

### **OMITTING START AND STOP**

````python
# You don't always need both.

sales = [1200, 1500, 900, 2000, 1800]

print(sales[:3])
# First 3 elements

print(sales[2:])
# From index 2 to the end

print(sales[:])
# Entire list

# This is very common.
````

### **5. USING A STEP**

````python
# Slicing has a third component:
# [start : stop : step]

numbers = [0, 1, 2, 3, 4, 5, 6]

print(numbers[::2])

# We are taking every 2 element.

print(numbers[1::2])
````

### **6. REVERSING A SEQUENCE**

````python
# A very useful trick:

numbers = [1, 2, 3, 4, 5]

print(numbers[::-1])

# Why?

# step = -1
# → move backwards
````

### **7. INDEXING VS. SLICING**

````python
# This distinction must become automatic.

# Indexing

print(sales[2])

# One element.

# Slicing 

print(sales[1:4])

# A sequence of elements.
````

```
A sequence of elements.

Mental model:

Indexing
→ "Give me this one."

Slicing
→ "Give me this section."

```

### **8. STRING USE THE SAME SYSTEM**

`This is why learning the concept now matters.`

````python
text = "Python"

print(text[0])
# P

print(text[1:4])
# yth

print(text[-1])
# n

````

`Same rules.`

`That's because strings are sequences.`
