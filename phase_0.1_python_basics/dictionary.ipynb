`A dictionary soters data as:`

`key --> values`

`think of it like a real dictionary:`

`word -> meaning`

````python
student = {"name" : "Tom",
            "age" : 20,
            "city" :  "Mumbai"
}

print(f"Student Name: {student['name']}")
````

### **1. ACCESSING VALUES**

````python
print(f"Student Name: {student['name']}")
print(f"Student Age: {student['age']}")
````

`Unlike lists, you don't normally access dictionary data by position.`

`List : index -> value`

`Dictionary -> value`

### **2. WHY DICTIONARIES MATTERS IN DATA SCIENCE**

````python
customer = {
    "customer_id": 101,
    "age": 25,
    "income": 65000.0,
    "is_premium": True
}

print(f"Customer Income: {customer['income']}")
````

`This key-value idea appears everywhere in the Data Science ecosystem.`

`you can access what exatly you need.`

### **3. ADDING NEW DATA**

`Dictionaries are mutable`

````python
customer = {
    "name" : "Tom",
    "age" :  20
}

customer["city"] = "Mumbai"

print(f"Customer Data: {customer}")
````

### **4. UPDATING DATA**

````python
customer = {
    "name" :  "Tom",
    "age" : 20
}

customer["age"] = 40

print(f"Updated Customer data: {customer}")
````

### **5. REMOVING DATA**

````python
customer = {
    "customer_id": 101,
    "age": 25,
    "income": 65000,
    "is_premium": True
}

del customer["income"]
print(customer)
````

` You can also use pop()`

````python
age = customer.pop("age")
print(age)
````

### **6 . get() - AN IMPORTANT HABIT**

`Suppose the key doesen't exist:`

````python
customer = {
    "customer_id": 101,
    "age": 25,
    "income": 65000,
    "is_premium": True
}

print(customer["city"])
````

`This causes a KeyError.`

````python
print(customer.get("city", "Unknown"))
````

`This is safer when workiing with data where feilds may be missing.`

### 7. **GETTING KEYS AND VALUES**

````python
customer = {
    "customer_id": 101,
    "age": 25,
    "income": 65000,
    "is_premium": True
}

print(customer.keys())
print(customer.values())
print(customer.items())
````

`Conceptually: keys() -> names of feilds, values() -> actual values, items() -> key-value pairs.`

````python
for key, value in customer.items():
    print(f"{key}: {value}")
````

### **8. CHECKING WHETHER A KEY EXISTS**

````python
customer = {
    "customer_id": 101,
    "age": 25,
    "income": 65000,
    "is_premium": True
}

print("age" in customer)
````

`Notice: ' ( "age" in customer ) ' checks the keys, not the values.`

### **9. NESTED DICTIONARIES**`

````python
company = {
    "name": "ABC",
    "employee": {
        "name": "Tom",
        "age": 20
    }
}
````

`Access nested data:`

````python
print(company["employee"]["age"])
````

`You will encounter structures like this when dealing with JSON and APIs.`

### **10. DICTIONARY + LIST**

````python
customers = [
    {"name": "Tom", "age": 20},
    {"name": "Alice", "age": 25},
    {"name": "John", "age": 31}
]
````

customers
