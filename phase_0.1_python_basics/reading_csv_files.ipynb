` CSV stands for Comma-Separated Values.`

`Data scientist commonly receive data in CSV files because they are a simple way to store rows and columns.`

`You could read the entire file by using read(). It gives long string.`

`Python has a built-in module called csv that helps with this.`

````python
import csv

with open("data/sales.csv", "r", newline="") as file:
    reader = csv.reader(file)
    
    for row in reader:
        print(row)
````

### **Let's understand three new things:**

1. import csv: imports Python's built-in CSV module.

2. csv.reader(file): reads the CSV file row by row.

3. Each row is returned as a list of string.

### **2. Next concept: Write CSV files**

```

So far, you've learned how to read CSV files. Data scientists also need to create and save CSV files.

Imagine you calculate the total sales for several products and want to save the results for later analysis.

Python provides csv.writer() for this.


```

````python
import csv

with open("data/sales_summary.csv", "w", newline="") as file:
    writer = csv.writer(file)

    writer.writerow(["product", "total_sales"])
    writer.writerow(["Laptop", 100000])
    writer.writerow(["Mouse", 4000])
    
````

### **Let's understand each part**

- "w": opens file for writing.

- if the file doesn't exist, python creates it.

- if it already exists, Python overwrites its previous contents. 

- newline = "": is the recommended way to open CSV files for writing with Python's csv module.

- csv.writer(file):creates an object that helps write CSV rows.

- writer.writerrow(...): Writes one row at a time.

### **2. Next concept: csv.DictWriter()**

`Python also provides csv.DictWriter(), which writes CSV rows using dictionaries.`

````python
import csv 

with open("data/student_results.csv", "w", newline="") as file:
    fieldnames = ["student", "marks", "attendance"]

    writer = csv.DictWriter(file, fieldnames=fieldnames)

    writer.writeheader()

    writer.writerow({
        "student": "Aman",
        "marks" : 85,
        "attendance" : 90
    })

    writer.writerow({
    "student": "Priya",
    "marks" : 92,
    "attendance" : 95
       
        })
````

| Code                     | Meaning                                     |
| ------------------------ | ------------------------------------------- |
| `fieldnames`             | Defines the column names and their order.   |
| `csv.DictWriter(...)`    | Creates a writer that accepts dictionaries. |
| `writeheader()`          | Writes the column names as the first row.   |
| `writer.writerow({...})` | Writes one dictionary as a CSV row.         |

### **How to handle a missing CSV value**

````python
row = ["Mouse", "", "5"]

if row[1] != "":
    price = float(row[1])
    print(price)
else:
    print("Price is missing.")
````
