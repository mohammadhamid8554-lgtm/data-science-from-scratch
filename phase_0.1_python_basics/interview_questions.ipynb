Question 1: Data Manipulation & Edge Cases (Loops, Dictionaries, Exception Handling)

Scenario: You receive a raw list of customer transaction records as strings formatted as "user_id,amount". Some strings are corrupted (e.g., missing values, invalid numbers, or wrong format like "U101,45.5", "U102,invalid", "U101").

Question: Write a function using standard Python (no external libraries like Pandas) that parses this list and returns a dictionary with each user_id mapped to their total valid spent amount. Your code must gracefully handle exceptions without crashing and ignore corrupted entries.



```python
# Sample Input:
raw_data = ["101,45.50", "102,90.00", "101,bad_val", "103", "102,10.50"]

def transformed_data(raw_data: list) -> dict:
    my_dict = {}

    for record in raw_data:
        try:
             user_id, amount_text = record.split(",")
             amount = float(amount_text)
             my_dict[user_id] = my_dict.get(user_id, 0) + amount

        except (ValueError, TypeError):
            continue
    return my_dict

transformed_data(raw_data=raw_data)

```

Question: Write a single dictionary comprehension that filters for employees in the "Data" department with a score \ge 80, returning a dictionary where the keys are employee names and the values are their scores.

```python
employees = [
    {"name": "Alice", "score": 88, "department": "Data"},
    {"name": "Bob", "score": 55, "department": "Data"},
    {"name": "Charlie", "score": 92, "department": "Eng"},
    {"name": "Diana", "score": 79, "department": "Data"}
]

filtered_employees = {employee["name"]: employee['score'] for employee in employees if employee.get("department") == "Data" and employee.get("score", 0) >= 80}
print(f"Filtered Employees: {filtered_employees}")
```
