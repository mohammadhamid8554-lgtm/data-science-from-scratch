### **2. Creating a function**

`The basic structure is:`

```

def function_name(parameters):
    # logic
    return result

```

`````python
def calculate_revenue(price, quantity):
    return price * quantity
`````

```
 There are three important pieces:

def

Tells Python:

"I'm defining a function."

Parameters
price, quantity

These are the inputs the function expects.

return

Sends the result back to whoever called the function.


```

### **3. Calling a function**

`Defining a function does not execute it.`

`````python
def calculate_revenue(price, quantity):
    return price * quantity
`````

` Nothing happens yet.`

`You have to call it.`

`````python
revenue = calculate_revenue(100, 3)
print(revenue)
`````

### **4. Parameters vs Arguments**

`This distinction is worth knowing.`

`````python
def calculate_revenue(price, quantity):
    return price * quantity

# Price and quantity are parameters.

# When you call:

calculate_revenue(100, 3)

# 100 and 3 are arguments
`````

```

Think:

Parameters → placeholders
Arguments  → actual values

```

### **5. Why return matters**

`Compare these:`

### Version 1

`````python
def calculate_revenue(price, quantity):
    print(price * quantity)
`````

### Versioin 2

`````python
def calculate_revenue(price, quantity):
    return price * quantity
`````

```

The first prints the result.

The second gives the result back.

That difference is extremely important.

```

### **With Return**

`````python
revenue = calculate_revenue(100, 3)

def calculate_revenue(price, quantity):
    return price * quantity

revenue = calculate_revenue(100, 3)
profit = revenue - 100
print(f"Profit: {profit}")
`````

```

Now the function's output can be used somewhere else.

Data Scientist rule

Use return when the result needs to be reused.

```

### **6. A function can have multiple inputs**

`````python
def calculate_profit(revenue, cost):
    return revenue - cost

# Then:

profit = calculate_profit(1000, 350)
print(f"Profit: {profit}")
`````

### **7. A function can return multiple values**

`````python
# Python allows:

def summarize(values: list):
    return min(values), max(values)

minimum, maximum = summarize([12, 21, 3, 90, 40])
print(f"Minimum value: {minimum}\nMaximum value: {maximum}")
`````

```

This works because Python returns them together as a tuple.

Notice how your earlier knowledge of tuples comes back here.

```

### **8. Default parameters**

`Sometimes an input has a sensible default.`

`````python
def calculate_discount(price, discount = 0.10):
    return price * (1 - discount)

calculate_discount(1000)

# Uses the default 10% discount or you can provide another value:

calculate_discount(1000, 0.20)
`````

### **9. Data Science example**

`Suppose you repeatedly need to clean customer names.`

`````python
def clean_name(name):
    return name.strip().lower()

name1 = clean_name("Tom   ")

print(f"Name: {name1}")
`````

```

everywhere, you have one reusable rule.

This becomes extremely useful when you start cleaning real datasets.

```

### **Common Traps**

`1. Defining ≠ calling`

`````python
def add(a, b):
    return a + b

# Doesn't execute the function.
# You need:

addition = add(10, 30)

print(f"Sum of the numbers: {addition}")
`````

`2. print() ≠ return`

`````python
def add(a, b):
    print(a + b)

# prints the answer but doesn't return it.

def sub(a, b):
    return (a - b)

# produces a value that can be sotred and reused.
`````
