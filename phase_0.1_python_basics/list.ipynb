### **1. ACCESSING ELEMENTS OF A LIST**

```python
ages =  [21, 25, 30, 19, 32]
print(ages[0])
print(ages[1])
print(ages[-1])
print(ages[-3])
```

### **3. SLICING LISTS**

```python
sales =  [1200, 1500, 900, 2100]
print(sales[:3])
```

` Remember: [start: stop] --> start is included, stop is excluded`

### **3. LIST ARE MUTABLE**

```python
sales =  [1200, 1500, 900, 2100]
sales[1] = 2000
print(sales)
```

`The list itself changed.`

`Compare that mentally with a string`

```python
name = "Tom"

# You cannot directly change one character of a string.
# Strings are immutable.

```

`List -> mutable`

`String -> immutable`

`You need this distinction later when working with Python objects and data structure`

### **4. ADDING DATA**

### `.append()`


`Adds one item to the end.`

```python
sales =  [1200, 1500, 900, 2100]

sales.append(3000)

print(sales)
```

### `.extend()`

`Adds multiple items.`

```python
sales =  [1200, 1500, 900, 2100]
sales.extend([3000, 5000])
print(sales)
```

`Don't confuse them:`

`append one list inside the list`

`extend adds  TWO values`

###  **5. REMOVING DATA**

```python
sales =  [1200, 1500, 900, 2100]

# Remove a specific value
sales.remove(1200)

print(sales)
```

### `Remove by position`

```python
sales =  [1200, 1500, 900, 2100]
removed = sales.pop(1)

print(removed)

print(sales)
```

### **6. USEFUL BUILT-IN FUNCTION**

`Python gives you several useful fuctions for lists.`

```python
sales =  [1200, 1500, 900, 2100]

print(len(sales))
print(sum(sales))
print(min(sales))
print(max(sales))
```

`You can calculate an average:`

```python
average_sales = sum(sales)/len(sales)
print(f"Average Sales: {average_sales}")
```

`This is useful bridge toward statistics and numpy`

### **7. SORTING**

```python
sales =  [1200, 1500, 900, 2100]

sales.sort()

print(sales)
```

`Descending:`

```python
sales.sort(reverse=True)
print(sales)
```

**`Important distinction `**

`sort() chnages the original list.`

`sorted() creates a sorted result without modifying the orginal`

```python
sales =  [1200, 1500, 900, 2100]

new_sales = sorted(sales)
print(sales)

print(new_sales)
```

### **8. CHECKING MEMBERSHIP**

```python
sales =  [1200, 1500, 900, 2100]
print(1500 in sales)
print(5000 in sales)
```

`This becomes useful when checking whether a value/category exists.`
