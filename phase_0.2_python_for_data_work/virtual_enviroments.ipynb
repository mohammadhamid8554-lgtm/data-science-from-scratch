### **Next concept: Managing project dependencies**
Imagine you finish a data science project and share it with another developer.

Your project uses:
- Pandas 2.2.0
- NumPy 1.26.4
- Matplotlib 3.8.0
How will the other developer know which packages and versions your project needs?

This is where requirements.txt becomes useful.

### **First, understand pip freeze**

```python
%python -m pip freeze > requirements.txt
# Shows installed packages in a requirements-file format, using version specifications.
```

```python
%python -m pip list
# Shows installed packages and their versions in a table.
```

### **Creating requirements.txt file**

Suppose another developer receives your project and its requirements.txt.

They can create their own virtual environment and install the listed dependencies using:

```python
%python -m pip install -r requirements.txt
```

### **Two important professional practices**
#### **Practice 1: Don't share your virtual environment folder.**

1. Your Python source files.
2. Your datasets, where appropriate.
3. requirements.txt
- The other developer creates their own enviroments.

#### **Practice 2: Keep your enviroment separate from project files.**

- The venv/ folder is generally excluded from version control. The requirements file is shared.

Real-world: A teammate cannot recreate your environment using requirements.txt. What possible problems would you investigate?

1. Answer: A virtual environment is a Python workspace that isolates packages for different projects.

2. `pip list` shows a table of installed package names and their versions. `pip freeze` shows the exact package versions in a format suitable for `requirements.txt`.

3. The command installs the pandas package into the currently active Python interpreter in the activated virtual environment.

4. We should share the `requirements.txt` file so that other developers can install the required packages in their own environments. This is the professional way to manage dependencies instead of sharing the entire virtual environment folder in version control, such as on GitHub.
