# VS Code for Data Science — Mega Revision Cheat Sheet

## 1. What is VS Code?

Visual Studio Code (VS Code) is a code editor and development environment used to write, run, debug, and manage code.

For Data Science, we can use it to:
- Write Python programs (`.py`).
- Work with Jupyter Notebooks (`.ipynb`).
- Manage virtual environments.
- Install and manage Python packages.
- Run Python programs through the terminal.
- Debug errors and organize projects.

## 2. Important VS Code Components

| Component | Purpose |
|---|---|
| Editor | Write and edit code |
| Explorer | Manage files and folders |
| Terminal | Execute commands |
| Command Palette | Find and execute VS Code commands |
| Extensions | Add functionality |
| Python Extension | Python language support |
| Jupyter Extension | Notebook support |

## 3. Important Keyboard Shortcuts

| Shortcut | Action |

| `Ctrl + Shift + P` | Open Command Palette |
| `Ctrl + `` ` | Open/close integrated terminal |

| `Shift + Enter` | Run selected line/cell in supported contexts |

## 4. Python Interpreter

### What is a Python Interpreter?

A Python interpreter is the program that executes Python code.

One computer can have multiple Python interpreters:

- System Python
- Project virtual environment Python
- Python from another project environment

### Why does interpreter selection matter?

Different interpreters can have different installed packages and Python versions.

For example:

```text
Project/
│
├── venv/
│   └── Scripts/
│       └── python.exe
│
├── main.py
└── requirements.txt
```

If Pandas is installed inside `venv` but VS Code runs your script using system Python, you may get:

```text
ModuleNotFoundError: No module named 'pandas'
```

### How to select an interpreter

1. Press `Ctrl + Shift + P`.
2. Search for `Python: Select Interpreter`.
3. Select the interpreter inside your project's virtual environment.

## 5. Verify the Active Python Interpreter

Use the following code:

```python
import sys

print(sys.executable)
```

It displays the exact Python executable running your code.

Example:

```text
C:\project\venv\Scripts\python.exe
```

If the path points to your project's environment, the correct interpreter is running.

**Important:** Check `sys.executable` inside the actual script or notebook you are debugging.

## 6. Virtual Environment in VS Code

A virtual environment keeps project dependencies separate.

### Activate on Windows

```bash
venv\Scripts\activate
```

### Deactivate

```bash
deactivate
```

### Install a package

```bash
python -m pip install pandas
```

### Check installed packages

```bash

```

### Check package information


### Important distinction

| Activation | Interpreter Selection |
|---|---|
| Configures the terminal session | Selects the Python executable for VS Code Python features |
| Done through terminal commands | Done through Command Palette |
| Affects commands in that terminal | Does not automatically activate every terminal |

## 7. Run Button vs Terminal

### Run Button

- Convenient way to execute Python files.
- Reduces manual command typing.
- Uses the execution configuration and interpreter selected by VS Code.

### Terminal

Execute a Python file manually:

```bash
python main.py
```

The terminal gives greater control over commands, environment activation, package installation, and execution.

**Remember:** Both methods can execute Python code, but the terminal is essential for managing real development workflows.

## 8. Python Files vs Jupyter Notebooks

| Python File (`.py`) | Jupyter Notebook (`.ipynb`) |
|---|---|
| Usually executes as a script | Executes code in individual cells |
| Commonly run from top to bottom | Cells can execute out of order |
| Good for reusable programs | Good for exploration and analysis |
| Usually contains code | Can contain code, Markdown and outputs |

VS Code supports both formats.

## 9. Jupyter Kernel in VS Code

A kernel is the process that executes notebook cells and maintains the notebook's Python state.

### Important concepts

- Kernel stores variables in memory.
- Restarting the kernel clears in-memory variables.
- Interrupting stops currently running code.
- Notebook cells can execute out of order.
- The notebook kernel can differ from the selected script interpreter.

### Example

```python
# Cell 1
sales = [100, 200, 300]

# Cell 2
average = sum(sales) / len(sales)

# Cell 3
print(average)
```

If you change Cell 1 but don't rerun Cell 2, the `average` variable can contain an old result.

**Notebook visual order does not guarantee execution order.**

### Restart + Run All

Use this to check whether your notebook works from a fresh kernel.

It helps identify:
- Missing variable definitions.
- Incorrect cell dependencies.
- Out-of-order execution.
- Hidden dependencies on previous kernel state.

## 10. Python Interpreter vs Jupyter Kernel

| Python Interpreter | Jupyter Kernel |
|---|---|
| Executes Python code | Executes notebook cells |
| Used by Python scripts | Used by notebooks |
| Selected through Python interpreter settings | Selected through notebook kernel selector |
| Can run independently of Jupyter | Maintains notebook session state |

Selecting an interpreter for Python scripts does not always guarantee that your notebook uses the same environment.

## 11. ipykernel

`ipykernel` allows a Python environment to operate as a Jupyter kernel.

Install it inside the intended environment:

```bash
python -m pip install ipykernel
```

If the environment does not appear in the notebook kernel list, it may need to be registered.

```bash
python -m ipykernel install --user --name venv --display-name "Python (venv)"
```

Then select `Python (venv)` from the notebook kernel selector.

## 12. Debugging ModuleNotFoundError

Example error:

```text
ModuleNotFoundError: No module named 'pandas'
```

### Step-by-step debugging

**Step 1: Check the active interpreter**

```python
import sys

print(sys.executable)
```

**Step 2: Check whether Pandas is installed**

In the terminal:

```bash
python -m pip show pandas
```

Inside a notebook:

```python
%pip show pandas
```

**Step 3: Install Pandas in the correct environment**

In the terminal:

```bash
python -m pip install pandas
```

Inside a notebook:

```python
%pip install pandas
```

**Step 4: Retry the import**

```python
import pandas as pd
```

If necessary, restart the notebook kernel.

**Golden rule:** Install packages in the environment that actually executes your code.

## 13. Common VS Code Problems

| Problem | Possible Cause | What to Check |
|---|---|---|
| `ModuleNotFoundError` | Package missing from active environment | Interpreter and package installation |
| Script works but notebook fails | Different interpreter and kernel | Notebook kernel selection |
| Wrong Python version | Incorrect interpreter selected | `sys.executable` |
| Notebook cannot find kernel | Environment not registered or kernel support missing | `ipykernel` |
| Terminal uses unexpected Python | Wrong environment active | `python -m pip --version` and `where python` |
| Notebook produces old results | Cells executed out of order | Restart + Run All |

## 14. Complete Data Science Project Workflow

A typical Python project structure:

```text

│
├── venv/
│
├── data/
│   └── sales.csv
│
├── notebooks/

├── reports/
│

│

```

### Workflow

1. Open the project folder in VS Code.
2. Create or activate the project's virtual environment.
3. Select the correct Python interpreter.
4. Install the required packages.
5. Create Python scripts or Jupyter Notebooks.
6. Select the correct notebook kernel.
7. Write, execute, and debug your code.
8. Verify results.
9. Save project dependencies in `requirements.txt`.
