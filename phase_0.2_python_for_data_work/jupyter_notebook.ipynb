### **Jupyter Notebook — Interview Questions**

- What is the difference between a Jupyter Notebook and a Python .py file?



- What is a kernel, and why does it maintain state?



- What is the difference between restarting a kernel and interrupting it?



- Why can executing notebook cells out of order produce incorrect results?



- What is notebook reproducibility, and how would you verify it?



- Challenge: A notebook works perfectly on your computer but fails on another person's computer. Give three possible reasons.

1. A Jupyter Notebook lets you run cells interactively and in different orders, while a `.py` file is usually run as a script from top to bottom.

2. A kernel is the process that runs notebook code. It keeps the session’s state in memory, including variables and imported modules. Execution counts show when cells were run; the kernel does not use them to determine which cells have run.

3. Restarting a kernel clears its in-memory state. Interrupting a kernel stops the code currently running but usually keeps the existing state.

4. Running cells out of order can cause errors or incorrect results because a cell may depend on variables or changes from another cell. For example, using a variable before running the cell that defines it can raise a `NameError`.

5. Notebook reproducibility means someone can run the notebook from a fresh kernel, with the required data and dependencies, and obtain the same results. Restart the kernel and use **Run All** to check this.

6. A notebook might work on one computer but fail on another because:
    - Required packages or a compatible version of Python are missing.
    - Data files or other resources are missing or stored at different paths.
    - The notebook relies on cells having been run out of order or on variables left in the kernel’s state.
