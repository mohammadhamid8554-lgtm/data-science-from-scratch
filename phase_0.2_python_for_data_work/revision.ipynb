# 1. DATETIME — WORKING WITH DATES AND TIME

## 1.1 Import datetime

```python
from datetime import date, datetime, time, timedelta
```

| Class | Purpose |
|---|---|
| `date` | Stores a calendar date |
| `datetime` | Stores date + time |
| `time` | Stores time only |
| `timedelta` | Represents a duration or time difference |

## 1.2 Get today's date

```python
today = date.today()

print(today)
print(today.year)
print(today.month)
print(today.day)
```

Remember:
- `date.today()` → today's date.
- `.year`, `.month`, `.day` → date components.

## 1.3 Get current date and time

```python
now = datetime.now()

print(now)
print(now.year)
print(now.month)
print(now.day)
print(now.hour)
print(now.minute)
print(now.second)
```

## 1.4 Convert a string into a date

```python
date_string = "25-12-2026"

converted_date = datetime.strptime(
    date_string, "%d-%m-%Y"
)

print(converted_date)
```

`strptime()` = **string → datetime**

Common format codes:

| Code | Meaning | Example |
|---|---|---|
| `%d` | Day | 25 |
| `%m` | Month | 12 |
| `%Y` | 4-digit year | 2026 |
| `%y` | 2-digit year | 26 |
| `%H` | Hour (24-hour) | 18 |
| `%M` | Minute | 30 |
| `%S` | Second | 45 |

## 1.5 Convert datetime into a string

```python
now = datetime.now()

formatted = now.strftime("%d-%m-%Y")

print(formatted)
```

`strftime()` = **datetime → string**

### Must Remember

- `strptime()` parses a string.
- `strftime()` formats a datetime.
- `date` stores only the date.
- `datetime` stores date and time.
- `timedelta` represents a difference or duration.

# 2. CSV — WORKING WITH TABULAR DATA

## 2.1 What is CSV?

CSV = Comma-Separated Values.

It stores data in rows and columns.

Example:

```text
name,marks
Aman,85
Riya,92
Rahul,76
```

- First row → Header
- Remaining rows → Records
- Each row → One record
- Each column → One field

## 2.2 Import CSV

```python
import csv
```

## 2.3 Read CSV using csv.reader()

```python
with open("data/students.csv", "r") as file:
    reader = csv.reader(file)

    header = next(reader)

    for row in reader:
        print(row)
```

`csv.reader()` returns rows as lists.

Example:

```python
["Aman", "85"]
```

## 2.4 Read CSV using DictReader()

```python
with open("data/students.csv", "r") as file:
    reader = csv.DictReader(file)

    for student in reader:
        print(student["name"])
        print(student["marks"])
```

`csv.DictReader()` returns each row as a dictionary.

Example:

```python
{
    "name": "Aman",
    "marks": "85"
}
```

### Important

CSV values are generally read as strings.

```python
marks = int(student["marks"])
```

Convert values before numerical calculations.

## 2.5 Calculate statistics from CSV

```python
marks_list = []

with open("data/students.csv", "r") as file:
    reader = csv.DictReader(file)

    for student in reader:
        marks_list.append(int(student["marks"]))

if marks_list:
    print(sum(marks_list))
    print(sum(marks_list) / len(marks_list))
    print(max(marks_list))
    print(min(marks_list))
```

**Always check for an empty dataset before calculating an average.**

## 2.6 Write CSV using csv.writer()

```python
with open("data/report.csv", "w", newline="") as file:
    writer = csv.writer(file)

    writer.writerow(["name", "marks"])
    writer.writerow(["Aman", 85])
    writer.writerow(["Riya", 92])
```

- `writerow()` → Writes one row.
- `writerows()` → Writes multiple rows.
- `newline=""` → Helps prevent extra blank rows.

## 2.7 Write CSV using DictWriter()

```python
with open("data/report.csv", "w", newline="") as file:
    fieldnames = ["name", "marks"]

    writer = csv.DictWriter(
        file, fieldnames=fieldnames
    )

    writer.writeheader()

    writer.writerow({
        "name": "Aman",
        "marks": 85
    })
```

- `fieldnames` → Defines column names and order.
- `writeheader()` → Writes column headers.
- `writerow()` → Writes one dictionary record.

### Must Remember

| Method | Purpose |
|---|---|
| `csv.reader()` | Reads rows as lists |
| `csv.DictReader()` | Reads rows as dictionaries |
| `csv.writer()` | Writes list-based rows |
| `csv.DictWriter()` | Writes dictionary-based rows |
| `next(reader)` | Reads the next row |
| `writer.writeheader()` | Writes CSV header |

# 3. JSON — WORKING WITH STRUCTURED DATA

## 3.1 What is JSON?

JSON = JavaScript Object Notation.

It stores structured data using objects and arrays.

Example:

```json
{
    "name": "Aman",
    "age": 20,
    "skills": ["Python", "SQL"]
}
```

JSON is commonly used for:
- Configuration files
- API responses
- Structured records
- Data exchange

## 3.2 Import JSON

```python
import json
```

## 3.3 Read JSON from a file

```python
with open("data/customer.json", "r") as file:
    customer = json.load(file)
```

`json.load()` = **JSON file → Python object**

## 3.4 Convert JSON string into Python object

```python
json_string = '{"name": "Aman", "age": 20}'

customer = json.loads(json_string)

print(customer["name"])
```

`json.loads()` = **JSON string → Python object**

## 3.5 Write JSON to a file

```python
customer = {
    "name": "Aman",
    "age": 20
}

with open("data/customer.json", "w") as file:
    json.dump(customer, file, indent=4)
```

`json.dump()` = **Python object → JSON file**

`indent=4` makes JSON easier to read.

## 3.6 Convert Python object into JSON string

```python
customer = {"name": "Aman", "age": 20}

json_string = json.dumps(customer, indent=4)
```

`json.dumps()` = **Python object → JSON string**

## 3.7 Nested JSON

```python
customer = {
    "name": "Aman",
    "address": {
        "city": "Mumbai",
        "pin": 400001
    }
}

print(customer["address"]["city"])
```

Use multiple keys to access nested dictionaries.

## 3.8 Handle invalid JSON

```python
try:
    with open("data/customer.json", "r") as file:
        customer = json.load(file)

except json.JSONDecodeError:
    print("Invalid JSON format")

except FileNotFoundError:
    print("File not found")
```

### Must Remember

| Function | Meaning |
|---|---|
| `json.load()` | Read JSON file |
| `json.loads()` | Parse JSON string |
| `json.dump()` | Write JSON file |
| `json.dumps()` | Create JSON string |

**Easy memory trick:**

- `load` → File
- `loads` → String
- `dump` → File
- `dumps` → String

# 4. BASIC FILE HANDLING

## 4.1 pathlib

```python
from pathlib import Path
```

`Path` provides an object-oriented way to work with files and folders.

## 4.2 Create a path

```python
data_folder = Path("data")
file_path = data_folder / "students.csv"
```

The `/` operator joins path components.

## 4.3 Check whether a path exists

| Method | Meaning |

- `exist_ok=True` → Don't raise an error if directory exists.

## 4.5 List directory contents

```python
folder = Path("data")

for item in folder.iterdir():
    print(item.name)
```

`iterdir()` → Immediate contents only.

## 4.6 Search files recursively

```python
folder = Path("data")

for file in folder.rglob("*.csv"):
    print(file)
```

`rglob()` searches inside subdirectories too.

| Method | Search |

| `iterdir()` | Immediate contents |
| `rglob()` | Recursive search |

## 4.7 Get file information

```python
file = Path("data/students.csv")

print(file.name)
print(file.suffix)
print(file.suffix.lower())
print(file.stat().st_size)
```

- `.name` → Filename.

- `.stat().st_size` → File size in bytes.

| Mode | Meaning |

| `"a"` | Append to existing content |

```python
with open("data/sales.txt", "r") as file:
    content = file.read()
```

# 5. REUSABLE ANALYSIS FUNCTIONS

## 5.1 Why reusable functions?

A reusable function allows you to write logic once and use it with different data.

Instead of repeating the same analysis, create one function.

## 5.2 Basic analysis function

```python
def analyze_marks(marks):
    if not marks:
        return None

    total = sum(marks)
    average = total / len(marks)
    highest = max(marks)
    lowest = min(marks)

    return total, average, highest, lowest
```

Use it:

```python
result = analyze_marks([80, 90, 70])

print(result)
```

## 5.3 Multiple return values

```python
total, average, highest, lowest = analyze_marks(
    [80, 90, 70]
)
```

Multiple values returned by a function are packaged into a tuple.

## 5.4 Validate data separately

```python
def validate_marks(marks):
    if not isinstance(marks, list):
        return False

    if not marks:
        return False

    if any(
        isinstance(mark, (bool, str))
        or not isinstance(mark, (int, float))
        for mark in marks
    ):
        return False

    if any(mark < 0 or mark > 100 for mark in marks):
        return False

    return True
```

## 5.5 Reuse validation

```python
def analyze_marks(marks):
    if not validate_marks(marks):
        return None

    total = sum(marks)
    average = total / len(marks)

    return total, average
```

This separates:
- Validation → Is the data acceptable?
- Analysis → What does the data tell us?

## 5.6 Count passing students

```python
def count_students(marks):
    return sum(
        1 for mark in marks if mark >= 35
    )
```

## 5.7 Calculate pass percentage

```python
def calculate_pass_percentage(marks):
    if not marks:
        return 0

    passed = count_students(marks)

    return passed / len(marks) * 100
```

## 5.8 Combine functions

```python
def analysis_report(marks):
    if not validate_marks(marks):
        return None

    total, average = analyze_marks(marks)

    passed = count_students(marks)

    pass_percentage = calculate_pass_percentage(marks)

    return {
        "total": total,
        "average": average,
        "passed_students": passed,
        "pass_percentage": pass_percentage
    }
```

### Must Remember

- Functions reduce repeated code.
- Separate validation from analysis.
- Use `return` to send results back.
- Use `print()` to display results.
- Reuse small functions inside larger functions.
- Empty data must be handled.
- Validate input before analysis.
- Clear function responsibilities improve maintainability.

A virtual environment is an isolated Python environment for a project.

Different projects can use different package versions without interfering with one another.

Example:

Each project can have its own environment.

# 7. JUPYTER NOTEBOOK

## 7.1 What is Jupyter Notebook?

Jupyter Notebook is an interactive environment where code is executed in separate cells.

It is useful for:
- Experimenting with data.
- Exploring results.
- Visualizing information.
- Documenting analysis.

## 7.2 Types of cells

| Cell | Purpose |
|---|---|
| Code | Execute Python |
| Markdown | Write notes, headings, explanations |

## 7.3 Kernel

A kernel is the process that executes notebook code.

It maintains session state, including:
- Variables
- Imported modules
- Objects created during execution

## 7.4 Cell execution

Useful shortcuts:

| Shortcut | Action |
|---|---|
| `Shift + Enter` | Run cell and move to next |
| `Ctrl + Enter` | Run cell and stay |
| `Alt + Enter` | Run cell and insert below |

## 7.5 Execution order

Notebook cells can execute out of order.

Example:

```python
# Cell 1
sales = [100, 200, 300]
```

```python
# Cell 2
total = sum(sales)
```

```python
# Cell 3
print(total)
```

Changing Cell 1 doesn't automatically recalculate `total`.

## 7.6 Kernel restart

Restarting the kernel clears in-memory variables.

After restarting, execute required cells again.

## 7.7 Restart + Run All

Use this to check whether your notebook works from a fresh kernel.

It helps identify:
- Missing variables
- Hidden dependencies
- Incorrect execution order
- Stale calculations

## 7.8 Reproducibility

A reproducible notebook can be executed from a fresh kernel using the required data and dependencies to obtain consistent results.

### Must Remember

- Code cells execute Python.
- Markdown cells document analysis.
- Kernel maintains state.
- Restart clears variables.
- Cell order doesn't guarantee execution order.
- Restart + Run All helps verify reproducibility.
- Notebooks should be organized and documented.

| Package | Purpose |

## 9.2 Module vs package vs library

- **Module:** Usually a single Python file.
- **Package:** Organized collection of related modules.
- **Library:** Reusable functionality designed for a particular purpose.

## 9.3 pip

`pip` is Python's package installer.

Install:

```bash
python -m pip install pandas
```

## 9.4 Package version

Install a specific version:

```bash
python -m pip install pandas==2.2.3
```

| Symbol | Meaning |
|---|---|
| `==` | Exact version |
| `>=` | Minimum version |
| `<=` | Maximum version |
| `!=` | Exclude version |

## 9.5 Version pinning

Version pinning means specifying a particular package version.

Example:

```text
pandas==2.2.3
numpy==2.1.3
```

It helps make projects more reproducible.

## 9.6 Important commands

| Command | Purpose |
