### **Lesson 1: What is a CSV file?**

Imagine a shop keeps a list of products. Instead of writing everything on paper, it stores the information in a file.

A CSV file stores data in rows and columns, like a table.

### **1. Understand the three important parts**

- Row: One product's information.
- Column: One type of information, such as price.
- Header: The first row containing column names.

````python
import csv

with open("data/new_students.csv", "r", newline = "") as file:
    reader = csv.reader(file)

    for row in reader:
        print(row)
````

### **csv.DictReader**

````python
import csv 

with open("data/new_students.csv", "r", newline = "") as file:
        reader = csv.DictReader(file)

        list_of_marks = []

        for student in reader:
            print(f"Student name: {student['name']} | Subject: {student['subject']} | Marks: {student['marks']}")
            list_of_marks.append(int(student['marks']))
        
        if len(list_of_marks) > 0:
            average_marks = sum(list_of_marks)/len(list_of_marks)
            print(f"Average marks: {average_marks:.2f}")
        
        else:
            print("Error: No student record found.")
````

### **Your program must:**
1. Read the input using csv.DictReader().
2. Check whether marks are valid integers between 0 and 100.
3. Assign Pass if marks are at least 35; otherwise assign Fail.
4. Write valid records into the output using csv.DictWriter().
5. Skip invalid records and continue processing the remaining students.

````python
with open("data/students.csv", "r", newline="") as input_file, \
    open("data/student_report.csv", "w", newline="") as output_file:

    fieldnames = ["name", "marks", "result"]
    reader = csv.DictReader(input_file)
    writer = csv.DictWriter(output_file, fieldnames=fieldnames)
    writer.writeheader()

    for student in reader:
       try:
          marks = int(student['marks'])
       except (ValueError, TypeError):
          print("Skipping record: marks must be a valid number.")
          continue

       if not 0 <= marks <= 100:
          print("Skipping record: marks must be between 0 and 100.")
          continue

       result = "Pass" if marks >= 35 else "Fail"

       writer.writerow({
          "name": student.get("name", ""),
          "marks": marks,
          "result": result
       })

print("Valid records written to data/student_report.csv.")
````

### **4. Must Remember — Data Scientist Cheat List**

- JSON stands for JavaScript Object Notation.
- JSON stores structured data.
- JSON objects use key-value pairs.
- JSON arrays store ordered collections of values.
- JSON keys and strings use double quotes.

### **Reading JSON files using json.load()**

- json.load reads JSON data from a file and converts it into Python objects.

````python
import json

with open("data/product.json", "r") as file:
    product = json.load(file)

print(product)
````

### **Notice:**

- JSON object → Python dictionary
- JSON array → Python list
- JSON true → Python True
- JSON false → Python False
- JSON null → Python None
This conversion is called deserialization.

### **Writing JSON files using json.dumps()**

- json.dump() → Python object to JSON file.
- json.load() → JSON file to Python object.
- json.dumps() → Python object to JSON string.

````python
import json

customer_data = {
    "name": "Rahul",
    "age": 22
}

json_string = json.dumps(customer_data, indent=4)

print(json_string)
print(type(json_string))

print("\njson.dumps() returns a Python string. It does not create a file.")
````
