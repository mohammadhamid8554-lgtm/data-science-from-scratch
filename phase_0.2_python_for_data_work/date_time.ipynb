## Topic 1: Python `datetime` — Getting Today's Date

### 1. Why do we need `datetime`?
The `datetime` module helps us work with dates and times. In data science, it is useful for analyzing sales dates, tracking orders, and preparing time-based reports.

### 2. Import the `date` class
```python
from datetime import date
```
- `datetime` is a Python module.
- `date` is a class inside that module.
- `from ... import ...` lets us import a specific class.

### 3. Get today's date
```python
today = date.today()
print(today)
```
Example output: `2026-10-28`

**Remember:** `date.today()` returns the current date when the code runs.

### 4. Extract date components
```python
print(today.year)
print(today.month)
print(today.day)
```
- `.year` → year
- `.month` → month (1–12)
- `.day` → day of the month (1–31)

These are **attributes**, not methods. Use `today.year`, not `today.year()`.

### 5. Format the date using an f-string
```python
print(f"Today is {today.day}/{today.month}/{today.year}")
```
Example output: `Today is 28/10/2026`

Note: This produces `D/M/YYYY` without leading zeros. We will learn dedicated date formatting later.

### 6. Key terms
- **Module:** A file or library containing Python code.
- **Class:** A blueprint for creating objects.
- **Attribute:** A value associated with an object.
- **Method:** A function associated with an object.
- **Object:** A value created or returned by Python code.

### 7. Common mistakes
| Incorrect | Correct |
|---|---|
| `date.today` | `date.today()` |
| `today.year()` | `today.year` |
| `from date import datetime` | `from datetime import date` |

### 8. Data scientist's workflow
1. Import the required tool.
2. Get the date.
3. Extract the required components.
4. Format the output for the report.

**Core takeaway:** `date.today()` gets today's date, while `.year`, `.month`, and `.day` access its components.

### **Lesson 1: Date and time handling**

````python
from datetime import date

today = date.today()
print(f"Today's date: {today}")
print(f"Year: {today.year}")
print(f"Month: {today.month}")
print(f"Day: {today.day}")
print(f"Formatted Date: {today.day:02d}/{today.month:02d}/{today.year}")
# Here, :02d means display an integer using at least two digits, adding a leading zero when necessary.
````

### **Lesson 2: Converting text into dates**

````python
from datetime import datetime


date_text = "15-03-2025"

converted_date = datetime.strptime(date_text, "%d-%m-%Y")
print(f"Converted Date: {converted_date}")
print(f"Year: {converted_date.year}")
print(f"Month: {converted_date.month}")
````

### **Lesson 3: Calculating the difference between dates**

### **1. The problem**
Imagine a company wants to know how many days it took to deliver an order.
```
   - Order date: 1 March 2025
   - Delivery date: 8 March 2025
```
Instead of counting days manually, Python can calculate the difference.
2. The concept
Subtracting one date object from another produces a timedelta object.

Think of timedelta as a container that tells you the amount of time between two dates.

### **Example** 

`Dilivery date(8 March 2025) - Order date(1 March 2025) = 7 days `

````python
from datetime import date

order_date = date(2025, 3, 1)
delivery_date = date(2025, 3, 8)

difference = delivery_date - order_date
print(f"Difference: {difference.days}")
print(f"Remaining hours: {difference.seconds//3600}")
````

### **Understand these three parts:**

- date(2025, 3, 1) creates a date using year, month, and day.
- delivery_date - order_date calculates the difference.
- .days extracts the number of days from the resulting timedelta

### **Lesson 5: Formatting dates with strftime()**


1. The problem
Suppose Python gives you this date:

2025-03-15 00:00:00


But your report needs this format:

15/03/2025


You could build the string manually with f-strings, as you've already learned. But Python also provides strftime() for formatting dates and times.

2. The mental model

Think of strftime() as a date formatter. python date object(2025-03-15) -> strftime("%d/%m/%Y") = formatted text(15/03/2025)

### **Example**

````python
from datetime import date, datetime

today = date(2025, 3, 5)

formatted_date1 = today.strftime("%d/%m/%Y")
formatted_date2 = today.strftime("%d/%m/%y")
formatted_date3 = today.strftime("%b/%m/%y")
formatted_date4 = today.strftime("%B/%m/%y")
print(formatted_date1)
print(formatted_date2)
print(formatted_date3)
print(formatted_date4)
````

```
strptime() converts text into a datetime object.

strftime() converts a date/time object into text.

```

### **Lesson 6: Current date and time with datetime.now()**

### **1. The problem**

Imagine you have a sales system that needs to record exactly when a transaction happens.

You could manually enter the date and time, but that would be unreliable. Python can get the current date and time from your computer's clock.

### **2. The basic idea**

````python
from datetime import  datetime as dt

current_time = dt.now()

print(current_time)
print(current_time.time())
print(current_time.date())
````

What each part does:
- datetime.now() gets the current local date and time.
- .date() extracts the date portion.
- .time() extracts the time portion.

````python
from datetime import datetime

deadline = datetime(2026, 12, 31, 23, 59)

if current_time < deadline:
    print("Deadline has not passed.")

else:
    print("Deadline has passed.")
````

### **Lesson 7: Calculating time remaining with timedelta**

### **1. The problem**

Imagine a customer has a payment deadline on 5 November 2026. Your program needs to calculate how many days remain until that deadline.

Instead of counting manually, Python can subtract the current date from the deadline.

### **2. The concept**

When you subtract two datetime objects, Python returns a timedelta object.

````python
from datetime import datetime as dt

current_time = dt.now()

deadline = dt(2026, 11, 5, 12)

time_remaining = deadline - current_time

print(f"Remaining time: {time_remaining.days}")
````

## **Final Test: Payment Deadline Checker**

Scenario: A business stores payment deadlines as text. Your job is to convert each deadline into a date, validate it, and determine whether the payment deadline has passed.

### **Requirements**

Write a Python program that:

1. Imports datetime.
2. Takes a deadline from the user in the format DD-MM-YYYY.
3. Converts the text into a datetime object using strptime().
4. Handles invalid dates or incorrect formats using try/except ValueError.
5. Gets the current date and time using datetime.now().
6. Compares the deadline with the current date and time.
7. Prints one of these results:
   - Payment deadline has passed.
   - Payment deadline has not passed.
8. If the deadline has not passed, also print the remaining whole days.

````python
from datetime import datetime

try: 
    # 1. Take the deadline date from the user as an input.
    
    deadline = input("Enter deadline date(DD-MM-YYYY): ")
    converted_deadline = datetime.strptime(deadline, "%d-%m-%Y")
    current_date_time = datetime.now()

    # Verify whether the deadline has passed or not.
    # If not, calculate remaining days and hours else display deadline date has passed.

    if current_date_time.date() < converted_deadline.date():
        print("Payment deadline has not passed.")
        time_remaining = converted_deadline.date() - current_date_time.date()
        print(f"Remaining days: {time_remaining.days}")

    else:
        print("Payment deadline has passed.")


except ValueError:
    print("Error: Invalid date or format.")
````
