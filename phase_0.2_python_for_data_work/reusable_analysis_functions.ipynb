### **2. What is a reusable analysis function?**

A reusable function is a block of code that performs a specific task and can be used repeatedly with different data.

- A function should perform a clear task.



- Parameters allow different inputs.



- return sends results back to the caller.



- Reuse the same function with different datasets.



- Avoid repeating the same analysis logic.

```python
def analyze_sales(sales):
    total = sum(sales)
    average = total / len(sales)
    highest = max(sales)
    lowest = min(sales)

    return total, average, highest, lowest

sales = [1200, 420, 500, 350]

result = analyze_sales(sales)
print(f"Total sales: {result[0]}\nAverage sales: {result[1]:.2f}\nHighest sale: {result[2]}\nLowest sale: {result[3]}")

new_sales = [4000, 2000, 3500]

result = analyze_sales(new_sales)
print("\nNew Sales Analysis")
print(f"Total sales: {result[0]}\nAverage sales: {result[1]:.2f}\nHighest sale: {result[2]}\nLowest sale: {result[3]}")
```

### **5. Your first exercise**
Create a function named analyze_marks().

It should:
1. Accept a list of student marks.
2. Calculate the total marks.
3. Calculate the average marks.
4. Find the highest marks.
5. Find the lowest marks.
6. Return all four results.

### **Validate the data**

```python
def validate_marks(student_marks:list):

    if not isinstance(student_marks, list):
        return False
    
    if not student_marks:
        return False

    if any(isinstance(mark, (bool, str)) or not isinstance(mark, (int, float)) for mark in student_marks):
        return False

    if any(mark < 0 or mark > 100 for mark in student_marks):
        return False

    return True

print(validate_marks([80, 90, 75]))
print(validate_marks([]))
print(validate_marks([80, "90", 75]))
print(validate_marks([True, 90]))
print(validate_marks([-5, 80, 90]))
print(validate_marks([0, 50, 100]))
```

### **Calculate summary statistics**

```python
def analyze_marks(student_marks:list):

    validation = validate_marks(student_marks)
    if not validation:
            return None
    
    total_marks = sum(student_marks)
    average_marks = total_marks/len(student_marks)
    highest_marks = max(student_marks)
    lowest_marks = min(student_marks)
    return total_marks, average_marks, highest_marks, lowest_marks


marks = [78, 85, 92, 67,0, 100]

result = analyze_marks(marks)
if result is None:
    print("Cannot analyze invalid marks.")
else:
    total, average, highest, lowest = result
    print(f"Total marks: {total}")
    print(f"Average marks: {average:.2f}")
    print(f"Highest marks: {highest}")
    print(f"Lowest marks: {lowest}")
```

### **Calculate passed students**

```python
def count_students(student_marks):
    validation = validate_marks(student_marks)

    if not validation:
        return None

    count = sum(1 for mark in student_marks if mark >= 35)
    return count
```

### **Calculate passed student's percentage**


```python
def calculate_pass_percentage(student_marks):

    passed_students = count_students(student_marks)
    
    if passed_students is None:
        return None

    percentage = round(passed_students / len(student_marks) * 100, 2)

    return percentage

marks = [80, 25, 65, 30, 90, 40]
result = calculate_pass_percentage(marks)
print(result)
```

### **Create analysis report**

```python
def analysis_report(student_marks:list):
    validation = validate_marks(student_marks)

    if not validation:
        return None

    summary_statistics = analyze_marks(student_marks)
    total, average, highest, lowest = summary_statistics
   


    passed_students = count_students(student_marks)
    percentage =  calculate_pass_percentage(student_marks)

    return{
        "total": total,
        "average": average,
        "highest": highest,
        "lowest": lowest,
        "passed_students": passed_students,
        "pass_percentage": percentage
    }

marks = [80, 25, 65, 30, 90, 40]

report = analysis_report(marks)
print(report)
```

### **Interview-style questions**

1. Fundamental: What is a reusable function, and why is it useful in data analysis?

    **Answer:** A reusable function performs a specific task and can be used with different data for similar tasks. Instead of repeating the same logic, we can write a function once and use it with different datasets. For example, we may create a function to find a summary of last month’s sales. If we want to calculate the same summary for this month’s sales, we can use the function we already created instead of writing the same logic again. This is how a reusable function helps with repetitive tasks.

2. Fundamental: What is the difference between `return` and `print()` inside a function?

    Answer: The major difference between print() and `return is that print() only displays the output, while `return` gives us a value that can be used in further operations if needed.

3. Practical: Why should we validate data before calculating its average?

    **Answer:** Real data is not always clean and ready to use. Before working with data, we need to validate edge cases, including the following: 1. Check the data type. 2. Check business-related edge cases. 3. Validate user input. For example, if a user provides a list of numbers as strings, calculating the average will raise a `TypeError`. That is why we need to validate our data before working with it.

4. Practical: What is the advantage of returning multiple values from `analyze_marks()`?

    **Answer:** Returning multiple values helps us understand our data clearly. In our previous program, we calculated the total, average, highest, and lowest marks. This information can help us analyze student marks, such as identifying how many students received high or low marks. After the analysis, we can make recommendations to students who did not receive good marks and motivate them to do their best next time.

5. Challenging: Why is calling `count_students()` inside `calculate_pass_percentage()` better than repeating the counting logic?

    **Answer:** Calling `count_students()` inside `calculate_pass_percentage()` helps avoid repeating the same logic. We have already written the validation and counting operations inside `count_students()`, so repeating that logic in another function would be unnecessary. We are learning about reusable functions to avoid writing code again and again. We can call the function inside the percentage function instead of repeating the same logic.

6. Challenging: Suppose `analysis_report()` calls four different functions, and each function independently validates the same dataset. What inefficiency could this create? How might you improve the design?

    **Answer:** It could make the function less efficient because the same logic is repeated in every function. Instead, we can validate the data once inside the analysis report function and get the other calculations from functions such as `analyze_marks()`, the percentage function, and the counting function. This could improve the function’s efficiency, which is important when working on production-level projects.

7. Real-world: If you receive 100,000 student records, why might reusable analysis functions make your code easier to maintain?

    **Answer:** Since we have created different functions for different tasks, whenever we need to analyze the data we receive, we can call the desired functions and get the results without much effort. For example, if we want a statistical summary of the data, we can use our analysis function. Similarly, we can use other functions. The validation function is especially important because we cannot analyze the data if it is not appropriate for our analysis goals.
