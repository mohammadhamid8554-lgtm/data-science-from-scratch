```python
from pathlib import Path

folder_path = Path('data')

count = 0
if folder_path.is_dir():
    for item in folder_path.iterdir():
        if item.is_file() and item.suffix.lower() == ".csv":
            print(f"File: {item.name}")
            count += 1
    print(f"Total csv files: {count}")
else:
    print("Directory not found.")
```

### **Next exercise: Combine your skills**
Write a program that:
1. Checks whether the data directory exists.
2. Creates a directory named reports if it doesn't exist.
3. Finds all CSV files inside data.
4. Prints each CSV filename and its file size in bytes.
5. Counts and prints the total number of CSV files.

```python
from pathlib import Path

folder_path = Path("data")


count = 0
if folder_path.is_dir():
    new_folder_path = Path("reports")
    new_folder_path.mkdir(exist_ok=True)
    summary = []
    for item in folder_path.iterdir():
        if item.is_file() and item.suffix.lower() == ".csv":
            count += 1
            size = item.stat().st_size
            summary.append((item.name, size))
            print(f"File name: {item.name} | File size: {item.stat().st_size} bytes")
    print(f"Total csv files: {count}")

    with open("reports/csv_summary.txt", "w") as file:
        file.write("CSV File Summary\n")
        file.write(f"Total CSV Files: {count}\n")

        for filename, size in summary:
            file.write(f"{filename}: {size} bytes\n")
else: 
    print("Directory not found.")
```

- stat() gets information about a file.



- .st_size gives its size in bytes.

### **Final practical challenge: File handling**
Before we mark Basic File Handling complete, solve this scenario:

You have a folder named data containing CSV files.

Build a program that:
1. Checks whether data exists.
2. Creates reports if necessary.
3. Finds all CSV files.
4. Collects each filename and file size.
5. Creates reports/csv_summary.txt.
6. Writes the report without crashing if there are no CSV files.
7. Handles a missing data directory clearly.

```python
from pathlib import Path

folder_path = Path("data")

summary = []

if folder_path.is_dir():

    for item in folder_path.iterdir():

        if item.is_file() and item.suffix.lower() == ".csv":

            size = item.stat().st_size
            summary.append((item.name, size))
    reports_path = Path("reports")
    reports_path.mkdir(parents=True, exist_ok=True)

    try:

        with open("reports/csv_summary.txt", "w") as file:

            file.write("CSV File Summary\n")
            file.write(f"Total CSV Files: {len(summary)}\n")

            for filename, size in summary:
                file.write(f"{filename}: {size} bytes\n")
                
    except FileNotFoundError:
        print("Error: Required file or directory not found.")

    except PermissionError:
        print("Error: Permission denied.")

else:
    print("Directory not found.")
```

- iterdir()	Returns immediate files and folders only

- rglob("*.csv")	Searches recursively for CSV files in the directory and its subdirectories

### **Final practical exercise**
Write a program that:
1. Checks whether data exists.
2. Searches for all CSV files, including those inside nested folders.
3. Prints each CSV file's name and size in bytes.
4. Counts the total number of CSV files.

```python
from pathlib import Path

folder_path = Path("data")

count = 0
if folder_path.is_dir():
    for item in folder_path.rglob("*.csv"):
        if item.is_file():
            size = item.stat().st_size
            print(f"File name: {item.name} | File size: {size}")
            count += 1
        elif count == 0:
            print("Error: no CSV files found.")

    print(f"Total CSV Files: {count}")
else:
    print("Directory not found")
```
