### **Business scenario**
Imagine a company receives a CSV file containing sales transactions.

Unfortunately, the file contains errors:
- Some customer names are missing.
- Some quantities are invalid.
- Some prices are negative.
- Some transaction IDs are duplicated.
The company cannot trust this data until it is checked and cleaned.

Your job: Build a Python program that checks the data, separates valid and invalid records, and generates a summary report.

```python
import csv
# Part-1
def validate_record(transaction_records):
    try:

         # Check required fields and buisness rule.
        if not transaction_records["transaction_id"]:
            return False, "Error: Missing transaction Id."
                
        if not transaction_records['customer_name']:
            return False, "Error: Missing customer name."
        
        if not transaction_records["product"]:
             return False, "Error: Missing product."
        # Convert CSV Strings into numbers.
        quantity = int(transaction_records["quantity"])

        if  quantity <= 0:
                    return False, "Error: Quantity must be greater than zero."
           
        unit_price = float(transaction_records["unit_price"])

        if  unit_price < 0:
                        return False, "Error: Unit price cannot be negative."
        
        return True, None
        
    except ValueError:
        return False, "Unit Pirce and quantity must be a valid number."

    except TypeError:
         return False, "Cannot convert unit price and quantity."

# Part-2
try:

    # Tas 1: Read and Inspect the Dataset
    with open("data/sales_raw.csv", "r", newline="") as file:
        records = csv.DictReader(file)
        header = records.fieldnames
        list_of_records = list(records)

        valid_record = []
        invalid_record = []
        unique_ids = set()
        duplicate_ids = []

        for record in list_of_records:
            is_valid, reason = validate_record(record)
            transaction_ids = record['transaction_id']

            if is_valid and transaction_ids not in unique_ids:
                valid_record.append(record)
                unique_ids.add(transaction_ids)

            elif transaction_ids in unique_ids:
                invalid_record.append(record)
                reason = "Duplicate transaction ID"
                record['error_reason'] = reason
                

            else:
                duplicate_ids.append(transaction_ids)
                record['error_reason'] = reason
                invalid_record.append(record)

        total_revenue = 0
        for record in valid_record:
            quantity = int(record['quantity'])
            price = float(record['unit_price'])

            total_revenue += price * quantity

        print(f"Total Revenue: {total_revenue:.2f}")
        print(f"Tota Records: {len(list_of_records)}")
        print(f"Valid records: {len(valid_record)}")
        print(f"Invalid records: {len(invalid_record)}")
        print(f"Unique IDs: {unique_ids}")
        print(f"Duplicate IDs: {duplicate_ids}")
           
except FileNotFoundError:
    print("File not found")
```
