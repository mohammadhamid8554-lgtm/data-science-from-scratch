### **Why do we need this topic?**

```python
import numpy as np
sales = np.array([1000, 2000, np.nan, 4000, 5000])
```

One value is missing.

In real Data Science datasets, missing values happen because:
- A customer didn't provide information.
- A sensor failed to record a measurement.
- A data entry mistake occurred.
- A system didn't collect a value.

```python
print(np.mean(sales))
```

`Problem: One missing value can make the entire average become NaN.`

### **What is NaN?**

```python
arr = np.array([10, 20, np.nan, 40])
print(arr)
```

`Notice that the array becomes floating-point because NaN is a floating-point value.`

### **Detecting missing values — np.isnan()**

```python
arr = np.array([10, 20, np.nan, 40, np.nan])
np.isnan(arr)
```

True → element is NaN.



False → element is not NaN.

### **Count missing values**

```python
np.sum(np.isnan(arr))
```

This tells us there are two missing numerical values.

### **Calculating averages with missing values — np.nanmean()**

```python
np.mean(arr)
```

```python
round(np.nanmean(arr), 2)
```

`np.nanmean()` ignores NaN values when calculating the average.

### **Replacing NaN — np.nan_to_num()**
Sometimes you want to replace missing numerical values with another number

```python
arr = np.array([10, 20, np.nan, 40])
np.nan_to_num(arr, nan=0)
np.nan_to_num(arr, nan=100)
```

### **Other special values — inf and -inf**

```python
arr = np.array([10, np.inf, -np.inf])
```

`np.inf` → positive infinity.



`-np.inf` → negative infinity.

```python
np.isinf(arr)
```

Detect all non-finite values:

```python
np.isfinite(arr)
```

`np.isfinite()` returns `False` for NaN and both positive and negative infinity.
