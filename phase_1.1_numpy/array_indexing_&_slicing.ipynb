### **1. What is indexing?**

`Means Access a single emelement from a sequence like list or array or string.`

```python
import numpy as np

sales = np.array([100, 200, 300, 400, 500])
print(sales[0])
print(sales[4])
print(sales[-5])
print(sales[-1])
```

### **What is Slicing?**

`Means to extract a portion of a sequence like list or string or array.`

```python
import numpy as np

sales = np.array([100, 200, 300, 400, 500])

print(sales[:])
print(sales[:3])
print(sales[::2])
print(sales[::-1])
```

### **Indexing and Slicing in 2D array**

```python
employees = np.array([
    [101, 25000, 2],
    [102, 30000, 3],
    [103, 35000, 4],
    [104, 40000, 5]
])

print(employees[0, 0])
print(employees[-1, -1])
print(employees[1, 2])
print(employees[0, 1])
print(employees[2, 2])
print(employees[0])
print(employees[:, 1])
```

### **One important Data Science concept: Views**
`NumPy slicing usually returns a view, not an independent copy.`


```python
prices = np.array([100, 200, 300, 400])

selected = prices[1:3]
selected = prices[1:3].copy()
```
