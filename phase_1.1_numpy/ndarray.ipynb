### **1. What is an Array**

```python
import numpy as np

marks = [75, 80, 65, 90, 85] # This is a Python list.

marks = np.array(marks)   # This is Numpy array.
```

`Array: Structured container for numerical values.`

### **2. What is `ndarry`?**

- ndarray stands for N-dimensional array.
- N = any number of dimensions.
- array = a collection of values arranged in a structure.
- The NumPy array object is called ndarry.

```python
marks = np.array([75, 80, 65, 90, 85]) # Numpy creates an ndarray object.
```

### **3. Understanding Array Dimensions**

### **A. One-dimensional array(1D)**

```python
marks = np.array([75, 80, 65, 90, 85]) # It has one direction of data.
```

### **B. Two-dimensional array(2D)**

```python
marks = np.array([
    [80, 75, 90],
    [60, 85, 70],
    [95, 0, 85]
])

# A 2D array has two dimensions: Rows and Columns

```

### **C. Three-dimennsional array(3D)**

`3D array =  Multiple tables stacked together.`

### **4. The Most Important Difference: Dimension vs Number of Values**

### **1. 1D array**

```python
a = np.array([10, 20, 30, 40])
```

- Dimensions: 1
- Number of values: 4

### **2. 2D array**

```python
b = np.array([
    [10, 20, 30],
    [40, 50, 60]
])
```

- Dimensions: 2
- Number of values: 6

### **3. 3D array**

```python
c = np.array([
    [[1, 2], [3, 4]],
    [[5, 6], [7, 8]]
])
```

- Dimensions: 3
- Number of values: 8

### **Remember:**
- Dimension tells you how many directions the data is arranged in.
- Number of values tells you how many individual elements exist

### **4. How to Check Array Dimensions**

```python
a = np.array([10, 20, 30])

print(f"Dimensions of a: {a.ndim}")
```

```python
b = np.array([
    [10, 20],
    [30, 40]
])

print(f"Dimensions of b: {b.ndim}") # ndim is an attribute, not a function.
```
