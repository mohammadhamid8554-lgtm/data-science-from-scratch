### **Broadcasting**
`Performing operations between two different shapes of arrays.`

### **Broadcating with scalar**

```python
import numpy as np

sales = np.array([100, 200, 300, 400])

charge = 50

print(sales + charge)
```

```python
sales = np.array([[200, 300],[400, 500]])
charge = np.array([20, 30])

new_sales = sales + charge     # Compitable Shape: (2, 2)
print(new_sales)
```

```python
sales = np.array([[200, 300],
                 [400, 500]])
charge = np.array([20, 30, 40])

new_sales = sales + charge         # Incompitable
print(new_sales)
```

```python
import numpy as np

prices = np.array([
    [100, 200, 300],
    [400, 500, 600]
])

tax = np.array([10, 20, 30])

final_prices = prices + tax

print(final_prices)
print(final_prices.shape)
```

### 1D array across 2D rows
data = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

adjustment = np.array([1, 2, 3])

data + adjustment

 Broadcasting rules:
 1. Compare shapes from right to left.
 2. Dimensions must be equal or one must be 1.
 3. Missing dimensions are treated as 1.
 4. Incompatible shapes produce ValueError.
