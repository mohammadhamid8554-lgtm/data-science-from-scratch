Why is understanding array dimensions and shape important in Data Science?

1. ndarray stands for n-dimensional arrary. n can be any like 1,2, and 3. ndarray is a structured container that stores values efficiently. This is the main data structure in NumPy.

2. ndim gives the dimensions of an array while shape gives the number of rows and columns along each dimensions.

3. Shape gives us how many rows and columns a dataset contains along each dimension while, size gives total number of elements of an array.

4. Data type tells us what kind of data is an array containing. Data type matters most in data science before perfoming any calculations.

5. 4 represents 4 different groups, 3 represents each group contains 3 rows and 2 represents that each group contains 2 columns in each.

6. Understanding array dimensions helps us to perform operations like array boardcasting and helps in the process of training and testing concept in machine learning. If we know dimensions and shape of an array, it will help us in array manupilation and reshaping.
