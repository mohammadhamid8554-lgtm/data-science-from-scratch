#### **Why would a Data Scientist create an array filled with zeros instead of manually writing zero values?**

A Data Scientist uses np.zeros() because:
1. Efficiency: Creates large arrays without manually writing values.
2. Initialization: Provides a starting structure for storing or updating numerical values.
3. Convenience: Easily creates arrays of different dimensions.
4. Performance: Avoids manually constructing large collections of zero values in Python.

### **5. Creatig array using np.linspace()**

Imagine you need exactly 5 evenly spaced numbers between 0 and 100.

With np.arange(), you choose the step size. But what if you want to choose the number of values instead?

That's what np.linspace() does.

```python
numbers = np.linspace(0, 100, 5)
print(numbers)
```

### **8. Creating array using np.asarray()**
`np.asarray() converts array-like data into a NumPy array.`

```python
prices = [100, 200, 300, 400]

price_array = np.asarray(prices)

print(price_array)
```

```python
marks = (75, 80, 90, 85)

marks_array = np.asarray(marks)

print(marks_array)
```

```python
data = [[10, 20], [30, 40]]

array = np.asarray(data)

print(array)
```
