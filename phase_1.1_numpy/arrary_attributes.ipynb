```python
import numpy as np

employees = np.array([
    [101, 25000, 2],
    [102, 30000, 3],
    [103, 35000, 4],
    [104, 40000, 5],
    [105, 45000, 6]
])
```

### **Your task:**

Write code to:
1. Print the number of dimensions.
2. Print the shape.
3. Print the total number of elements.
4. Print the data type.
5. Calculate how many employees are in the dataset using array attributes (don't manually count rows).
6. Calculate how many columns the dataset contains using array attributes.

```python
print(f"Dimensions: {employees.ndim}")
print(f"Shape: {employees.shape}")
print(f"Total number of elements: {employees.size}")
print(f"Data type: {employees.dtype}")

rows, cols = employees.shape
print(f"Total number of employees: {rows}")
print(f"Total number of columns: {cols}")
```
