## 1. Introduction to NumPy

**NumPy = Numerical Python**

NumPy is a Python library used for fast numerical calculations and working with numerical datasets.

```python
import numpy as np
```

### Why NumPy?

- Faster numerical calculations than many Python-level loops.
- Performs operations on entire arrays.
- Supports multidimensional numerical data.
- Provides mathematical and statistical functions.
- Used heavily in Pandas, Machine Learning, and scientific computing.

### Python List vs NumPy Array

```python
numbers = [10, 20, 30]
numbers * 2
# [10, 20, 30, 10, 20, 30]

arr = np.array([10, 20, 30])
arr * 2
# [20, 40, 60]
```

**Vectorization:** Applying operations to entire arrays instead of manually looping through individual elements.

## 3. Array Creation

```python
np.array([1, 2, 3])
```

| Function | Purpose |
|---|---|
| `np.zeros((2,3))` | Creates zeros |
| `np.ones((2,3))` | Creates ones |
| `np.full((2,3), 7)` | Fills with chosen value |
| `np.arange(1,10,2)` | Creates sequence with step |
| `np.linspace(0,10,5)` | Creates fixed number of evenly spaced values |
| `np.empty((2,3))` | Allocates uninitialized values |
| `np.eye(3)` | Creates identity matrix |
| `np.asarray(data)` | Converts compatible data into an array |

### Important Differences

```python
np.arange(1, 10, 2)
# [1 3 5 7 9]

np.linspace(0, 10, 5)
# [0.  2.5  5.  7.5  10.]
```

- `arange()` → controls step size.
- `linspace()` → controls number of values.
- `empty()` → values are uninitialized; assign before using.
- `eye()` → diagonal contains ones, other elements zero.

## 5. Indexing and Slicing

### 1D Indexing

```python
arr = np.array([10, 20, 30, 40, 50])

arr[0]       # First element
arr[-1]      # Last element
arr[1:4]     # Index 1 to 3
arr[::2]     # Every second element
arr[::-1]    # Reverse array
```

### 2D Indexing

```python
data = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

data[0]        # First row
data[1, 2]     # Row 1, column 2
data[:, 1]     # Entire second column
data[1:, :]    # Rows from index 1 onward
data[:, 0:2]  # First two columns
```

### View vs Copy

```python
view = arr[1:4]
copy = arr[1:4].copy()
```

- Slicing often creates a view.
- A view may share memory with the original.
- `.copy()` creates an independent copy.

## 7. Broadcasting

Broadcasting allows NumPy to perform operations between arrays of compatible shapes.

### Example

```python
data = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

bonus = np.array([1, 2, 3])

data + bonus
```

Result:

```python
[[11, 22, 33],
 [41, 52, 63]]
```

### Broadcasting Rules

Compare dimensions from right to left.

Two dimensions are compatible when:

- They are equal, OR
- One of them is 1.

Missing dimensions are treated as 1.

Example:

```python
(2, 3) + (3,)      # Compatible
(5, 1) + (1, 3)    # Compatible → (5, 3)
(2, 3) + (2,)       # Not compatible
```

## 8. Aggregation Functions

Aggregation functions summarize numerical data.

```python
np.sum(arr)
np.mean(arr)
np.min(arr)
np.max(arr)
np.median(arr)
np.std(arr)
np.var(arr)
```

### Index of Extremes

```python
np.argmax(arr)   # Index of maximum
np.argmin(arr)  # Index of minimum
```

### Percentiles

```python
np.percentile(arr, 25)
np.percentile(arr, 50)
np.percentile(arr, 75)
```

- 25th percentile → approximately 25% of values are at or below it.
- 50th percentile → median.
- 75th percentile → approximately 75% of values are at or below it.

### Statistical Relationships

- Mean → average.
- Median → middle value; often robust to outliers.
- Variance → average squared deviation from mean.
- Standard deviation → square root of variance.

NumPy uses `ddof=0` by default for variance and standard deviation.

## 9. Axis — Rows and Columns

```python
sales = np.array([
    [100, 200, 300],
    [400, 500, 600]
])
```

### `axis=0`

```python
sales.sum(axis=0)
# [500 700 900]
```

Calculates down rows → one result per column.

### `axis=1`

```python
sales.sum(axis=1)
# [600 1500]
```

Calculates across columns → one result per row.

| Operation | Meaning |
|---|---|
| `axis=0` | Column-wise results |
| `axis=1` | Row-wise results |

**Remember:** The axis you reduce disappears from the output shape.

## 10. Boolean Indexing and Filtering

### Conditions

```python
arr > 50
arr < 50
arr >= 50
arr <= 50
arr == 50
arr != 50
```

### Filtering

```python
arr[arr > 50]
```

### Multiple Conditions

```python
arr[(arr > 20) & (arr < 80)]
arr[(arr < 20) | (arr > 80)]
arr[~(arr > 50)]
```

### Operators

- `&` → AND
- `|` → OR
- `~` → NOT

Use parentheses around each condition.

### Count Matching Values

```python
np.sum(arr > 50)
```

### Conditional Modification

```python
arr[arr < 50] += 10
```

**Remember:** NumPy arrays use `&`, `|`, and `~` for element-wise Boolean operations, not Python's `and` or `or`.

- Total element count must remain the same.
- `-1` allows NumPy to calculate one dimension.

### Flatten

```python
arr.flatten()
```

Converts array to 1D and always creates a copy.

### Reshape to 1D

```python
arr.reshape(-1)
```

## 12. Handling Missing and Special Values

### NaN

```python
np.nan
```

Represents a missing or undefined numerical value.

### Detect Missing Values

```python
np.isnan(arr)
np.sum(np.isnan(arr))
```

Never use:

```python
arr == np.nan
```

Use:

```python
np.isnan(arr)
```

### NaN-Aware Aggregations

```python
np.nanmean(arr)
np.nansum(arr)
np.nanmin(arr)
np.nanmax(arr)
np.nanmedian(arr)
```

These ignore NaN values.

### Replace NaN

```python
np.nan_to_num(arr, nan=0)
```

Replaces NaN with zero by default in this example.

### Infinity

```python
np.inf
-np.inf
```

### Detect Special Values

```python
np.isinf(arr)
np.isfinite(arr)
```

- `isinf()` → detects positive and negative infinity.
- `isfinite()` → True for finite numbers; False for NaN and infinity.

**Important:** Missing values should not automatically become zero. Investigate their meaning before deciding how to handle them.

## 13. Random Number Generation

### Create Generator

```python
rng = np.random.default_rng()
```

### Reproducible Generator

```python
rng = np.random.default_rng(seed=42)
```

Same seed and same sequence of operations → reproducible random sequence.

### Random Decimals

```python
rng.random(5)
```

Generates values from 0 inclusive to 1 exclusive.

### Random Integers

```python
rng.integers(1, 11, size=5)
```

Generates integers from 1 to 10 inclusive.

### Normal Distribution

```python
rng.normal(loc=50, scale=10, size=5)
```

- `loc` → mean.
- `scale` → standard deviation.
- Values are not restricted to a fixed interval.

### Random Sampling

```python
rng.choice(arr, size=3)
rng.choice(arr, size=3, replace=False)
```

- `choice()` selects elements.
- `replace=False` prevents selecting the same position repeatedly.
- `replace=True` allows repeated selection.

### Shuffle

```python
rng.shuffle(arr)
```

Modifies the original array.

### Permutation

```python
rng.permutation(arr)
```

Returns a shuffled copy.

# NumPy Mental Model

Remember these 5 principles:

1. **Arrays** store numerical data efficiently.
2. **Vectorization** applies operations to entire arrays.
3. **Axis** determines the direction of aggregation.
4. **Boolean indexing** selects values based on conditions.
5. **Broadcasting** allows compatible shapes to work together.

### Data Scientist's Thinking Pattern

When analyzing a dataset:

1. Understand its structure using `shape`, `ndim`, and `dtype`.
2. Calculate summaries using aggregation functions.
3. Filter values using Boolean indexing.
4. Handle missing and unusual values carefully.
5. Use numerical operations to answer business questions.
6. Explain what the results mean, not just what the code returns.

**NumPy is not just a library to memorize. It is a tool for thinking and working with numerical data.**
