### **What is Random Number Generation?**
Random number generation means creating numbers without manually deciding each value.

### **Why Data Scientists use randomness**
- Randomly select samples from large datasets.
- Create simulated data for experiments.
- Generate random test data.
- Shuffle datasets before training ML models.
- Initialize values in some machine learning algorithms.

### **The Modern NumPy Method**

```python
import numpy as np

rng = np.random.default_rng()
```

np.random → NumPy's random module.



default_rng() → creates a random generator.



rng → variable storing that generator.

### **3. Generate Random Numbers**

### **A. Random decimal numbers**

```python
rng.random(5)
```

Generates 5 random decimal values.



Values are between 0 and 1.



0 is possible; 1 is excluded.

### **B. Random integers**

```python
rng.integers(1, 10, size=5)
```

1 = starting value, included.



10 = stopping value, excluded.



size=5 = generate five numbers.

### **C. Random numbers from a normal distribution**

```python
rng.normal(loc=50, scale=10, size=5)
```

loc=50 → mean of the distribution.



scale=10 → standard deviation.



size=5 → generate five values.

`Values are not restricted to 40–60. They can be lower or higher.`

### **Must Remember**

random() generates decimal values, integers() generates whole numbers, and normal() generates values from a normal distribution.

### **Randomness vs Reproducibility**
This is one of the most important concepts in Data Science.

Suppose you generate random numbers today and get:

```python
[4, 8, 2, 6, 1]
```

Tomorrow, you run the same code and get:

```python
[7, 3, 9, 5, 2]
```

This makes debugging and experiments difficult.

### **Solution: Seed**

```python
rng = np.random.default_rng(seed=42)
rng.integers(1, 10, size=5)
```

`The seed makes the random sequence reproducible.`

### **Random Sampling — rng.choice()**

```python
customers = ["A", "B", "C", "D", "E"]
print(rng.choice(customers, size=3))
```

### **Importand parameters**

```python
rng.choice(customers, size=3, replace= False)
```

customers → source data.



size=3 → select 3 values.



replace=False → do not select the same item more than once.

### **Must Remember**
- replace=True → selection with replacement; duplicates are possible.
- replace=False → selection without replacement; duplicates are not possible.

### **Shuffling — rng.shuffle()**

```python
marks = np.array([45, 67, 89, 32, 76])

# Now you want to randomly rearrange their order.

rng.shuffle(marks)
print(marks)
```

```python
rng.permutation(marks)
print(marks)
```

Original marks remains unchanged.
