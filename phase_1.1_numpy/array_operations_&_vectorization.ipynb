### **Your next challenge — Business sales analysis**

```python
branch_a = np.array([10000, 15000, 20000, 25000])
branch_b = np.array([8000, 12000, 18000, 22000])
```

Combined sales for each corresponding period.



Difference between the two branches' sales.



Total combined sales for all periods.



Average combined sales per period.



The combined sales after a 10% increase.

```python
print(f"Combined Sales: {branch_a + branch_b}")
print(f"Difference between two sales: {branch_a - branch_b}")
total_sales = branch_a + branch_b
print(f"Total Sales: {np.sum(total_sales)}")
average_sales = np.mean(total_sales)
print(f"Average Sales: {average_sales:.2f}")
print(f"Combined sales after 10% increase: {total_sales * 1.10}")
```

```python
marks = np.array([45, 60, 75, 80, 90])
```

Round the values of np.array([12.456, 23.789, 34.567]) to 2 decimal places.

Total marks.

Standard deviation of marks.



Round the values of np.array([12.456, 23.789, 34.567]) to 2 decimal places.

```python
print(f"Total marks: {marks.sum()}")
print(f"Average marks: {marks.mean()}")
print(f"Highest marks: {marks.max()}")
print(f"Lowest marks: {marks.min()}")
print(f"Square root of each marks: {np.sqrt(marks)}")
print(f"Standard deviation of marks: {marks.std():.2f}")
print(f"Rounded Values: {np.round(np.array([12.456, 23.789, 34.567]), 2)}")
```
