### **Aggregation Functions**

`Aggregation functions take many values and calculate one summary result, such as total, average, minimum, or maximum.`

```python
import numpy as np

sales = np.array([1000, 2000, 1500, 3000, 2500])
```

How much did you sell in total?



What was your average daily sale?



What was your highest sale?



What was your lowest sale?



How many sales values are there?

```python
total_sales = sales.sum()
average_sales = sales.mean()
highest_sale = sales.max()
lowest_sale = sales.min()
total_elements = len(sales)
std = sales.std()
print(total_sales)
print(average_sales)
print(highest_sale)
print(lowest_sale)
print(total_elements)
print(std)
```

- np.sum(sales)
- sales.sum()

The first calls a NumPy function. The second calls an array method.

### **Other useful aggregation functions**

#### **1. np.std() — Standard deviation**

```python
marks = np.array([60, 65, 70, 75, 80])

print(f"Standard Deviation: {marks.std():.2f}")
```

`A smaller standard deviation means values are closer to the mean. A larger one means they are more spread out.`

#### **2. np.var() — Variance**

`Measures the average squared distance from the mean.`

```python
print(np.var(marks))
```

#### **3. np.median() — Median**
`Finds the middle value after arranging the data in sorted order.`

```python
import numpy as np
values = np.array([10, 20, 30, 40, 100, 200])

print(np.median(values))
```

### **np.argmax() and np.argmin()**

- np.max() returns the largest value.



- np.argmax() returns the index (position) of the largest value.



- np.min() returns the smallest value.



- np.argmin() returns the index (position) of the smallest value.

```python
import numpy as np

sales = np.array([1000, 3000, 1500, 5000, 2000])

print(sales.max())         # Largest value
print(sales.argmax())      # Position of largest value
print(sales.min())         # Lowest value
print(sales.argmin())      # Position of lowest value
```

### **One more useful function: np.percentile()**

`Percentiles tell us the value below which a given percentage of observations falls.`

```python
marks = np.array([10, 20, 30, 40, 50])

print(np.percentile(marks, 45))
```

`The 50th percentile is the median.`

- 25th percentile: lower quartile (Q1).



- 50th percentile: median (Q2).



- 75th percentile: upper quartile (Q3).

```python
scores = np.array([10, 20, 30, 40, 50])
```

np.percentile(scores, 25)



np.percentile(scores, 50)



np.percentile(scores, 75)

```python
np.percentile(scores, 75)
```

`It does not mean 75% of values equal 40. It means approximately 75% of observations are at or below that percentile value.`

1. 20
2. 30
3. 40

4.The 75th percentile is a value below which approximately 75% of observations fall.
