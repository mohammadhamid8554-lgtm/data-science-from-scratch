`ndim tells us how many dimensions an array has. .shape tells us how many elements exist along each dimension.`

### **1. What is shape?**

```python
import numpy as np

marks = np.array([          # shape = (3, 4)
    [90, 75, 60, 55],
    [76, 66, 85, 45],
    [84, 43, 61, 97]
])


print(f"Shape of marks: {marks.shape}") # Means: (3:Rows, 4:Columns)


```

### **2. Difference Between ndim and shape**

#### **1D-array**

```python
a = np.array([10, 20, 30, 40])
print(f"Shape of a: {a.shape}")
print(f"Dimensions of a: {a.ndim}")
```

#### **2D-array**

```python
b = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

print(f"Dimensions of b: {b.ndim}")
print(f"Shape of b: {b.shape}")
```

#### **3D-array**

```python
sales_data = np.array([
    [[2000, 3000], [1500, 500], [2300, 4000]],
    [[350, 600], [560, 500], [3000, 1200]]
])

print(f"Dimensions of sales data: {sales_data.ndim}")
print(f"Shape of sales data: {sales_data.shape}")
```
