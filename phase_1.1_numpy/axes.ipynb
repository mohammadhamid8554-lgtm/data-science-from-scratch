### **Axis (Understanding Rows and Columns)**

### **What is an axis?**

`An axis tells NumPy which direction to perform an operation`

For a 2D array:
- axis=0 → move down the rows (operate column-wise).
- axis=1 → move across the columns (operate row-wise).

```python
import numpy as np

sales = np.array([
    [100, 200, 300],
    [400, 500, 600]
])

column_wise_total = sales.sum(axis = 0)   # Column-wise operation

row_wise_total = sales.sum(axis = 1)      # Row-wise operation

print(f"Row-wise total: {row_wise_total}")
print(f"Column-wise total: {column_wise_total}")

sales.mean(axis=0)  # Average of each product
sales.mean(axis=1)  # Average of each month

sales.max(axis=0)   # Highest sale per product
sales.max(axis=1)   # Highest sale per month

sales.min(axis=0)   # Lowest sale per product
sales.min(axis=1)   # Lowest sale per month
 
```
