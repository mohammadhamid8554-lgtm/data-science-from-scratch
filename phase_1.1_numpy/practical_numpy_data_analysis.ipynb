### **Our Dataset — Monthly Product Sales**

```python
import numpy as np

sales = np.array([
    [120, 150, 180],
    [100, 130, 160],
    [140, 170, 200],
    [110, 160, 190]
])
```

### **Dataset Structure**
- Rows = months
- Columns = products
- Column 0 = Product A
- Column 1 = Product B
- Column 2 = Product C

```python
print(f"Dataset structure: {sales.shape}")
print(f"Dimensions: {sales.ndim}")
print(f"Total number of elements: {sales.size}")
print(f"Dataset data type: {sales.dtype}")
```

### **Practical Challenge 3 — Final Analysis**

```python
import numpy as np

sales = np.array([
    [120, 150, 180],
    [100, 130, 160],
    [140, 170, 200],
    [110, 160, 190]
])

prices = np.array([200, 300, 400])
```

Calculate each product's percentage contribution to total company revenue.



Find which products generated revenue greater than ₹150,000.



Calculate the revenue difference between the highest-revenue and lowest-revenue months.



Calculate the revenue growth from Month 1 to Month 3 as a percentage of Month 1 revenue.



Calculate total revenue if the company gives a 10% discount on every product.



Identify the month with the lowest revenue using argmin().

```python

total_revenue = np.sum(sales * prices)
total_revenue_per_product = sales.sum(axis = 0) * prices
revenue_difference = np.max(np.sum(sales * prices, axis = 1)) - np.min(np.sum(sales * prices, axis = 1))
revenue_growth = (np.sum(sales[2, :] * prices) - np.sum(sales[0, :] * prices))/np.sum(sales[0, :] * prices) * 100
discounted_total_revenue=np.sum(np.sum(sales * prices, axis = 0) * (1-0.10))
lowest_revenue_month = np.argmin(np.sum(sales * prices, axis = 1))

print(f"Each product contribution: {(total_revenue_per_product/total_revenue) * 100}")
print(f"Product revenue greater than 150000: {total_revenue_per_product[total_revenue_per_product > 150000]}")
print(f"Revenue difference between highest and lowest revenue months: {revenue_difference}")
print(f"Revenue growth(%): {revenue_growth:.2f}%")
print(f"Total discounted(10%) revenue: {discounted_total_revenue}")
print(f"Month with lowest revenue: {lowest_revenue_month}")
```
