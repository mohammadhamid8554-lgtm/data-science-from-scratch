### **What is filtering?**
`Boolean indexing means selecting array elements using conditions such as:`
- Greater than (>)



- Less than (<)



- Equal to (==)



- Not equal to (!=)



- Greater than or equal to (>=)



- Less than or equal to (<=

### **First understand Boolean arrays**

```python
import numpy as np

marks = np.array([45, 80, 30, 90, 65])
print(marks >= 60)
print(marks[marks>=60])
```

- marks >= 60 creates a Boolean array.



- NumPy checks that Boolean array.



- It selects elements where the condition is True.



- It ignores elements where the condition is False.

### **Boolean indexing — Selecting elements**

```python
# Find marks greater than 80
marks[marks > 80]

# Find marks less than 40
marks[marks < 40]

# Find marks equal to 90
marks[marks == 90]

# Find marks not equal to 45
marks[marks != 45]
```

### **Combining Multiple Conditions**

### **Important NumPy Operators**
&	AND — both conditions must be true

\|	OR — at least one condition must be true

~	NOT

```python
sales[(sales > 1000) & (sales < 4000)]
```

```python
sales[(sales < 1000) | (sales > 4000)]
```

```python
sales[~(sales > 1000)]
```

```python
# Correct
sales[(sales > 1000) & (sales < 4000)]

# Incorrect
sales[sales > 1000 and sales < 4000]
```

Why? Python's and and or expect single truth values. NumPy comparisons produce arrays of Boolean values.

### **Filtering and Updating**
`Boolean indexing is not just for selecting data. You can also use it to update selected values.`

```python
marks = np.array([25, 45, 30, 80, 90])
```

```python
marks[marks < 35] += 5
print(marks)
```

Notice:
- 25 became 30.
- 30 became 35.
- Other values remained unchanged.
This is called conditional modification.
