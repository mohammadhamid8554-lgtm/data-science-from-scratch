```python
import csv
from pathlib import Path

try:
    # Task 1. Take user input and validate.
    customer_id = int(input("Enter Customer ID: "))
    customer_name = input("Enter Customer Name: ").strip()
    customer_age = int(input("Enter Customer Age: "))
    customer_email = input("Enter Customer Email: ").strip()

   
    if "@" in customer_email and "." in customer_email:
         
        if  (customer_id > 0 ) and (customer_age > 0) and (customer_email != "") and (customer_name != "") :
        
            # Task 2. Open and write csv file.

            if  not Path("data/customers.csv").exists() or Path("data/customers.csv").stat().st_size == 0:
                with open("data/customers.csv", "w", newline="") as file:
                    writer = csv.writer(file)
                    writer.writerow(["customer_id", "customer_name", "customer_age", "customer_email"])

       # Task 4. Check for duplicate customer IDs.
        
            with open("data/customers.csv", "r", newline="") as file:
                    reader = csv.reader(file)
                    header = next(reader, None)

        
                    duplicate_found = False
                    for customer_ID in reader:
                        if str(customer_id) == customer_ID[0]:
                            duplicate_found = True
                            break
                    
            if  duplicate_found:
                print("Error: Customer ID already exists.")

            else: 
                with open("data/customers.csv", "a", newline="") as file:
                    writer = csv.writer(file)
                    writer.writerow([customer_id, customer_name, customer_age, customer_email])

        else:
             print("Error: Customer ID and age must be positive, and name and email cannot be empty.")

    else:
        print("Error: Invalid Email.")

    with open("data/customers.csv", "r", newline="") as file:
                reader = csv.reader(file)
                header = next(reader, None)
        
                if header is None:
                    print("No customer records found.")
        
                else:
                    print("Customer Records")
                    print("=" * 20)
                    records_found = False
        
                    for record in reader:
                        records_found = True
                        formatted_record = f"ID: {record[0]} | Name: {record[1]} | Age: {record[2]} | Email: {record[3]}"
                        print(formatted_record)
        
                    if not records_found:
                        print("No customer records found.")
    
except ValueError:
    print("Error: Customer ID and age must be integers.")

except OSError:
    print("Error: Could not access or write the customer file.")
```
