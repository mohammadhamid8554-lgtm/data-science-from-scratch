```python
try:

    product_price = float(input("Enter Product Price: "))
    discount_percentage = float(input("Enter Discount Percentage(0-100): "))

    if product_price > 0 and 0 <= discount_percentage <= 100:
        print("\nProceed with calculation.\n")

        discount_amount = product_price * discount_percentage / 100
        final_price = product_price - discount_amount

        print(f"Original Product Price: {product_price}₹.")
        print(f"Discount Percentage: {discount_percentage}%")
        print(f"Discount Amount: {discount_amount:.2f}₹.")
        print(f"Final Price: {final_price:.2f}₹.")

    else:
        print("Error: Price or discount is outside the allowed range.")
    

except ValueError:
    print("Error: Invalid User Input.")
```

### **Must Remember — Data Scientist Cheat List**

- float() converts numeric text into a number that can contain decimals.

- try/except handles expected errors, such as invalid numeric input.

- Validation checks that values follow the rules before calculations begin.

- and requires both conditions to be true.

- Percentage formula: amount × percentage ÷ 100.

- F-strings display calculated values clearly; :.2f formats a number to two decimal places.

- Boundary testing checks important limits, such as 0% and 100%.
