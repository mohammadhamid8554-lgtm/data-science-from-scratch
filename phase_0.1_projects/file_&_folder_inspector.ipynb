````python
from pathlib import Path


try:
    folder_path = Path(input("Enter a Folder Path: "))

# Task 1.

    if folder_path.exists() and folder_path.is_dir():
        print("Path Exists and it is a Folder.")

# Task 2. 
        count_file = 0
        count_folder = 0
        extension_counts = {}

# Task 3.
        for item in folder_path.iterdir():

            if item.is_file():
                                        # Task 4 and 5. 
                if item.suffix:
                    file_type = item.suffix
                    print(f"{item.name} is a file. Extension: {item.suffix}")
                else: 
                    file_type = "No extension"
                    print(f"{item.name} is a file. Estension: {file_type}")

                if file_type in extension_counts:
                        extension_counts[file_type] += 1

                else: 
                    extension_counts[file_type] = 1
                count_file += 1

            elif item.is_dir():
                 print(f"{item.name} is a folder.")
                 count_folder += 1

        print("File type summary:")

        for extension, count in extension_counts.items():
            print(f"{extension}: {count}")
        
            # Task 6.

        if count_file == 0 and count_folder == 0:
            print("The folder is empty.")

        else:
            print(f"Total Files: {count_file}")
            print(f"Total Folders: {count_folder}")

        
        with open("data/inspection_report.txt", "w" ) as file:
           file.write("FILE AND FOLDER INSPECTION\n")
           file.write("=" * 20 + "\n")
           file.write(f"Folder: {folder_path}\n")

           file.write(f"Total Files: {count_file}\n")
           file.write(f"Total Folders: {count_folder}\n")

           file.write("\nFile type summary:\n")
           for extension, count in extension_counts.items():
               file.write(f"{extension}: {count}\n")
        


    elif folder_path.exists() and not folder_path.is_dir():
        print("Folder Path Exists but it is not a Folder.")

    else: 
        print("Folder Paht doesn't Exists.")
        

except OSError:
    print("Error: Folder cannot be accessed.")

````
