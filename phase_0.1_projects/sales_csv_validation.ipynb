````python
import csv

try:

    with open("data/sales.csv", "r", newline="") as file:
        reader = csv.reader(file)
        header = next(reader, None)
        expected_header = ["product", "price", "quantity"]

        print("=== SALES CSV VALIDATION REPORT ===")

        if header == expected_header:

            valid_rows = 0
            invalid_rows = 0
            row_counts = 1

            for record in reader:
                row_is_valid = True
                row_counts += 1

                if  record[0].strip() != "":
                    print(f"Valid Product: {record[0].strip()}")

                else:
                    row_is_valid = False
                    print(f"Row {row_counts}: Error: Product name is missing.") 

                try:
                    price = float(record[1])
                    if price > 0:
                        print(f"Valid price. {price}")

                    else:
                        row_is_valid = False
                        print(f"Row {row_counts}: Error: Price must be greater than zero.")

                except ValueError:
                        row_is_valid = False
                        print(f"Row {row_counts}: Error: Price must be greater than zero and a valid number.") 

                try:
                    quantity = int(record[2])

                    if quantity > 0:
                        print(f"Valid Quantity: {quantity}")

                    else:
                        row_is_valid = False
                        print(f"Row {row_counts}: Error: Quantity must be greater than zero.")

                except ValueError:
                    row_is_valid = False
                    print(f"Row {row_counts}: Error: Quantity must be greater than zero and a valid number.")

                if row_is_valid:
                    valid_rows += 1
                else:
                    invalid_rows += 1 

            print("=== SUMMARY ===")
            print(f"Valid Rows: {valid_rows}")
            print(f"Invalid Rows: {invalid_rows}")
            print(f"Total Rows: {valid_rows + invalid_rows}")


        else:
            print("Error: Invalid CSV header.")

except FileNotFoundError:
    print("Error: The sales.csv file was not found.")
````

# Project 9: Sales CSV Validator — Cheat Sheet

## 2. Python Concepts Learned

| Concept | What to remember |
|---|---|
| `import csv` | Imports Python's CSV-handling module. |
| `open()` | Opens a file for reading or writing. |
| `with open(...)` | Manages the file safely and closes it automatically. |
| `"r"` mode | Opens a file for reading without writing to it. |
| `csv.reader(file)` | Reads CSV records as lists of strings. |

| `strip()` | Removes whitespace from the beginning and end of a string. |

## 3. CSV Data Basics

Example file:

```csv
product,price,quantity
Laptop,50000,2
Mouse,800,5
Keyboard,1500,3
Monitor,12000,2
```

- The first row contains the column names, called the **header**.
- Each following row represents one sales record.
- Each record contains fields separated by commas.
- `csv.reader()` initially returns the fields as strings.

For example:

```python
['Laptop', '50000', '2']
```

The price and quantity must be converted before numeric checks or calculations.

## 4. Data Validation Rules

| Field | Validation rule | Invalid examples |
|---|---|---|
| Product | Must not be empty or whitespace-only | `""`, `"   "` |
| Price | Must be numeric and greater than zero | `"abc"`, `""`, `0`, `-10` |
| Quantity | Must be a whole number greater than zero | `"abc"`, `""`, `0`, `-3`, `"2.5"` |
| Header | Must match the required names and order | `product,quantity,price` |

Remember:

- `float("12.5")` is valid.
- `int("5")` is valid.
- `int("2.5")` raises `ValueError`.
- `"  Mouse  ".strip()` produces `"Mouse"`.
- A valid field does not automatically make the entire row valid.

## 5. The Most Important Logic: Row Validation

```python
row_is_valid = True
```

Start every row as valid. If any field fails, change the flag:

```python
row_is_valid = False
```

After checking every field:

```python
if row_is_valid:
    valid_rows += 1
else:
    invalid_rows += 1
```

**Critical rule:** Never set `row_is_valid = True` again just because a later field passes. Otherwise, a valid price could accidentally overwrite an earlier product-name error.

Initialize counters before the loop, but reset the row-validity flag inside the loop.

## 6. Row Number Tracking

The header is CSV row 1, so the first data record is row 2.

```python
row_counts = 1

for record in reader:
    row_counts += 1
```

Use an f-string to identify the location of an error:

```python
print(f"Row {row_counts}: Error: Product name is missing.")
```

**Why this matters:** A useful data-quality report tells us both what is wrong and where it is wrong.
