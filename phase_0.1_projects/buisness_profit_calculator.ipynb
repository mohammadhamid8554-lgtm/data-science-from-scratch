```python
try:
    revenue = float(input("Enter Revenue: "))
    product_costs = float(input("Enter Product Costs: "))
    other_expenses = float(input("Enter Other Expenses: "))

    if revenue >= 0 and product_costs >= 0 and other_expenses >= 0:
        print("Proceed with calculaion.")

        total_costs = product_costs + other_expenses
        profit = revenue - total_costs

        print(f"Revenue: ₹{revenue:.2f}")
        print(f"Total Costs: ₹{total_costs:.2f}")
        print(f"Profit/Loss: ₹{profit:.2f}")

        if revenue > 0:
             profit_margin = profit / revenue * 100
             print(f"Profit Margin: {profit_margin:2f}%.")
        
        else:
            print("Profit margin cannot be calculated because revenue is zero.")

    else:
        print("Error: Business values are not in allowed range.")

except ValueError:
    print("Error: Invalid User Input.")
```

### **Must Remember — Data Scientist Cheat List**
- Revenue: Money earned from sales.
- Total costs: Product costs plus other expenses.
- Profit or loss: Revenue minus total costs.
- Profit margin: Profit divided by revenue, multiplied by 100.
- Validation: Reject values that break the business rules.
- Division by zero: Check the denominator before dividing.
- Negative profit margin: Indicates a loss when revenue is positive.
- Testing: Include profitable, loss-making, zero-revenue, and invalid-input cases.
