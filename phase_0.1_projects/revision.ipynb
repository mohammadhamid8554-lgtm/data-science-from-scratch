## 3. Conditional Statements

Used to make decisions.

```python
if condition:
    # Runs when True

elif another_condition:
    # Runs when the previous condition is False
    # and this condition is True

else:
    # Runs when all previous conditions are False
```

Example:

```python
marks = 85

if marks >= 90:
    grade = "A"
elif marks >= 75:
    grade = "B"
elif marks >= 60:
    grade = "C"
else:
    grade = "F"
```

**Important:** Python checks conditions from top to bottom.

## 4. Loops

Loops repeat instructions.

### For loop

Used to iterate over a sequence.

```python
numbers = [10, 20, 30]

for number in numbers:
    print(number)
```

### While loop

Repeats while a condition remains True.

```python
count = 1

while count <= 5:
    print(count)
    count += 1
```

### Loop control

```python
break       # Stops the loop
continue    # Skips the current iteration
```

## 5. Functions

Functions organize reusable code.

```python
def calculate_total(price, quantity):
    total = price * quantity
    return total

result = calculate_total(100, 3)
```

### Important terms

- `def`: Defines a function.
- Parameters: Variables in the function definition.
- Arguments: Actual values passed to the function.
- `return`: Sends a result back to the caller.
- `print()`: Displays information; it does not return that displayed value.

## 6. Data Structures

### List

Ordered, mutable, allows duplicates.

```python
numbers = [10, 20, 30]

numbers.append(40)
numbers.remove(20)
numbers[0]       # 10
numbers[-1]      # 40
```

### Tuple

Ordered, immutable, allows duplicates.

```python
coordinates = (10, 20)
```

### Dictionary

Stores data as key-value pairs.

```python
student = {
    "name": "Aarav",
    "marks": 85
}

student["name"]          # Access value
student["marks"] = 90    # Update value
student.get("age", 0)    # Safe access with default
student.keys()           # All keys
student.values()         # All values
student.items()          # Key-value pairs
```

### Set

Unordered collection of unique values.

```python
numbers = {10, 20, 20, 30}

numbers.add(40)
numbers.remove(20)
```

**Remember:** Sets remove duplicates. `len()` counts unique elements in a set.

## 7. Indexing and Slicing

Indexing accesses one item. Slicing extracts multiple items.

```python
name = "Python"

name[0]       # P
name[-1]      # n
name[0:3]     # Pyt
name[:3]      # Pyt
name[3:]      # hon
name[::-1]    # nohtyP
```

The ending index in slicing is excluded.

## 8. Comprehensions

A shorter way to create collections.

### List comprehension

```python
squares = [x**2 for x in range(1, 6)]
```

### List comprehension with condition

```python
even_numbers = [x for x in range(10) if x % 2 == 0]
```

### Dictionary comprehension

```python
squares = {x: x**2 for x in range(1, 6)}
```

## 12. Exception Handling

Prevents common errors from crashing a program unexpectedly.

```python
try:
    number = int(input("Enter a number: "))
    print(number)

except ValueError:
    print("Please enter a valid whole number.")
```

### Common exceptions

| Exception | Meaning |
|---|---|
| `ValueError` | Correct type expected, but invalid value |
| `FileNotFoundError` | File does not exist |
| `JSONDecodeError` | Invalid JSON format |
| `KeyError` | Dictionary key does not exist |
| `TypeError` | Operation used with an unsuitable type |

**Remember:** `try` contains risky operations; `except` handles specified errors.

## 15. CSV Handling

CSV means Comma-Separated Values. It stores tabular data.

### Reading CSV with DictReader

```python
import csv

with open("data/students.csv", "r") as file:
    reader = csv.DictReader(file)

    for student in reader:
        print(student["name"])
        print(student["marks"])
```

### Writing CSV with DictWriter

```python
import csv

with open("data/report.csv", "w", newline="") as file:
    fieldnames = ["name", "marks"]

    writer = csv.DictWriter(file, fieldnames=fieldnames)

    writer.writeheader()

    writer.writerow({
        "name": "Aarav",
        "marks": 85
    })
```

**Important CSV rules:**

- CSV values are generally read as strings.
- Convert numeric values before calculations.
- `DictReader()` accesses columns by their header names.
- `writer.writeheader()` writes column names.
- `newline=""` helps avoid extra blank rows when writing CSV files.

Example:

```python
marks = int(student["marks"])
```

## 16. JSON Handling

JSON means JavaScript Object Notation. It is commonly used to store structured data.

### Reading JSON

```python
import json

with open("data/product.json", "r") as file:
    product = json.load(file)
```

### Writing JSON

```python
with open("data/product.json", "w") as file:
    json.dump(product, file, indent=4)
```

### Updating JSON data

```python
product["price"] = 500

with open("data/product.json", "w") as file:
    json.dump(product, file, indent=4)
```

**Remember:**

- `json.load()` reads JSON from a file.
- `json.dump()` writes Python data into a JSON file.
- `indent=4` makes the JSON easier to read.
- JSON objects become Python dictionaries.
- JSON arrays become Python lists.

## 17. Data Validation Basics

Data validation checks whether input data is acceptable before using it.

### Numeric conversion

```python
try:
    marks = int("85")
except ValueError:
    print("Invalid number")
```

### Range validation

```python
if 0 <= marks <= 100:
    print("Valid marks")
else:
    print("Invalid marks")
```

### Missing dictionary key

```python
student.get("marks")
```

**Important distinction:**

- `ValueError`: The value cannot be converted as expected.
- Range check: The value is numeric but outside the permitted range.
- Missing value: The expected data is absent or empty.

For data-processing tasks, a useful general workflow is:

**Read → Validate → Handle invalid data → Process → Check results → Save.**

## 18. Important Data Science Programming Principles

1. **Readable code:** Use meaningful variable and function names.
2. **Reusable code:** Put repeated logic into functions.
3. **Input validation:** Never blindly trust external data.
4. **Error handling:** Anticipate common failures.
5. **Correct data types:** Convert values before calculations.
6. **File safety:** Use `with open()` and understand file modes.
7. **Data integrity:** Avoid silently accepting invalid values.
8. **Logical order:** Validate data before performing dependent calculations.
9. **Test edge cases:** Check empty files, missing values, invalid text, and unexpected numbers.
10. **Understand the logic:** Do not rely on memorizing code.
