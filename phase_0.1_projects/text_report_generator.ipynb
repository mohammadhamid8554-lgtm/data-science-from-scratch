# Project 7: Text Report Generator

## 1. Project Goal
Build a Python program that collects a report title and report text, validates the inputs, calculates text statistics, formats a readable report, and saves it to a text file.

## 2. Concepts Learned

### A. Collect and Validate User Input
- `input()` collects information from the user.
- `.strip()` removes whitespace from the beginning and end of text.
- `if` checks whether the title and report text are empty.

```python
report_title = input("Enter report title: ").strip()
report_text = input("Enter report information: ").strip()

if report_title != "" and report_text != "":
    print("Valid input")
else:
    print("Error: Title or text cannot be empty.")
```

### B. Count Characters, Words, and Lines

```python
characters = len(report_text)
words = len(report_text.split())
lines = len(report_text.splitlines())
```

- `len(text)` counts characters, including spaces and punctuation.
- `split()` separates text into words using whitespace.
- `splitlines()` separates text into lines.
- `len()` counts the items returned by `split()` and `splitlines()`.

### C. Format a Report Using an f-string

```python
formatted_report = (
    f"{'=' * 20}\n"
    f"{report_title}\n"
    f"{'=' * 20}\n"
    f"{report_text}\n\n"
    f"Character count: {characters}\n"
    f"Word count: {words}\n"
    f"Line count: {lines}"
)

print(formatted_report)
```

- An f-string inserts variable values into text.
- `{'=' * 20}` creates a separator.
- `\n` moves the following text to a new line.
- Parentheses allow a long string expression to span multiple lines.

### D. Save the Report to a File

```python
try:
    with open("data/report.txt", "w") as file:
        file.write(formatted_report)

    print("Report saved successfully.")

except OSError:
    print("Error: Could not save the report.")
```

- `open()` opens a file.
- `"w"` creates a file or overwrites its existing contents.
- `with` automatically closes the file when the block finishes.
- `file.write()` writes text into the file.
- `OSError` handles common file-system errors, such as a missing parent folder or
