````python
import json

with open("data/inventory.json", "r") as file:
    inventory = json.load(file)

    total_stock_value = 0
    for product, items in inventory.items():

        total_stock_value += items["price"] * items["quantity"]
        print(f"Product: {product} | Price: {items['price']} | Quantity: {items['quantity']}")
    print(f"Total Inventory Value: {total_stock_value}")


    product_name = "Mouse"
    added_quantity = -5

    if product_name in inventory and added_quantity > 0:
        inventory[product_name]["quantity"] += added_quantity
        print(f"Updated Mouse Quantity: {inventory['Mouse']['quantity']}")

    else:
        print("Error: Product must exists and added quantity must be greater than zero..")

    sold_quantity = 100

    if product_name in inventory and inventory[product_name]["quantity"] >= sold_quantity:
        inventory[product_name]["quantity"] -= sold_quantity
        print("Sale Successful.")
        print(f"Remaining Mouse Quantity: {inventory[product_name]['quantity']}")

    else:
        print("Error: Product is missing or insufficient stock.")

with open("data/inventory.json", "w") as file:
    json.dump(inventory, file, indent= 4)
````

# Project 10: Product Inventory Manager — Revision Cheat Sheet

## 1. Project Goal
Build a Python program that manages product inventory using a JSON file.

- Store product names, prices, and quantities.
- Display product details.
- Calculate total inventory value.
- Add new stock.
- Record product sales.
- Validate inputs and prevent invalid stock updates.
- Save updated inventory to a JSON file.

## 2. Important Concepts and Syntax

### A. Read JSON data
```python
with open("data/inventory.json", "r") as file:
    inventory = json.load(file)
```
- `json.load(file)` reads JSON data into a Python object.
- `"r"` means read mode.
- The inventory is stored as a nested dictionary.

### B. Access product details
```python
inventory["Mouse"]["price"]
inventory["Mouse"]["quantity"]
```
- The first key selects the product.
- The second key selects its price or quantity.

### C. Loop through products
```python
for product_name, details in inventory.items():
    print(product_name, details["price"], details["quantity"])
```
- `.items()` gives each dictionary key and value.
- A loop processes every product.

### D. Calculate inventory value
```python
total_value = 0

for details in inventory.values():
    total_value += details["price"] * details["quantity"]
```
**Formula:**

`Product value = Price × Quantity`

`Total inventory value = Sum of all product values`

### E. Add stock safely
```python
if product_name in inventory and added_quantity > 0:
    inventory[product_name]["quantity"] += added_quantity
else:
    print("Invalid product or quantity.")
```
- Check that the product exists.
- Added quantity must be greater than zero.
- `+=` increases the existing quantity.

### F. Record a sale safely
```python
if product_name in inventory and inventory[product_name]["quantity"] >= sold_quantity:
    inventory[product_name]["quantity"] -= sold_quantity
    print("Sale successful.")
else:
    print("Product missing or insufficient stock.")
```
- Check that the product exists.
- Ensure enough stock is available.
- `-=` reduces the quantity.
- The condition prevents selling more stock than is available.

**Important:** In a more complete program, also validate that `sold_quantity > 0`.

### G. Save updated inventory
```python
with open("data/inventory.json", "w") as file:
    json.dump(inventory, file, indent=4)
```
- `json.dump()` writes the Python data to the JSON file.
- `"w"` means write mode and replaces the file's existing contents.
- `indent=4` makes the JSON easier to read.

## 3. Common Mistakes to Avoid

1. Accessing a product before checking whether it exists.
2. Allowing zero or negative quantities during stock addition.
3. Selling more products than are available.
4. Forgetting to save changes to the JSON file.
5. Opening a file in write mode before finishing reading its contents.
6. Confusing `json.load()` (read from a file) with `json.dump()` (write to a file).
7. Forgetting to test the actual saved file after the program runs.

## 4. Tests Performed

- Normal stock addition and sale.
- Unknown product handling.
- Negative stock addition rejected.
- Insufficient stock sale rejected.
- JSON file checked to confirm rejected operations did not change the saved quantity.

## 6. Must Remember as a Data Scientist

- **Load → Process → Validate → Save:** a useful pattern for working with stored data.
- **Validate before updating:** bad data can produce incorrect results.
- **Protect data integrity:** rejected operations should not change the inventory.
- **Test edge cases:** normal cases alone are not enough.
- **Verify persistence:** confirm that the saved file matches the expected result.
- **Understand the logic:** do not rely on memorizing code without knowing what each condition does.
