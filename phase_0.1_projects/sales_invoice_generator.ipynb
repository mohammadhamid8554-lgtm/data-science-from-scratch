# Project 11: Sales Invoice Generator — Cheat Sheet

## 1. Project Goal
Build a Python program that collects customer and product details, calculates the total amount, validates input, displays an invoice, and saves it to a text file.

## 2. Taking User Input

```python
customer_name = input("Enter Your Name: ").strip()
product_name = input("Enter Product Name: ").strip()
price = float(input("Enter Product Price: "))
quantity = int(input("Enter Product Quantity: "))
```

- `input()` collects information from the user as text.
- `.strip()` removes spaces from the beginning and end of a string.
- `float()` converts input into a decimal number.
- `int()` converts input into a whole number.

## 3. Handling Invalid Numeric Input

```python
try:
    price = float(input("Enter Product Price: "))
    quantity = int(input("Enter Product Quantity: "))

except ValueError:
    print("Error: Price or quantity must be valid numbers.")
```

**Remember:** `ValueError` occurs when a value cannot be converted to the requested type, such as converting `"abc"` to a number.

## 4. Validating Price and Quantity

```python
if price > 0 and quantity > 0:
    print("Valid input")

else:
    print("Error: Price and Quantity must be greater than zero.")
```

- `>` means greater than.
- `and` means both conditions must be true.
- A price or quantity of zero or less is rejected.

## 5. Calculating the Total Amount

```python
total_amount = price * quantity
```

**Formula:**

Total Amount = Price per Item × Quantity

Example: ₹800 × 4 = ₹3,200.

## 6. Displaying a Formatted Invoice

```python
print("SALES INVOICE")
print("-" * 30)
print(f"Customer Name: {customer_name}")
print(f"Product Name: {product_name}")
print(f"Price per item: ₹{price}")
print(f"Quantity: {quantity}")
print(f"Total Product Amount: ₹{total_amount}")
```

- An f-string inserts variable values into text.
- `"-" * 30` prints 30 hyphens.
- `\n` creates a new line when writing text to a file.

## 7. Saving the Invoice to a Text File

```python
with open("data/sales_invoice.txt", "w") as file:
    file.write("SALES INVOICE\n")
    file.write("-" * 30 + "\n")
    file.write(f"Customer Name: {customer_name}\n")
    file.write(f"Product Name: {product_name}\n")
    file.write(f"Price per item: {price}\n")
    file.write(f"Quantity: {quantity}\n")
    file.write(f"Total Product Amount: {total_amount}\n")

print("Invoice saved successfully.")
```

- `open()` opens a file.
- `"w"` opens a file for writing. It creates a new file or replaces existing contents.
- `file.write()` writes text to the file.
- `with` automatically closes the file after the block finishes.
- `"data/sales_invoice.txt"` is the file path.

**Important:** Put the file-writing block inside the valid-input branch so invalid input does not overwrite the saved invoice.

## 8. Complete Program Structure

```python
try:
    # 1. Collect input
    customer_name = input("Enter Your Name: ").strip()
    product_name = input("Enter Product Name: ").strip()
    price = float(input("Enter Product Price: "))

        print(f"Quantity: {quantity}")

        # 5. Save invoice
        with open("data/sales_invoice.txt", "w") as file:
            file.write("SALES INVOICE\n")
            file.write("-" * 30 + "\n")
            file.write(f"Customer Name: {customer_name}\n")
            file.write(f"Product Name: {product_name}\n")
            file.write(f"Price per item: {price}\n")
            file.write(f"Quantity: {quantity}\n")
            file.write(f"Total Product Amount: {total_amount}\n")

```
