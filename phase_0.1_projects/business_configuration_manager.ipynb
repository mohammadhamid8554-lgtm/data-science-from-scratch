## 1. Project Objective
Build a Python program that stores, loads, displays, and validates business settings using a JSON file.

## 2. Project Requirements
The program should be able to:

1. Store business settings in a JSON file.
2. Load the settings into Python using `json.load()`.
3. Display the saved business settings.
4. Check whether required settings exist.
5. Validate settings to ensure their values are acceptable.
6. Handle missing files and invalid JSON data appropriately.

## 3. Configuration Settings
The program will manage these settings:

- **Shop name:** Smart Shop
- **Currency:** INR
- **Tax rate:** 5%
- **Minimum stock level:** 5 items

## 4. Tools and Concepts
- Python dictionaries
- JSON data format
- `json.load()` and `json.dump()`
- File handling with `with open()`
- Conditional statements
- Input and data validation
- Exception handling using `try/except`

## 5. Expected Outcome
At the end of the project, the program should be able to read business settings from a JSON file, display them, detect missing or invalid settings, and handle common file-related errors.

## 6. Learning Rules

```python
import json
try:
    with open("data/business_config.json", "r")  as file:
        config = json.load(file)
        if isinstance(config, dict):

            if "currency" in config:
                print(f"Currency: {config['currency']}")

            else:
                print("Error: Currency setting is missing.")

            if "shop_name" in config:
                print(f"Shop Name: {config['shop_name']}")

            else:
                print("Error: Shop name setting is missing.")
                
            if "tax_rate" in config:
                print(f"Tax rate: {config['tax_rate']}%")

                if isinstance(config["tax_rate"], (int, float)) and type(config['tax_rate']) is not bool and config['tax_rate'] >= 0 and config['tax_rate'] <= 100:
                    print("Tax rate is valid.")

                else:
                    print("Error: Tax rate is invalid.")

            else:
                print("Error: Tax rate setting is missing.")

            if "minimum_stock_level" in config :
                print(f"Minimum stock level: {config['minimum_stock_level']}")

                if  type(config['minimum_stock_level']) is int  and config["minimum_stock_level"] > 0:
                    print("Minimum stock level setting is valid.")

                else:
                    print("Error: Minimum stock level setting is invalid.")
            else:
                print("Error: Minimum stock level setting is missing.")
        else:
            print("Error: Configuration data must be a dictionary.")

except TypeError:
    print("Error: Tax rate and minimum stock level setting must be valid numbers.")

except json.JSONDecodeError:
    print("Error: File contains invalid JSON.")

except FileNotFoundError:
    print("Error: Business configuration file not found.")
```

### **Must Remember — Data Scientist Cheat List**

- json.load(file) reads JSON data from a file.

- isinstance(data, dict) checks whether data is a dictionary.

- "key" in data checks whether a key exists.

- isinstance(value, (int, float)) checks for numeric types, but remember that Booleans also count as integers in Python.

- type(value) is int checks for an actual integer, excluding Booleans.

- try and except help handle errors without an unexpected crash.

- Always validate data before trusting it.
