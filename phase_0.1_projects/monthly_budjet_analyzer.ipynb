### **1. Project objective**

Imagine a student has ₹15,000 each month. They spend money on food, transport, and shopping.

Your program must calculate their total expenses, determine how much money remains, and tell them whether they stayed within their budget.

### **2. Requirements**

Your program must do these six things:

1. Take user input
Ask for the monthly budget and three expenses: food, transport, and shopping.



2. Validate the input
Convert the values into numbers and handle invalid input using try/except.



3. Calculate total expenses
Add food, transport, and shopping expenses.



4. Calculate the remaining budget
Subtract total expenses from the monthly budget.



5. Make a decision
Use if/elif/else to distinguish between overspending, spending the exact budget, and staying under budget.



6. Display a clear report
Show the total expenses, remaining budget, and an appropriate message.

```python
try:

    monthly_budget = float(input("Enter your monthly budget: "))
    food_expense = float(input("Enter your food expense: "))
    transport_expense = float(input("Enter your transport expense: "))
    shopping_expense = float(input("Enter your shopping expense: "))

    # Handle negative input amount.

    if monthly_budget < 0 or food_expense < 0 or transport_expense < 0 or shopping_expense < 0:
        print("Error: Amound can not be negative.")

    else:

        total_expense = food_expense + transport_expense + shopping_expense
        remaining_budget = monthly_budget - total_expense

        if total_expense == monthly_budget:
            print("Total expense is equal to monthly budget.")

        elif total_expense > monthly_budget:
            print("Total expense is more than monthly budget.")

        else:
            print("Total expense is within monthly budget.")

        print(f"Monthly Budget:   {monthly_budget}.")
        print(f"Total Expense:    {total_expense}")
        print(f"Remaining Budget: {remaining_budget}")


except ValueError:
    print("Invalid input amount.")
```

### **Must Remember — Data Scientist Cheat List**

- input() returns a string.



- float() converts suitable strings into numbers.



- ValueError occurs when a conversion cannot be performed.



- try/except lets a program handle expected errors.



- Validate data before using it in calculations.



- A valid value is not necessarily a meaningful value; context determines what is acceptable.



- Test normal inputs, boundary values such as zero, and invalid inputs.
