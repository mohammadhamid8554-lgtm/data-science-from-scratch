```python
import csv

with open("data/students.csv", "r") as file:
        reader = csv.DictReader(file)
        for student in reader:
            
            try:
                if 0 <= int(student["maths"]) <= 100  and 0 <= int(student["science"]) <= 100 and 0 <= int(student["english"]) <= 100:

                  print(f"Student Name: {student['name']}")
                  print(f"Student Id: {student['student_id']}")
                  print(f"Math Marks: {int(student['maths'])}")
                  print(f"Science Marks: {int(student['science'])}")
                  print(f"English Marks: {int(student['english'])}")
                  total_marks = int(student["maths"]) + int(student["science"]) + int(student["english"])
                  print(f"Total marks: {total_marks}")
  
                  percentage = total_marks/300 * 100
                  print(f"Overall Percentage: {percentage:.2f}%")

                  if int(student["maths"]) >= 35 and int(student["science"]) >= 35 and int(student["english"]) >= 35:
                    result = "Pass" 
                  else:
                    result = "Fail"
                
                  print(f"Result: {result}")

                  if percentage >= 90 and result == "Pass":
                     grade = "A"

                  elif 75 <= percentage < 90 and result ==  "Pass":
                     grade = "B"

                  elif 60 <= percentage < 75 and result == "Pass":
                     grade = "C"

                  elif  35 <= percentage < 60 and result == "Pass":
                     grade = "D"

                  else:
                     grade = "F"
                
                  print(f"Grade: {grade}")

                else:
                   print("Error: Marks must be between 0 and 100")
 
            except ValueError:
                print("Error: Studnet Marks must be whole a numbers.") 
```
