```python
import csv
expenses = []

try:
    # 1. Collect first user input.

    number_of_expenses = int(input("Enter Number of Expenses: "))

    if number_of_expenses <= 0:
        raise ValueError("Expense number must be greater than zero.")

    for expense_number in range(number_of_expenses):
        expense_amount = float(input("Enter Expense Amount: "))
        expense_category = input("Enter Expense Category: ").strip().title()
        expense_description = input("Enter Description: ").strip().title()

        if expense_category  == "" or expense_description == "":
            raise ValueError("Category or description cannot be empty.")

        if expense_amount <=0:
            raise ValueError("Expense amount must be greater than zero.")

        expense = {
        "amount" : expense_amount,
        "category" : expense_category,
        "description" : expense_description
        }

        expenses.append(expense)


    total_expenses = sum(expense_record["amount"] for expense_record in expenses)
    print(f"Total Spending:  ₹{total_expenses:.2f}")
    
    # 3. Summarize expenses by category.

    category_totals = {}

    for expense_record in expenses:
        category = expense_record["category"]
        amount = expense_record["amount"]

        if category in category_totals:
            category_totals[category] += amount
        else:
            category_totals[category] = amount

    print("\nSpending by category:")

    for category, amount in category_totals.items():
        print(f"{category}:  ₹{amount:.2f}")


    # 4. Save the expense in a csv file.
    
    with open("data/expenses.csv", "w", newline="") as file:
          fieldnames = [ "amount", "category", "description" ]
        
          writer  = csv.DictWriter(file, fieldnames=fieldnames)

          writer.writeheader()

          writer.writerows(expenses)

    print("File saved to data/expenses.csv.")

    
except ValueError:
    print("Error: Invalid User Input.")
```

## Must Remember — Personal Expense Journal

| Concept | What to Remember | Python Example |
|---|---|---|
| Empty list | Create a list to store multiple records. | `expenses = []` |
| Dictionary | Store related details about one expense. | `expense = {"amount": 120, "category": "Food", "description": "Lunch"}` |
| List of dictionaries | Store multiple records in one list. | `expenses.append(expense)` |
| User input | Convert numeric input to the required type. | `amount = float(input("Enter amount: "))` |
| Integer input | Use `int()` when you need a whole number. | `count = int(input("Number of expenses: "))` |
| Input cleaning | Remove extra spaces and standardize capitalization. | `category = input("Category: ").strip().title()` |
| Input validation | Reject amounts that are zero or negative. | `if amount <= 0: raise ValueError("Amount must be positive")` |
| Empty text validation | Prevent empty categories or descriptions. | `if not category.strip(): raise ValueError("Category cannot be empty")` |
| Exception handling | Handle invalid numeric input and validation errors. | `try: ... except ValueError: ...` |
| Repetition | Use `range()` when the user specifies how many records to enter. | `for i in range(count):` |
| Access dictionary values | Use a key to retrieve a value from a record. | `expense["amount"]` |
| Calculate total | Use `sum()` with a generator expression to add values. | `total = sum(e["amount"] for e in expenses)` |
| Category summary | Use a dictionary to store each category's running total. | `category_totals[category] = category_totals.get(category, 0) + amount` |
| Dictionary iteration | Use `.items()` to access keys and values together. | `for category, total in category_totals.items():` |
| Number formatting | Display money with two decimal places. | `print(f"₹{total:.2f}")` |
| CSV module | Import Python's built-in CSV tools. | `import csv` |
| Open a CSV file | Use `with open()` to manage the file safely. | `with open("data/expenses.csv", "w", newline="") as file:` |
| CSV column names | Define the columns your records should contain. | `fieldnames = ["amount", "category", "description"]` |
| CSV writer | Create a writer that writes dictionary records. | `writer = csv.DictWriter(file, fieldnames=fieldnames)` |
| CSV header | Write column names as the first row. | `writer.writeheader()` |
| Write records | Save all dictionaries as CSV rows. | `writer.writerows(expenses)` |
| File mode `"w"` | Creates a file or replaces its existing contents. | `open("data/expenses.csv", "w")` |
| Folder awareness | The `data` folder must exist before opening a file inside it this way. | `"data/expenses.csv"` |
| Financial precision | `float` is suitable for practice, but exact monetary calculations may require `Decimal`. | `from decimal import Decimal` |
| Data quality | Validate data before saving or analyzing it. | Check amounts, categories, and descriptions before appending records. |

### Key Lessons

- A **dictionary** represents one record; a **list of dictionaries** represents many records.
- A **loop** processes each expense; a **summary dictionary** groups expenses by category.
- Validate data before storing it, and check that totals match the original records.
- CSV files store data in rows and columns, making them useful for later analysis with Pandas.
- Opening a file in `"w"` mode overwrites previous contents, so choose the file mode carefully.
