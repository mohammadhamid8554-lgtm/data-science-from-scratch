### **Lesson 4: Handling Messy Numeric Data with pd.to_numeric()**
astype() works when values are already compatible with the target type. pd.to_numeric() gives us better options when real-world data contains invalid values.

### **1. The problem**

````python
sales = pd.Series(["5000", "3000", "Unknown", "7000"])
````

Try converting sales column:

````python
sales.astype(float)
# it will raise a ValueError because "Unknown" is not a valid number.
````

### **2. Meet pd.to_numeric()**

````python
pd.to_numeric(sales)
# It converts valid numeric values into numbers.
````

### **3. The important parameter: errors**
| Option | What it does |
|---|---|
| `errors="raise"` | Raises an error for invalid values (default). |
| `errors="coerce"` | Converts invalid values into `NaN`. |
| `errors="ignore"` | Keeps the original values when conversion fails. |

Important: coerce does not fix the bad value. It converts it into a missing value so you can investigate and handle it.

### **4. Backward, current, forward, practical**
- Backward: astype() converts compatible values.
- Current: pd.to_numeric() handles invalid numeric entries.
- Forward: You'll use this during data cleaning and preprocessing.
- Practical: Sales datasets often contain entries such as "Unknown", "N/A", or "Not Available".

### **5. Your exercise**

````python
sales = pd.Series(["1200", "2500", "Unknown", "4000", "N/A"])
display(sales)
````

1. Convert it using pd.to_numeric() with errors="coerce".
2. Display the converted Series.
3. Count how many values became missing.
4. Calculate the mean of the valid numeric values.

````python
sales_numeric = pd.to_numeric(sales, errors="coerce")

print("Converted sales:")
display(sales_numeric.to_frame(name="Sales"))

missing_count = sales_numeric.isna().sum()
average_sales = sales_numeric.mean()

print(f"Missing values: {missing_count}")
print(f"Average of valid sales: {average_sales:,.2f}")
````

### **Lesson 5: Handling Missing Values After Conversion**

errors="coerce" converts invalid values into NaN. Now we must decide what to do with those missing values.

**1. The problem**

````python
sales = pd.Series(["1200", "2500", "Unknown", "4000", "N/A"])

sales = pd.to_numeric(sales, errors="coerce")

display(sales)
````

But should we delete them? Not automatically.

### **2. Three ways to handle missing values**

### **Method 1: Identify missing values**

````python
sales.isna()
# Returns True where values are missing.
````

````python
sales.isna().sum()
# Counts missing values.
````

### **Method 2: Remove missing values**

````python
sales.dropna()
# Returns a Series without missing values.
````

Remember: this does not automatically modify the original Series.

### **Method 3: Replace missing values**

````python
sales.fillna(0)
# Replaces missing values with zero.
````

You can also use the mean

````python
sales.fillna(sales.mean())
#This replaces missing values with the average of available value
````

### **3. Important: Don't blindly replace missing values**
Imagine a sales dataset:
- `0` means no sales; `NaN` means unknown sales.
- Replacing `NaN` with `0` can mislead, so choose a treatment based on what the missing value represents.



### **4. Connections**
- Backward: pd.to_numeric() helped identify invalid numeric entries.
- Current: You're learning to handle the resulting missing values.
- Forward: Missing-value treatment is important in data preprocessing and machine learning.
- Practical: Incorrect handling can distort sales averages and business reports

### **5. Your exercise**

````python
sales = pd.Series([1200, 2500, None, 4000, None, 3000])

display(sales)
````

Complete these tasks:
1. Count the missing values.
2. Create a new Series without missing values using dropna().
3. Replace missing values with 0 in another Series.
4. Replace missing values with the mean in a third Series.
5. Calculate the mean before and after replacing missing values with zero. Explain why they differ.

````python
# 1. Count missing values
missing_count = sales.isna().sum()
print(f"Missing values: {missing_count}")

# 2. Remove missing values
clean_sales = sales.dropna()
print("\nClean sales:")
display(clean_sales)

# 3. Replace missing values with 0
filled_with_zero = sales.fillna(0)
print("\nSales with missing values filled by zero:")
display(filled_with_zero)

# 4. Replace missing values with the mean
filled_with_mean = sales.fillna(sales.mean())
print("\nSales with missing values filled by the mean:")
display(filled_with_mean)

# 5. Compare means
mean_before = sales.mean()
mean_after_zero = filled_with_zero.mean()

print(f"\nMean before filling: {mean_before:,.2f}")
print(f"Mean after filling with zero: {mean_after_zero:,.2f}")

# When calculating the mean before filling, the missing values are ignored. After filling them with zero, they are included in the calculation, which lowers the mean.
````

### **Lesson 6: Converting Columns with Nullable Integer Types**
Sometimes a numeric column contains missing values. Normal integer types cannot represent NaN, so Pandas provides a nullable integer type called Int64.

### **1. The problem**

````python
import pandas as pd

data = {
    "Product": ["Laptop", "Phone", "Tablet"],
    "Quantity": ["2", "3", "Unknown"]
}

df = pd.DataFrame(data)

display(df)
````

````python
# Convert the quantity column

df["Quantity"] = pd.to_numeric(df["Quantity"], errors="coerce")
display(df)
````

Why float?

Because traditional NumPy integer types cannot store NaN. Pandas uses a floating-point type here.

### **2. Meet Pandas nullable integers**

`Pandas provides a special integer type: Int64. Notice that capital 'I'`

````python
df["Quantity"] = df["Quantity"].astype("Int64")
display(df)
````

Now the column can contain both integers and missing values.

### **3. Important difference**

| Type | Supports missing values? |
|---|---|
| `int64` | No |
| `float64` | Yes, using `NaN` |
| `Int64` | Yes, using `<NA>` |

### **4. Connections**
- Backward: You learned pd.to_numeric() and errors="coerce".
- Current: You're learning how to preserve integer data when missing values exist.
- Forward: This helps you prepare reliable datasets for analysis and machine learning.
- Practical: Customer counts, product quantities, and number of orders are usually integers, even when some values are missing.

A date stored as text is not automatically a real date. Pandas provides pd.to_datetime() to convert it into a datetime type.
### **1. Understand the problem**

````python
import pandas as pd

data = {
    "Product": ["Laptop", "Phone", "Tablet"],
    "Sale_Date": ["2026-01-15", "2026-02-20", "2026-03-10"]
}

df = pd.DataFrame(data)
display(df)
````

check the data types:

We want to:
- Find sales between two dates.

````python
dates = pd.Series([
    "2026-01-15",
    "2026-02-20",
    "Invalid",
    "2026-03-10"
])

display(dates)
````

This will fail:

### **4. Important distinction**
| Numeric conversion | Datetime conversion |
|---|---|
| `pd.to_numeric()` | `pd.to_datetime()` |
| Invalid values become `NaN` | Invalid values become `NaT` |
| Produces a numeric dtype | Produces a datetime dtype |

### **Lesson 9: Extracting Information from Datetime Columns**

Once a column becomes datetime, Pandas lets us extract its year, month, day, weekday, and more using .dt.

### **1. Meet the .dt accessor**

````python
import pandas as pd

df = pd.DataFrame({
    "Product": ["Laptop", "Phone", "Tablet"],
    "Sale_Date": ["2026-01-15", "2026-02-20", "2026-03-10"]
})

df["Sale_Date"] = pd.to_datetime(df["Sale_Date"])
````

### **2. Common datetime properties**
| Code | Meaning |
|---|---|
| `df["Sale_Date"].dt.year` | Year |
| `df["Sale_Date"].dt.month` | Month number |
| `df["Sale_Date"].dt.day` | Day |
| `df["Sale_Date"].dt.day_name()` | Weekday name |
| `df["Sale_Date"].dt.month_name()` | Month name |
| `df["Sale_Date"].dt.dayofweek` | Weekday number (Monday = 0) |

Example:

````python
df["Year"] = df["Sale_Date"].dt.year
df["Month"] = df["Sale_Date"].dt.month
df["Day"] = df["Sale_Date"].dt.day
df["Month_Name"] = df["Sale_Date"].dt.month_name()

df
````

Now your DataFrame contains separate date features.

```python
df.fillna(df.mean(numeric_only=True))  # Fill using mean
```

## 5. Nullable Integer Type

```python
df["Quantity"] = df["Quantity"].astype("Int64")
```

| Type | Missing values |
|---|---|
| `int64` | Not supported |
| `float64` | Supports NaN |
| `Int64` | Supports <NA> |

## 6. Convert Text into Datetime

```python
df["Sale_Date"] = pd.to_datetime(
    df["Sale_Date"],
    format="%d-%m-%Y",
    errors="coerce"
)
```

- `NaT` = Missing datetime value.
- `NaN` = Missing numeric value.

## 7. Extract Date Features

```python
df["Year"] = df["Sale_Date"].dt.year
df["Month"] = df["Sale_Date"].dt.month
df["Day"] = df["Sale_Date"].dt.day

df["Month_Name"] = df["Sale_Date"].dt.month_name()
df["Day_Name"] = df["Sale_Date"].dt.day_name()
```

## 8. Clean Text Columns

```python
df["Name"] = (
    df["Name"]
    .astype("string")
    .str.strip()
    .str.lower()
)
```

- `.str.strip()` removes extra spaces at the beginning and end.
- `.str.lower()` converts text to lowercase.

## 9. Create Calculated Columns

```python
df["Total_Sales"] = df["Price"] * df["Quantity"]
```

Pandas performs vectorized calculations across rows.

## Golden Rules

1. Inspect data types before calculations.
2. Choose types based on what the data represents.
3. Use `pd.to_numeric()` for messy numeric data.
4. Use `Int64` when integers need to support missing values.
5. Use `pd.to_datetime()` for date conversion.
6. Never replace missing values blindly.
7. Preserve identifiers according to their meaning and usage.
8. Always inspect the result after conversion.
