# Pandas Topic 9 — Duplicate Data

## 1. What is duplicate data?

Duplicate data means the same record appears more than once when it should represent only one unique record.

Duplicates can cause:
- Incorrect totals and averages.
- Inflated revenue.
- Incorrect customer and transaction counts.
- Misleading analysis and machine-learning results.

**Important:** Repeated values do not always mean duplicate records.

Example: A customer can make multiple purchases, so their `Customer_ID` can appear multiple times legitimately.

## 2. Detecting duplicate rows — `duplicated()`

Identifies duplicate occurrences and returns Boolean values (`True` or `False`).

```python
df.duplicated()
```

Count duplicate occurrences:

```python
df.duplicated().sum()
```

Display duplicate rows:

```python
df[df.duplicated()]
```

By default, Pandas keeps the first occurrence as `False` and marks later occurrences as `True`.

## 3. Understanding the `keep` parameter

### `keep="first"` (default)
- Keeps the first occurrence as `False`.
- Marks later occurrences as `True`.

```python
df.duplicated(keep="first")
```

### `keep="last"`
- Keeps the last occurrence as `False`.
- Marks earlier occurrences as `True`.

```python
df.duplicated(keep="last")
```

### `keep=False`
- Marks every occurrence belonging to a duplicate group as `True`.
- Unique records remain `False`.

```python
df.duplicated(keep=False)
```

Count all records belonging to duplicate groups:

```python
df.duplicated(keep=False).sum()
```

## 4. Detecting duplicates using specific columns — `subset`

Check duplicates using one column:

```python
df.duplicated(subset=["Customer_ID"])
```

Check duplicates using multiple columns:

```python
df.duplicated(subset=["Name", "City"])
```

Display all records belonging to repeated customer IDs:

```python
df[df.duplicated(subset=["Customer_ID"], keep=False)]
```

**Remember:** Choose columns based on business meaning, not just appearance.

## 5. Removing duplicate rows — `drop_duplicates()`

Remove completely duplicated rows:

```python
df.drop_duplicates()
```

Remove duplicates based on a specific column:

```python
df.drop_duplicates(subset=["Customer_ID"])
```

Keep the first occurrence:

```python
df.drop_duplicates(subset=["Customer_ID"], keep="first")
```

Keep the last occurrence:

```python
df.drop_duplicates(subset=["Customer_ID"], keep="last")
```

Remove every record belonging to a duplicate group:

```python
df.drop_duplicates(subset=["Customer_ID"], keep=False)
```

**Important:** `drop_duplicates()` returns a new DataFrame by default. Assign the result if you want to save the cleaned data.

```python
clean_df = df.drop_duplicates()
```

## 6. Text standardization

Visually similar text can contain different characters.

Example:

- `"Delhi"`
- `" Delhi "`
- `"DELHI"`

Remove leading and trailing spaces:

```python
df["City"].str.strip()
```

Convert text to lowercase:

```python
df["City"].str.lower()
```

Combine both operations:

```python
df["City"].str.strip().str.lower()
```

Create a standardized column without overwriting the original:

```python
df["City_Clean"] = df["City"].str.strip().str.lower()
```

**Remember:** Standardization helps compare text consistently, but matching names alone does not prove two records represent the same person.

## 7. Preserve original data

Create a copy before cleaning:

```python
clean_df = df.copy()
```

This protects the original DataFrame.

## 8. Real-world duplicate-cleaning workflow

1. Preserve the original dataset.
2. Standardize relevant text columns.
3. Detect completely duplicated rows.
4. Investigate duplicates using meaningful keys.
5. Display all records belonging to duplicate groups.
6. Understand the business meaning of the records.
7. Decide which records are genuinely duplicates.
8. Remove only confirmed duplicates.
9. Validate the cleaned dataset.

## 9. Critical business concepts

- Customer_ID identifies a customer.
- Order_ID generally identifies an order.
- One customer can have multiple orders.
- A repeated customer ID is not automatically a duplicate.
- A repeated order ID requires investigation.
- The last occurrence is not necessarily the latest or most accurate record.
- Verify uniqueness rules and dataset grain before deleting records.

## 10. Most important distinction

`duplicated()` = Detect duplicates.

`drop_duplicates()` = Remove duplicates.

**Golden Rule: Duplicate detection is a technical operation; duplicate identification is a data decision.**

1. The Double-Click Trap (Exact Duplicates)

Problem: Users double-clicking checkout buttons generated exact row clones (e.g., transaction 101).

Question: Write code to identify and drop these exact duplicates, retaining the first occurrence. How many rows remain?

````python
clean_df = df.copy()

duplicates = df.duplicated().sum()
dropped_duplicates = df.drop_duplicates(keep="first").reset_index(drop=True)

print(f"Total number of duplicate records: {duplicates}")
print("Clean dataset after removal of duplicate records: ")
display(dropped_duplicates)
print(f"Remaining rows: {len(dropped_duplicates)}")
````
