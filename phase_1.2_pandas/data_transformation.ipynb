### **Lesson 1 — What Is Data Transformation?**

| Product  | Price | Quantity |
|----------|------:|---------:|

| Mouse    | 500   | 5        |

The business asks: **How much revenue did each product generate?**

### **Lesson 3: Mapping Values with .map()**

`.map() lets you replace values in a column using a mapping rule.`
| Code | Meaning |
|------|---------|
| L    | Laptop |
| M    | Mouse  |
| K    | Keyboard |


### **1. The pattern**

```python
mapping = {
    "L": "Laptop",
    "M": "Mouse",
    "K": "Keyboard"
}
```

- A dictionary stores the conversion rules: old value → new value.

- Pandas .map() applies those rules to a Series.

### **2. Your exercise — attempt first**

```python
import pandas as pd

df = pd.DataFrame({
    "Product_Code": ["L", "M", "K", "L"],
    "Quantity": [2, 5, 3, 1]
})

display(df)
```

Create a dictionary named product_mapping that converts:
- "L" → "Laptop"
- "M" → "Mouse"
- "K" → "Keyboard"

Then create a new column named Product_Name using .map().

Finally, display the DataFrame.

```python
product_mapping = {
    "L": "Laptop",
    "M": "Mouse",
    "K": "Keyboard"
}

df["Product_Name"] = df["Product_Code"].map(product_mapping)
display(df)
```

### **Why it works**
1. product_mapping stores the conversion rules.
2. .map(product_mapping) checks each value in Product_Code.
3. It replaces each code with its mapped name.
4. The result becomes the new Product_Name column.

### **Important rule**
If a value is missing from the mapping dictionary, .map() returns NaN for that value.

Example: if the dataset contains "X" but your dictionary has no "X" key, its mapped result will be missing.

### Key Concepts

- **Dictionary:** Maps old values to new values  
    `{"L": "Laptop", "M": "Mouse", "K": "Keyboard"}`
- **`.map()`:** Uses that dictionary to convert values in a column.
- **New column:** The converted values are stored in `Product_Name`.
- **Unmapped values:** Unknown values not included in the dictionary become `NaN` (missing).

### **Lesson 4: Replacing Values with .replace()**
`.replace()` changes specific values into other values. It's useful when cleaning or standardizing data.

### **1. The idea**
Suppose a dataset contains inconsistent payment labels:
| Payment |
|---------|
| Cash |
| cash |
| CASH |
| Card |

These three cash labels mean the same thing, but Pandas sees them as different strings.

We can standardize them.

### **2. The pattern**

```python
df["Payment"] = df["Payment"].replace({
    "cash": "Cash",
    "CASH": "Cash"
})

# This replaces the exact values specified in the dictionary.
```

### **3. Your exercise — attempt first**

```python
import pandas as pd

df = pd.DataFrame({
    "Payment": ["Cash", "cash", "CASH", "Card", "card"],
    "Amount": [500, 300, 700, 1200, 400]
})
```

**Standardize the Payment column so that:**
- "cash" and "CASH" become "Cash"
- "card" becomes "Card"
- Existing "Cash" and "Card" values remain unchanged

Then display the DataFrame.

```python
df["Payment"] = df["Payment"].replace(
    {
        "CASH": "Cash",
        "cash": "Cash",
        "card": "Card"
    }
)

# Additional method(my approch)
df["Clean_Payment"] = df["Payment"].str.title()
display(df)

print("Final DataFrame:")
display(df)

# Quick Inspection
df["Clean_Payment"].value_counts()
```
