### **Lesson 3: Mapping Values with .map()**

`.map() lets you replace values in a column using a mapping rule.`
| Code | Meaning |
|------|---------|
| L    | Laptop |
| M    | Mouse  |
| K    | Keyboard |


### **1. The pattern**

````python
mapping = {
    "L": "Laptop",
    "M": "Mouse",
    "K": "Keyboard"
}
````

- A dictionary stores the conversion rules: old value → new value.

- Pandas .map() applies those rules to a Series.

### **2. Your exercise — attempt first**

````python
import pandas as pd

df = pd.DataFrame({
    "Product_Code": ["L", "M", "K", "L"],
    "Quantity": [2, 5, 3, 1]
})

display(df)
````

Create a dictionary named product_mapping that converts:
- "L" → "Laptop"
- "M" → "Mouse"
- "K" → "Keyboard"

Then create a new column named Product_Name using .map().

Finally, display the DataFrame.

````python
product_mapping = {
    "L": "Laptop",
    "M": "Mouse",
    "K": "Keyboard"
}

df["Product_Name"] = df["Product_Code"].map(product_mapping)
display(df)
````

### **Why it works**
1. product_mapping stores the conversion rules.
2. .map(product_mapping) checks each value in Product_Code.
3. It replaces each code with its mapped name.
4. The result becomes the new Product_Name column.

### **Important rule**
If a value is missing from the mapping dictionary, .map() returns NaN for that value.

Example: if the dataset contains "X" but your dictionary has no "X" key, its mapped result will be missing.

### Key Concepts

- **Dictionary:** Maps old values to new values  
    `{"L": "Laptop", "M": "Mouse", "K": "Keyboard"}`
- **`.map()`:** Uses that dictionary to convert values in a column.
- **New column:** The converted values are stored in `Product_Name`.
- **Unmapped values:** Unknown values not included in the dictionary become `NaN` (missing).

### **Lesson 5: Applying a Function with .apply()**

`.apply() lets you apply a function to values in a Series or across rows/columns of a DataFrame.`

### **1. Connect the concepts**
- Previous: .map() transforms Series values using a mapping.
- Now: .apply() applies a function or rule.
- Later: This helps when cleaning data and creating features.
- Real use: Convert raw marks into grades, or calculate a business category.

### **2. Understand the idea**
Suppose students have marks:
| Student | Marks |
|---------|------:|
| A       | 85    |
| B       | 42    |
| C       | 29    |

You want to assign grades:
- Marks ≥ 80 → "A"
- Marks ≥ 40 → "B"
- Otherwise → "C"

````python
import pandas as pd

df = pd.DataFrame({
    "Student": ["Aman", "Sara", "Ravi", "Neha"],
    "Marks": [92, 67, 35, 78]
})
````

````python
def assign_grade(marks):
    if marks >= 80:
        return "A"
    elif marks >= 40:
        return "B"

    else:
        return "C"
````

````python
df["Grade"] = df["Marks"].apply(assign_grade)
display(df)
````

### **1. Why use it?**

````python
import numpy as np

condition = (

    (df["order_status"] == "Delivered")
    & (df["quantity"] > 0)
    & (df["unit_price"] >= 0)
    & (df["discount_pct"] >= 0)
    & (df["discount_pct"] <= 100)
    & ~(df[["quantity", "unit_price", "discount_pct"]].isna().any(axis=1))
)

df["Recognized_Revenue"] = np.where(condition, df["Revenue"], np.nan)
display(df.head(2))
````

# Pandas Topic 12 — Data Transformation
## Subtopic 1: `apply()` vs Vectorized Code

### 1. Vectorization

Vectorization means applying an operation to an entire Series or array without writing a Python loop to process each value manually.

```python
df["Line_Total"] = df["unit_price"] * df["quantity"]
```

**Why prefer it?**
- Usually faster for simple calculations.
- Code is shorter and easier to read.
- Uses optimized operations in Pandas and NumPy.

### 2. `apply()`

`apply()` lets us apply a custom function to a Series or DataFrame.

For row-wise DataFrame logic:

```python
df.apply(function, axis=1)
```

- `axis=1` → function receives each row.
- `axis=0` → function operates on each column.

Row-wise `apply()` can be slower because it repeatedly invokes Python-level logic.

### 3. Choose the Right Approach

Prefer vectorized operations when possible:

- Arithmetic: `*`, `+`, `/`
- Boolean masks
- `.between()`
- `np.where()`
- `np.select()`
- `.where()`

Use `apply()` when custom logic is difficult to express conveniently with built-in vectorized operations.

### 4. Business Rules and Vectorization

A business rule describes **what** the result should mean.

Vectorization or `apply()` describes **how** we calculate it.

A business rule does not automatically require `apply()`.

### 5. Validation Rule

Code running without an error does not prove the result is correct.

Validate:
- Shape and row count
- Data types
- Missing values
- Sample calculations
- Invalid values
- Business-rule correctness

### Remember

**Vectorized code first. `apply()` when genuinely needed. Always validate the output.**

I have used vectorized method to calculate Line_Total as this is an expressible condition and we can use other vectorized method too. But in the case of handling Quantity_Category, the condition was not expressible with vectorized method. So, we used np.select() function instead and we can also use apply() fucntion too.
