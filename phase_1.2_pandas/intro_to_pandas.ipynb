# Pandas — Topic 1: Introduction

## What is Pandas?
Pandas is a Python library used to load, inspect, manipulate, clean, and analyze structured data.

## Why Pandas?
- Specialized tools for tabular data.
- Reduces manual coding.
- Makes data cleaning easier.
- Supports efficient data analysis.
- Helps prepare datasets for visualization and ML.

## Tabular Data
- Row → Record / Observation.
- Column → Variable.
- Cell → Individual value.
- Index → Row label.

## Main Data Structures
- Series → 1D labeled data.
- DataFrame → 2D labeled table.

## NumPy vs Pandas
- NumPy → Numerical arrays and computation.
- Pandas → Structured data and analysis.
- Pandas uses NumPy in important ways.

## Data Types
- Numerical → Age, Spending, Marks.
- Categorical → City, Department, Product.
- IDs may look numerical but represent identifiers.

## Data Scientist Workflow
Raw Data → Inspect → Clean → Transform → Analyze → Insights

## Important Principle
Understand what each column represents before performing operations.
