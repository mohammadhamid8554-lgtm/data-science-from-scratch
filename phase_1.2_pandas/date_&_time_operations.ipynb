```python
orders["converted_date"] = pd.to_datetime(orders["order_date"], errors="coerce", format="%Y-%m-%d")
# print(orders["converted_date"].isna().sum())
# print(orders['order_date'].isna().sum())

print(orders["converted_date"].head())
```

### **Extract year from converted_date**

```python
orders["year"] = orders["converted_date"].dt.year
print(orders["year"])
```

### **Check for missing values in year**

```python
orders["year"].isna().sum()
```

### **Extract Month**

```python
orders["month"] = orders["converted_date"].dt.month
print(orders['month'])
print(f"Missing count: {orders['month'].isna().sum()}")
```

### **Extract weekday names**

```python
orders["weekday_names"] = orders["converted_date"].dt.day_name()
print(orders['weekday_names'])
print(f"Missing count: {orders['weekday_names'].isna().sum()}")
```
