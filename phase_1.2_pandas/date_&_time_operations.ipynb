### **resample()**

### **The intuition**
You used groupby() to count orders by the Order_Year and Order_Month columns.

resample() is another way to summarize data over time. It is especially useful when your data has a proper datetime column and you want time-based buckets such as:
- Orders per month
- Sales per week
- Revenue per day

One key difference:
- groupby() groups by values in columns.
- resample() groups by time intervals, usually using a datetime index.
