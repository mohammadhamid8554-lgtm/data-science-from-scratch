### **Selecting and Filtering Data**

### **1. Why do we need selection and filtering?**

```
Selection helps you choose the data you need.

Filtering helps you find records that satisfy a condition.

```

### **First concept: Selecting columns***
Let's use this example DataFrame.

````python
import pandas as pd

df = pd.DataFrame({
    "Product": ["Laptop", "Phone", "Tablet", "Monitor"],
    "Category": ["Electronics", "Electronics", "Electronics", "Electronics"],
    "Quantity": [5, 10, 8, 4],
    "Revenue": [250000, 150000, 80000, 60000]
})

df["Product"]
````

This selects the Product column.

Important: Selecting one column using a single pair of square brackets returns a Series.

### **B. Select multiple columns**

````python
df[["Product", "Revenue"]]
````

This selects two columns: Product and Revenue.

Why two square brackets?
- Outer [] → Pandas column selection.
- Inner [] → Python list containing column names.

This returns a DataFrame, not a Series.

### **Selecting Rows with loc and iloc**

First understand the difference:
- loc → selects using labels (row names/index labels).
- iloc → selects using positions (0, 1, 2, 3...).

````python
import pandas as pd

df = pd.DataFrame({
    "Product": ["Laptop", "Phone", "Tablet", "Monitor"],
    "Quantity": [5, 10, 8, 4],
    "Revenue": [250000, 150000, 80000, 60000]
})
df
````

### **1. Using iloc — Position-based selection**


````python
df.iloc[0]
````

````python
df.iloc[2]
````

### **Using loc — Label-based selection**

````python
df.loc[0]
````

````python
df.loc[2]
````

With the default index, loc[2] and iloc[2] appear to give the same row. But their meaning is different.
- loc[2] → Find label 2.
- iloc[2] → Find position 2.

# Pandas Topic 6 — Selecting and Filtering Data

## 1. Selecting Columns

```python
df["Product"]                  # Single column → Series
df[["Product", "Revenue"]]      # Multiple columns → DataFrame
df[["Product"]]                 # One column → DataFrame
```

- Series = 1D data
- DataFrame = 2D table

## 2. Selecting Rows

### iloc — Position-based selection

```python
df.iloc[0]       # First row
df.iloc[2]       # Third row
df.iloc[-1]      # Last row
```

### loc — Label-based selection

```python
df.loc[0]        # Row with label 0
df.loc[2]        # Row with label 2
```

## 3. Row Slicing

```python
df.iloc[:3]      # Positions 0, 1, 2
df.loc[:3]       # Labels 0, 1, 2, 3
```

- iloc slicing excludes ending position.
- loc slicing includes ending label.

## 4. Selecting Rows and Columns Together

```python
df.iloc[0:2, 0:2]

df.loc[0:2, ["Product", "Revenue"]]
```

Syntax:

```python
df.iloc[rows, columns]
df.loc[rows, columns]
```

## 5. Boolean Filtering

```python
df[df["Revenue"] > 100000]

df[df["Quantity"] < 8]

df[df["Revenue"] == 60000]

df[df["Revenue"] <= 150000]
```

Comparison operators:

| Operator | Meaning |
|---|---|
| `>` | Greater than |
| `<` | Less than |
| `>=` | Greater than or equal |
| `<=` | Less than or equal |
| `==` | Equal |
| `!=` | Not equal |

## 6. Multiple Conditions

```python
# AND
df[(df["Revenue"] > 100000) & (df["Quantity"] > 5)]

# OR
df[(df["Revenue"] > 200000) | (df["Quantity"] < 5)]

# NOT
df[~(df["Revenue"] > 100000)]
```

- `&` → Both conditions must be True.
- `|` → At least one condition must be True.
- `~` → Reverses Boolean values.

Always put each condition inside parentheses.

Use `&` and `|` instead of `and` and `or` with Pandas Series.

## 7. isin() — Matching Multiple Values

```python
df[df["Product"].isin(["Laptop", "Tablet"])]

df[df["Category"].isin(["Electronics", "Accessories"])]
```

Checks whether values belong to a specified collection.

## 8. between() — Filtering a Range

```python
df[df["Revenue"].between(60000, 150000)]

df[df["Quantity"].between(5, 10)]
```

By default, both boundaries are included.

```python
df[df["Revenue"].between(60000, 150000, inclusive="neither")]
```

Excludes both boundaries.

## 9. query() — Alternative Filtering Syntax

```python
df.query("Revenue > 100000")

df.query("Quantity < 8")

df.query("Revenue > 100000 and Quantity > 5")

df.query("Revenue > 200000 or Quantity < 5")
```

Filter and select columns:

```python
df.query("Revenue > 100000")[["Product", "Revenue"]]
```

## 10. Filtering + Column Selection

```python
df.loc[df["Revenue"] > 100000, ["Product", "Revenue"]]

df.loc[df["Quantity"] > 5, ["Product", "Quantity"]]

df.loc[
    (df["Revenue"] > 100000) & (df["Quantity"] > 5),
    ["Product", "Revenue", "Quantity"]
]
```

Structure:

```python
df.loc[row_condition, column_selection]
```

## Core Mental Model

- Selection → Which columns do I need?
- Filtering → Which rows satisfy my condition?
- loc → Labels
- iloc → Positions
- isin() → Match values from a list
- between() → Match values within a range
- query() → Readable alternative for filtering
- Boolean conditions → True keeps the row; False excludes it
