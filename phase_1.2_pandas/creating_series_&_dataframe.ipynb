Pandas provides:
- pd.Series() → create a Series.
- pd.DataFrame() → create a DataFrame.

### **Creating a Series**

### **Method A - Create Seriies from a list**

````python
import pandas as pd

marks = pd.Series([85, 92, 76, 88])
print(marks)
````

### **Notice:**
- 85, 92, 76, 88 → actual values.
- 0, 1, 2, 3 → automatically created index labels.
- int64 → integer data type.

### **Method B — Create Series with custom index**

````python
marks = pd.Series(
    [85, 92, 76,],
    index=["Rahul", "Sara", "Amit"]
)

print(marks)
````

### **Remember**

pd.Series(values, index=labels)
- values → actual data
- index → labels associated with values

### **Method C — Create Series from a dictionary**

````python
marks = pd.Series({

      "Ali" : 85,
      "Sara": 92,
      "Rahul": 76
       
})

print(marks)
````

### **Here:**
- Dictionary keys become Series index labels.
- Dictionary values become Series values.

### **Creating a DataFrame**
### **Method A — From a dictionary of lists**

````python
data = {
    "Name": ["Rahul", "Sara", "Tom"],
    "Age": [30, 23, 25],
    "Marks":[85, 92, 76]
}

df = pd.DataFrame(data, index=[1, 2, 3])
print(df)
````

### **Understand the structure:**
- Dictionary keys → column names.
- Dictionary values → column data.
- Each list → values for one column.
- Each position across lists → one row.
- Index → automatically created row labels.

### **Method B — From a list of lists**

````python
data = [
    ["Ali", 18, 85],
    ["Sara", 35, 92],
    ["Rahul", 18, 76]
]

df = pd.DataFrame(data, columns= ["Name", "Age", "Marks"], index= range(1, 4))
print(df)
````

### **Here:**
- Each inner list → one row.
- columns → column names.
- Each position inside a row → value for that column.

### **Creating a DataFrame from a list of dictionaries**

````python
data = [
    {"Name": "Rahul", "Age": 18},
    {"Name": "Sara", "Age": 20},
    {"Name": "Ali", "Age": 18}
]

df = pd.DataFrame(data, index=range(1, 4))
print(df)
````

### **Notice:**
- Each dictionary → one row.
- Dictionary keys → column names.
- Dictionary values → cell values.

If a dictionary is missing a key, Pandas fills that missing cell with NaN.

````python
data = [
    {"Name": "Ali", "Age": 18},
    {"Name": "Sara"},
]

df = pd.DataFrame(data)
print(df)
````

### **Creating an empty DataFrame**

````python
df = pd.DataFrame(columns=["Name", "Age", "Marks"])
print(df)
````

# Must Remember — Creating Series and DataFrames

## 1. Creating a Series

### Series from a List
```python
import pandas as pd

sales = pd.Series([1200, 2500, 1800, 3000])
```
- List values become Series values.
- Pandas automatically generates index labels: `0, 1, 2, ...`.

### Series with Custom Index
```python
marks = pd.Series(
    [85, 92, 76],
    index=["Ali", "Sara", "Rahul"]
)
```
- `values` → actual data.
- `index` → custom labels.

### Series from a Dictionary
```python
marks = pd.Series({
    "Ali": 85,
    "Sara": 92,
    "Rahul": 76
})
```
- Dictionary keys → Series index.
- Dictionary values → Series values.

---

## 2. Creating a DataFrame

### A. Dictionary of Lists
```python
data = {
    "Name": ["Ali", "Sara", "Rahul"],
    "Age": [18, 19, 18],
    "Marks": [85, 92, 76]
}

df = pd.DataFrame(data)
```
- Dictionary keys → column names.
- Lists → column values.
- Each position across lists → one row.
- All lists must have equal lengths.

### B. List of Lists
```python
data = [
    ["Ali", 18, 85],
    ["Sara", 19, 92],
    ["Rahul", 18, 76]
]

df = pd.DataFrame(
    data,
    columns=["Name", "Age", "Marks"]
)
```
- Each inner list → one row.
- `columns=` → column names.
- Each position inside a row → cell value.

### C. List of Dictionaries
```python
data = [
    {"Name": "Ali", "Age": 18},
    {"Name": "Sara", "Age": 19},
    {"Name": "Rahul", "Age": 20}
]

df = pd.DataFrame(data)
```
- Each dictionary → one row.
- Dictionary keys → column names.
- Dictionary values → cell values.
- Missing keys → `NaN`.

---

## 3. Custom DataFrame Index
```python
df = pd.DataFrame(
    data,
    index=["P101", "P102", "P103"]
)
```
- `index=` assigns row labels.
- It does NOT create a new column.

## 4. Empty DataFrame

### Completely Empty
```python
df = pd.DataFrame()
```

### Empty with Column Names
```python
df = pd.DataFrame(
    columns=["Name", "Age", "Marks"]
)
```

---

## 5. Inspect DataFrame Structure

```python
df.shape
df.index
df.columns
df.columns.tolist()
```

| Expression | Meaning |
|---|---|
| `df.shape` | `(rows, columns)` |
| `df.index` | Row labels |
| `df.columns` | Pandas Index containing column labels |
| `df.columns.tolist()` | Column labels as Python list |

Example:
```python
df.shape
```
Output:
```python
(100, 5)
```
Means 100 rows and 5 columns.

---

## 6. Important Rules

- Series = 1D labeled data.
- DataFrame = 2D labeled table.
- Dictionary of lists → data organized column-wise.
- List of dictionaries → data organized record-wise.
- List of lists → each inner list represents one row.
- Unequal list lengths in a dictionary of lists → `ValueError`.
- Missing dictionary keys in records → `NaN`.
- Pandas automatically generates an index if you don't provide one.
- A DataFrame column usually behaves like a Series.

---

## 7. Quick Memory Map

| Python Structure | Pandas Structure |
|---|---|
| List | Series |
| Dictionary | Series |
| Dictionary of lists | DataFrame |
| List of lists | DataFrame |
| List of dictionaries | DataFrame |

**Core Principle:**
Pandas interprets Python data according to its structure.
