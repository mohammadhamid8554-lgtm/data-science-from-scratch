### **What is a Pandas Series?**
`A Series is a one-dimensional labeled data structure in Pandas.`

````python
import numpy as np
import pandas as pd

marks = np.array([60, 90, 70, 80])
print(f"Without labels: {marks}")   # Without labels
print(f"With labels:\n{pd.Series(marks, index = range(1, 5))}")  
````

### **Pandas stores:**
- Values → actual data.
- Index → labels associated with those values.

### **Why do we need a Series?**

Suppose you have this NumPy array:

````python
marks = np.array([85, 90, 70])
marks[0]
````

But imagine you want to identify marks using meaningful labels instead of only positions.

````python
marks = pd.Series(
    [85, 90, 70],
    index = ["Rahul", "Aman", "Priya"]
)

print(marks)
print(marks["Aman"])
````

`Core idea: A Series connects values with labels.`

- Values:	  Actual data
- Index:	  Labels for each value
- Data type:  (dtype)	Type of values stored

Position tells us where the value is located.

Index label tells us how that value is identified.

### **Vectorized operation**

````python
marks = pd.Series([80, 90, 70], index = ["Rahul", "Aman", "Priya"])
print(marks + 5)
````

### **Important Series Properties**

````python
marks = pd.Series([80, 70, 95, 60])
print(marks.shape)
print(marks.size)
print(marks.dtype)
print(marks.index)
````

### **Practical Data Science Example**

````python
sales = pd.Series([1200, 1500, 900, 2000, 1800])
print(sales.sum())
print(sales.mean())
print(sales.max())
print(sales.min())
````

NumPy → numerical array analysis.



Pandas Series → numerical analysis with labeled data.

# Pandas Topic 2 — Series

## Definition
A Series is a one-dimensional labeled array containing values and an associated index.

## Structure
- Index → Labels identifying values.
- Values → Actual stored data.
- dtype → Data type of values.

## Creation
```python
import pandas as pd

marks = pd.Series([80, 90, 70])
```

## Properties
```python
marks.index
marks.values
marks.dtype
marks.shape
marks.size
```

## Aggregation
```python
marks.sum()
marks.mean()
marks.max()
marks.min()
marks.median()
```

## Vectorized Operations
```python
marks + 5
marks * 2
marks > 75
```

## Filtering
```python
marks[marks > 75]
```

## Important Rules
- Series is one-dimensional.
- Default index begins at 0.
- Index labels can be duplicated.
- Series supports vectorized operations.
- Series arithmetic aligns matching index labels.
- Series values can be converted/accessed as a NumPy array.
- `shape` returns a tuple.
- `size` returns the total number of elements.

## NumPy Connection
- Both support vectorized computation.
- Series adds labeled data and index-based alignment.
