### **Important parameters of read_csv()**

````python
df = pd.read_csv("data/students.csv", sep=";")
df
````

````python
df = pd.read_csv("data/students.csv", sep=";", usecols=["Name", "Marks"])
df
````

````python
df = pd.read_csv("data/students.csv", sep = ";", nrows=2)
df
````

### **index_col — use a column as index**

````python
df = pd.read_csv("data/students.csv", index_col="Student_ID")
df
````

# Pandas Topic 5 — Reading CSV & Excel Files

## 1. Reading CSV Files

```python
import pandas as pd

df = pd.read_csv("data/students.csv")
```

- `pd.read_csv()` loads a CSV file into a DataFrame.
- CSV files store tabular data separated by delimiters.

## 2. Reading Excel Files

```python
df = pd.read_excel("data/students.xlsx")
```

- `pd.read_excel()` loads an Excel worksheet.
- `openpyxl` is commonly used for `.xlsx` files.

```python
df = pd.read_excel(
    "data/students.xlsx",
    engine="openpyxl"
)
```

## 3. Selecting Specific Columns

```python
df = pd.read_csv(
    "data/product_sales.csv",
    usecols=["Product", "Revenue"]
)
```

- `usecols` loads only the required columns.
- It can reduce memory usage and unnecessary processing.

## 4. Loading Specific Rows

```python
df = pd.read_csv(
    "data/students.csv",
    nrows=5
)
```

- `nrows=5` loads the first five data rows.
- The header is not counted as a data row.

## 5. Setting a Column as Index

```python
df = pd.read_csv(
    "data/students.csv",
    index_col="Student_ID"
)
```

- `index_col` makes a specified column the DataFrame index.
- Index values do not have to be unique or sequential.

## 6. Selecting Excel Worksheets

By worksheet name:

```python
df = pd.read_excel(
    "data/students.xlsx",
    sheet_name="Results"
)
```

By worksheet position:

```python
df = pd.read_excel(
    "data/students.xlsx",
    sheet_name=2
)
```

- Worksheet positions start at 0.
- `sheet_name=2` means the third worksheet.

## 7. Important Parameters

| Parameter | Purpose |
|---|---|
| `usecols` | Select columns |
| `nrows` | Select number of data rows |
| `index_col` | Set index column |
| `sheet_name` | Select Excel worksheet |
| `engine` | Specify Excel processing library |
| `sep` | Specify CSV delimiter |

## 8. Common Errors

- `FileNotFoundError` → Check file path and working directory.
- `ParserError` → Check malformed CSV structure or delimiters.
- `UnicodeDecodeError` → Check file encoding.
- Missing Excel engine → Install required library such as `openpyxl`.

## 9. Data Scientist Best Practices

- Load only required columns when possible.
- Inspect unfamiliar datasets before analysis.
- Check file paths carefully.
- Understand what one row represents.
- Do not assume repeated identifiers mean duplicate records.
- Validate loaded data before starting analysis.

## Remember

**read_csv/read_excel → Load data**

**usecols → Select columns**

**nrows → Select rows**

**index_col → Set index**

**sheet_name → Select worksheet**
