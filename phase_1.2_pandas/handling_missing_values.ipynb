# Pandas — Topic 8: Handling Missing Values

## 1. What are Missing Values?

Missing values are empty or unavailable values in a dataset.

Examples:
- `NaN`
- `None`

Missing values can occur because of:
- Users not entering information.
- Data collection errors.
- Failed data merging.
- Unavailable information.

## 2. Detecting Missing Values

### `isna()`

Identifies missing values. Returns Boolean values (`True` or `False`).

```python
df.isna()
```

### `isnull()`

Works exactly like `isna()`.

```python
df.isnull()
```

### Count missing values in each column

```python
df.isna().sum()
```

### Count all missing cells

```python
df.isna().sum().sum()
```

### Find rows containing at least one missing value

```python
df[df.isna().any(axis=1)]
```

### Count rows containing missing values

```python
df.isna().any(axis=1).sum()
```

Remember:
- `axis=0`: Work column-wise.
- `axis=1`: Work row-wise.

## 3. Removing Missing Values — `dropna()`

### Remove rows containing any missing value

```python
df.dropna()
```

Default behaviour: `how="any"`.

### Remove columns containing missing values

```python
df.dropna(axis=1)
```

### Save the cleaned DataFrame

```python
clean_df = df.dropna()
```

By default, `dropna()` returns a new DataFrame. It does not modify the original DataFrame.

## 4. Understanding `how`

### `how="any"`

Removes a row if at least one value is missing.

```python
df.dropna(how="any")
```

### `how="all"`

Removes a row only if every value is missing.

```python
df.dropna(how="all")
```

## 5. Understanding `subset`

Use `subset` when you want to check missing values in specific columns.

```python
df.dropna(subset=["Age"])
```

Checks only the Age column.

```python
df.dropna(subset=["Age", "City"])
```

Checks Age and City.

Important: Other columns are ignored when deciding which rows to remove.

## 6. Understanding `thresh`

`thresh` specifies the minimum number of non-missing values required to keep a row.

```python
df.dropna(thresh=3)
```

Meaning: Keep rows containing at least 3 non-missing values.

Examples:
- `thresh=2`: At least 2 valid values.
- `thresh=3`: At least 3 valid values.
- `thresh=4`: At least 4 valid values.

Remember: `thresh` counts non-missing values, not missing values.

## 7. Combining Methods

You can chain multiple operations:

```python
clean_df = (
    df.dropna(how="all")
      .dropna(subset=["Purchase"])
      .dropna(thresh=3)
)
```

Each operation returns a DataFrame that becomes the input to the next operation.

## 8. Important Data Science Principles

1. Never delete missing values blindly.
2. Investigate why values are missing.
3. Consider the business problem before choosing a cleaning method.
4. Deleting rows can cause information loss.
5. Deleting rows can introduce bias.
6. Some incomplete records are still useful.
7. Preserve the original dataset whenever possible.

## Quick Revision Table

| Method | Purpose |
|---|---|
| `isna()` | Detect missing values |
| `isnull()` | Same as `isna()` |
| `sum()` | Count missing values |
| `dropna()` | Remove rows with missing values |
| `axis=1` | Remove columns with missing values |
| `how="any"` | Remove rows with any missing value |
| `how="all"` | Remove rows with all values missing |
| `subset` | Check selected columns |
| `thresh` | Require minimum non-missing values |

**Core lesson:** Missing-value handling is not just about deleting empty cells. It is about protecting useful information while preparing reliable data for analysis and machine learning.
