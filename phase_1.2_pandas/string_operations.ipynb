### **Meet the .str accessor**
`.str allows us to perform string operations on an entire Series.`

```python
import pandas as pd

names = pd.Series([
    "RAZA",
    "sara",
    "ALI",
    "rahul"
])

names.str.lower()
```

- Without .str, we would normally work with individual Python strings.

- With .str, we can apply the operation to every value in a Series.

- Think of .str as a tool that tells Pandas: "Perform this text operation on every item in this column."

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Customer": ["  ali khan", "SARA  ahmed ", "john.smith", "MINA-raza", np.nan],
    "City": ["  new york ", "LOS ANGELES", "chicago  ", None, " sAn FrAnCiSco"],
    "Phone": ["(021) 555-1234", "021.555.9876", "021 555 4567", "0215550000", "021-555-1111"],
    "Sales": ["$1,200", "$850", "$2,300.50", "$99", "$1,000"],
})

display(df)
```

- Clean_City: stripped, Title Case.
- Clean_Customer: stripped, Title Case, with . and - replaced by a space.
- Phone_Digits: digits only (10 digits).
- Sales_Num: the number as a float (remove $ and ,).
- Is_NY: True where the city contains “new york” (case-insensitive, NaN counts as False).
- Display the final DataFrame and df.dtypes.

```python
df["Clean_City"]  = df["City"].str.strip().str.title()

df["Clean_Customer"] = (
    df["Customer"]
    .str.replace(r"[.-]", " ", regex=True)
    .str.replace(r"\s+", " ", regex=True)
    .str.strip()
    .str.title()
)


df["Phone_Digits"] = df["Phone"].str.replace(r"\D", "", regex=True)
df["Sales_Num"] = pd.to_numeric(df["Sales"].str.replace(r"[$,]", "", regex=True), errors="coerce")
df["Is_NY"] = df["City"].str.contains("new york", case=False, na=False)

display(df.dtypes.to_frame(name="Data Type"))

print("Final DataFrame: ")
display(df)
```

### **2. extract() — Pull out a specific pattern**
| Product_Code |
|---|
| ELE101 |
| FUR205 |
| ELE309 |

You want only the numbers.

We can use extract() with a regular expression (regex).

A regex is a pattern that describes what text you want to find.


```python
df["Product_Name("].str.extract(r"(\d+)")
```

| Part | Meaning |
|---|---|
| `r` | Creates a raw string, useful for regex patterns |
| `\d` | Matches a digit from 0 to 9 |
| `+` | Matches one or more digits |
| `()` | Captures the matched part for extraction |

### **Extract email usernames**

```python
df["Email"].str.extract(r"([^@]+)")
```

Here:
- @ identifies the separator.
- [^@] means any character except @.
- (+) means one or more characters.
- () captures the matching text.

| Method | Main purpose | Example |
|---|---|---|
| `split()` | Break text using a separator | `"Raza Khan"` → `"Raza"`, `"Khan"` |
| `extract()` | Pull out text matching a pattern | `"ELE101"` → `"101"` |

Practical rule:
- Know the separator? Use split().
- Need to find a specific pattern? Use extract().

# Pandas String Operations — split() and extract()

## 1. `.str.split()`
- Breaks text into parts using a separator.
- Returns a Series of lists by default.

Example:
df["Name"].str.split(" ")

## 2. `expand=True`
- Places split parts into separate DataFrame columns.

Example:
df["Name"].str.split(" ", expand=True)

## 3. `n=1`
- Limits the number of splits to one.

Example:
df["Name"].str.split(" ", n=1, expand=True)

## 4. `.str.extract()`
- Retrieves text matching a regular expression.
- Requires a capture group `()`.

Example:
df["Code"].str.extract(r"(\d+)")

## 5. Basic Regex Symbols

| Symbol | Meaning |
|---|---|
| `\d` | A digit from 0 to 9 |
| `+` | One or more |
| `()` | Capture the matched pattern |
| `[^@]` | Any character except `@` |

## 6. Practical Examples

df["First_Name"] = parts[0]
df["Last_Name"] = parts[1]

# Extract product digits
df["Product_Digits"] = df["Product_Code"].str.extract(r"(\d+)")

# Extract email username
df["Email_Username"] = df["Email"].str.extract(r"([^@]+)")
