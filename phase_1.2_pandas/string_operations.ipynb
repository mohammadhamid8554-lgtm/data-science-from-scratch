### **Why do we need string operations?**
A Series is not a Python string, so `df["City"].strip()` fails. Use `df["City"].str.strip()` to apply `strip()` to every value. It skips `NaN` values instead of crashing and is vectorized, so no loops are needed.

| Method | Use |
|---|---|
| `.str.strip()` | Remove leading and trailing spaces |
| `.str.lower()` / `.str.upper()` / `.str.title()` | Standardize letter casing |
| `.str.replace(pat, repl, regex=...)` | Replace characters or patterns |
| `.str.contains(pat, na=False)` | Filter rows containing text |
| `.str.split(sep, expand=True)` | Split text into columns |
| `.str.extract(r"(\d+)")` | Extract a matching pattern |
| `.str.len()` | Check text length |


### **Meet the .str accessor**
`.str allows us to perform string operations on an entire Series.`

```python
import pandas as pd

names = pd.Series([
    "RAZA",
    "sara",
    "ALI",
    "rahul"
])

names.str.lower()
```

- Without .str, we would normally work with individual Python strings.

- With .str, we can apply the operation to every value in a Series.

- Think of .str as a tool that tells Pandas: "Perform this text operation on every item in this column."

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Customer": ["  ali khan", "SARA  ahmed ", "john.smith", "MINA-raza", np.nan],
    "City": ["  new york ", "LOS ANGELES", "chicago  ", None, " sAn FrAnCiSco"],
    "Phone": ["(021) 555-1234", "021.555.9876", "021 555 4567", "0215550000", "021-555-1111"],
    "Sales": ["$1,200", "$850", "$2,300.50", "$99", "$1,000"],
})

display(df)
```

- Clean_City: stripped, Title Case.
- Clean_Customer: stripped, Title Case, with . and - replaced by a space.
- Phone_Digits: digits only (10 digits).
- Sales_Num: the number as a float (remove $ and ,).
- Is_NY: True where the city contains “new york” (case-insensitive, NaN counts as False).
- Display the final DataFrame and df.dtypes.

```python
df["Clean_City"]  = df["City"].str.strip().str.title()

df["Clean_Customer"] = (
    df["Customer"]
    .str.replace(r"[.-]", " ", regex=True)
    .str.replace(r"\s+", " ", regex=True)
    .str.strip()
    .str.title()
)


df["Phone_Digits"] = df["Phone"].str.replace(r"\D", "", regex=True)
df["Sales_Num"] = pd.to_numeric(df["Sales"].str.replace(r"[$,]", "", regex=True), errors="coerce")
df["Is_NY"] = df["City"].str.contains("new york", case=False, na=False)

display(df.dtypes.to_frame(name="Data Type"))

print("Final DataFrame: ")
display(df)
```

## Pandas String Operations — Case Conversion

### 1. The `.str` Accessor
- `.str` lets us apply string methods to an entire Pandas Series.
- It is used for text columns in a DataFrame.

### 2. Case Conversion Methods

| Method | Purpose | Example |
|---|---|---|
| `.str.lower()` | Converts text to lowercase | `"RAZA"` → `"raza"` |
| `.str.upper()` | Converts text to uppercase | `"raza"` → `"RAZA"` |
| `.str.title()` | Capitalizes each word | `"raza khan"` → `"Raza Khan"` |

### 3. Important Rules
- These methods return a new Series.
- They do not automatically modify the original column.
- Assign the result to save it.
- `.str.title()` does not remove extra spaces.
- Use `.str.strip()` to remove leading and trailing spaces.

### 4. Practical Example

df["Clean_Name"] = df["Customer_Name"].str.title()

### 5. Data Science Application
- Standardize text formatting.
- Improve consistency in customer and product data.
- Help identify possible duplicates.
- Prepare text columns for analysis.

Remember:
`.str` = apply string operations to a Pandas Series.

### **2. The problem**

```python
"Raza" == " Raza "
```

Python treats these as different strings.

### **A. strip()**

```python
df["Customer_Name"].str.strip()
# Removes spaces from the beginning and end.
```

### **B. lstrip()**

```python
df["Customer_Name"].str.lstrip()
# Removes spaces only from the left side.
```

### **B. rstrip()**

```python
df["Customer_Name"].str.rstrip()
# Removes spaces only from the right side.
```

#### **Important: These methods do not remove spaces between words.**

```python
"Raza   Khan".strip()
# The internal spaces remain.
```
