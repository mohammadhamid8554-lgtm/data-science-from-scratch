### **What is DataFrame?**
A DataFrame is a table with rows and columns. It is one of the most important Pandas structures because real-world Data Science datasets usually look like tables.

### **Why DataFrames matter in Data Science**

With a DataFrame, you can:
- Find customers older than 25.
- Calculate average spending.
- Find the highest-spending customer.
- Detect missing values.
- Remove duplicate records.
- Group customers by city.
- Prepare data for Machine Learning.

This is why DataFrames are central to practical Data Science.

### **DataFrame = 2D labeled table**
- Rows → observations / records
- Columns → variables / features
- Index → row labels
- Column names → variable names
- Shape → (rows, columns)
- One column → usually a Series
- Multiple columns → DataFrame
- Numerical → quantities or measurements
- Categorical → labels, groups, identifiers
