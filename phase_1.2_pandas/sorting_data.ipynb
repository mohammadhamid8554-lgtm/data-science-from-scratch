### **Lesson 1 — Understanding Sorting**
Sorting means arranging data in a particular order.

For example:
- Products from highest revenue to lowest revenue.
- Students from lowest marks to highest marks.
- Customers from oldest to newest.
- Sales records alphabetically by product name.

### **Connection with previously learned topics**



````python
import numpy as np
import pandas as pd

# Sorting data in Python 
num = [3, 1, 2, 5]
num.sort()
print(num)

# Sorting data in NumPy
arr = np.array([2, 0, 1, 4, 3])
print(np.sort(arr))

# Sorting data in Pandas
series = pd.Series(arr)
print(series.sort_values())
````

**`Important difference:`** Python's sorted() sorts Python collections, while Pandas sort_values() sorts DataFrame rows based on selected columns.

Later, in EDA, sorting will help you identify top-selling products, highest-revenue customers, and unusual values.

### **Lesson 6 — Understanding inplace=True**

### **1. Default behavior**

````python
sorted_df = df.sort_values(by="Revenue")
````

Here:
- df remains unchanged.
- sorted_df stores the sorted result.

This is called returning a new result.

### **2. Using inplace=True**

````python
df.sort_values(by="Revenue", inplace=True)
````

Here:
- Pandas modifies df directly.
- The original DataFrame is rearranged.
- No new sorted DataFrame is returned.

### **Connection with Python**

- variables, Objects, Lists, Methods, and Fucntions returns some values.

- new_list = sorted(my_list) returns rearrged new list

- my_list.sort() modifies the original lsit.



### **4. Professional recommendation**

````python
# use:

sorted_df = df.sort_values(by="Revenue")

# Instead of:

df.sort_values(by="Revenue", inplace=True)
````

### **Why?**
- Easier to track transformations.
- Preserves the original dataset.
- Reduces accidental data changes.
- Makes debugging and reproducibility easier.

Do not use inplace=True just because it looks shorter.

# Pandas Topic 7 — Sorting Data

## 1. sort_values()

Sort rows based on column values.

```python
df.sort_values(by="Revenue")
```

- Default: `ascending=True`
- Smallest to largest.

```python
df.sort_values(by="Revenue", ascending=False)
```

- Largest to smallest.

## 2. Multi-column Sorting

Sort using multiple columns.

```python
df.sort_values(
    by=["Quantity", "Revenue"],
    ascending=[False, True]
)
```

- First column = primary sorting key.
- Second column = tie-breaker.
- Sorting uses the next column when previous values are equal.

## 3. sort_index()

Sort rows based on index labels.

```python
df.sort_index()
```

- `sort_values()` → sorts column values.
- `sort_index()` → sorts index labels.

## 4. Text Sorting

```python
df.sort_values(by="Product")
```

Alphabetical order.

```python
df.sort_values(by="Product", ascending=False)
```

Reverse alphabetical order.

## 5. ignore_index=True

Generate a fresh sequential index after sorting.

```python
df.sort_values(
    by="Revenue",
    ascending=False,
    ignore_index=True
)
```

Result index: 0, 1, 2, 3...

The original DataFrame is not automatically changed.

## 6. inplace=True

Modify the original DataFrame directly.

```python
df.sort_values(
    by="Revenue",
    ascending=False,
    inplace=True
)
```

- Changes original DataFrame.
- Returns `None`.
- Avoid assigning its result to another variable.

Preferred approach:

```python
sorted_df = df.sort_values(
    by="Revenue",
    ascending=False
)
```

Preserves the original DataFrame.

## 7. Filtering + Sorting

Filter first, then sort.

```python
df.loc[df["Revenue"] > 50000].sort_values(
    by="Revenue",
    ascending=False
)
```

Filter multiple conditions:

```python
df.loc[
    (df["Quantity"] > 5) &
    (df["Revenue"] > 50000)
].sort_values(
    by="Revenue",
    ascending=False
)
```

- `&` = AND
- `|` = OR
- Put each condition inside parentheses.

## 8. nlargest()

Find the top N records.

```python
df.nlargest(3, "Revenue")
```

Top 3 Revenue values.

## 9. nsmallest()

Find the bottom N records.

```python
df.nsmallest(2, "Revenue")
```

Bottom 2 Revenue values.

## 10. Top N + Selected Columns

```python
df.nlargest(3, "Revenue")[["Product", "Revenue"]]
```

Alternative:

```python
df.sort_values(
    by="Revenue",
    ascending=False
)[["Product", "Revenue"]].head(3)
```

## 11. Reset Index

```python
df.nsmallest(2, "Revenue").reset_index(drop=True)
```

- `reset_index()` → resets index.
- `drop=True` → discards old index instead of adding it as a column.

## Key Takeaway

- `sort_values()` → Sort by column.
- `sort_index()` → Sort by index.
- `ascending=False` → Largest to smallest.
- `nlargest()` → Top N.
- `nsmallest()` → Bottom N.
- `head()` → First N rows.
- `ignore_index=True` → Fresh sequential index.
- `inplace=True` → Modify original.
